```python
import cv2
import os
import numpy as np

# Define the video file path
video_path = "/content/drive/MyDrive/Satellite Project with Debojyoti Banerjee/desert-road-2.mp4"

# Define the output directory for extracted frames
output_dir = "extracted_frames"
os.makedirs(output_dir, exist_ok=True)

print(f"Extracting frames from: {video_path}")
print(f"Saving frames to: {output_dir}")

# Open the video file
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print(f"Error: Could not open video file {video_path}")
    print("This might be a codec issue or file corruption, even if the path is correct.")
else:
    frame_count = 0
    while True:
        # Read a frame from the video
        ret, frame = cap.read()

        # If 'ret' is False, it means we have reached the end of the video
        if not ret:
            break

        # Define the filename for the current frame
        frame_filename = os.path.join(output_dir, f"frame_{frame_count:05d}.png")

        # Save the frame as an image file
        cv2.imwrite(frame_filename, frame)

        frame_count += 1
        if frame_count % 100 == 0:
            print(f"Processed {frame_count} frames...")

    # Release the video capture object
    cap.release()
    print(f"Successfully extracted {frame_count} frames to {output_dir}.")
```

### Extracting Frames from Video

This code block will use `opencv-python` to read the video file frame by frame and save each frame as a PNG image in a designated output directory. This process allows us to treat each frame as a static image, which can then be used for denoising with the methods we've explored previously.

**Note:** This can generate a large number of image files, consuming disk space.

```python
import cv2
import os
import numpy as np

# Define the video file path
video_path = "/content/drive/MyDrive/Satellite Project with Debojyoti Banerjee/desert-road-2.mp4"

# Define the output directory for extracted frames
output_dir = "extracted_frames"
os.makedirs(output_dir, exist_ok=True)

print(f"Extracting frames from: {video_path}")
print(f"Saving frames to: {output_dir}")

# Check if the video file exists before attempting to open
if not os.path.exists(video_path):
    print(f"Error: Video file not found at {video_path}")
    print("Please ensure the path is correct and Google Drive is mounted.")

    # Optionally, list contents of the directory to help user find the correct path
    drive_folder = os.path.dirname(video_path)
    if os.path.isdir(drive_folder):
        print(f"Contents of '{drive_folder}':")
        for item in os.listdir(drive_folder):
            print(f"- {item}")
    else:
        print(f"Could not access directory '{drive_folder}'.")

else:
    # Open the video file
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"Error: Could not open video file {video_path} even though it exists.")
        print("This might be a codec issue or file corruption.")
    else:
        frame_count = 0
        while True:
            # Read a frame from the video
            ret, frame = cap.read()

            # If 'ret' is False, it means we have reached the end of the video
            if not ret:
                break

            # Define the filename for the current frame
            frame_filename = os.path.join(output_dir, f"frame_{frame_count:05d}.png")

            # Save the frame as an image file
            cv2.imwrite(frame_filename, frame)

            frame_count += 1
            if frame_count % 100 == 0:
                print(f"Processed {frame_count} frames...")

        # Release the video capture object
        cap.release()
        print(f"Successfully extracted {frame_count} frames to {output_dir}.")
```

```python
# PASTE THIS IN A NEW CELL - IT WILL WORK!

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy.ndimage import uniform_filter

# Load any image (replace with your path)
image_path = "/content/drive/MyDrive/Satellite Project with Debojyoti Banerjee/desert-road-2.mp4"  # CHANGE THIS

# Try to load
try:
    image = np.array(Image.open(image_path)).astype(np.float64)
    if len(image.shape) == 2:
        image = np.stack([image]*3, axis=-1)
    if image.shape[2] == 4:
        image = image[:,:,:3]
    print(f"✓ Loaded: {image.shape}")
except:
    # If fails, create test image
    image = np.random.rand(256, 256, 3) * 255
    print("Using test image")

# Add noise
noisy = image + np.random.normal(0, 25, image.shape)
noisy = np.clip(noisy, 0, 255)

# Denoise
filtered = np.zeros_like(noisy)
for i in range(3):
    filtered[:,:,i] = uniform_filter(noisy[:,:,i], size=5)

# Calculate PSNR
mse = np.mean((image - filtered) ** 2)
psnr_value = 20 * np.log10(255.0 / np.sqrt(mse))

# Show
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(image.astype(np.uint8))
axes[0].set_title('Original')
axes[1].imshow(noisy.astype(np.uint8))
axes[1].set_title('Noisy')
axes[2].imshow(filtered.astype(np.uint8))
axes[2].set_title(f'Filtered\nPSNR: {psnr_value:.2f} dB')
plt.show()

print(f"✓ PSNR: {psnr_value:.2f} dB")
```

```python
"""
GeoTIFF Satellite Image Denoising - FIXED FOR GEOTIFF FILES
Handles multi-band satellite imagery with proper libraries

Citations:
[1] GDAL/Rasterio: Gillies, S., et al. (2013-2024) "Rasterio: geospatial raster I/O for Python"
    https://github.com/rasterio/rasterio
[2] GeoTIFF format: Ritter, N., Ruth, M. (1997) "GeoTIFF Format Specification"
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import uniform_filter
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import warnings
warnings.filterwarnings('ignore')


def install_dependencies():
    """Install required libraries for GeoTIFF processing."""
    import subprocess
    import sys

    print("Installing required libraries...")
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install",
                              "rasterio", "-q"])
        print("Successfully installed rasterio")
    except:
        print("Could not install rasterio. Trying alternative method...")


# =====================================================================
# STEP 1: LOAD GEOTIFF IMAGE (MULTIPLE METHODS)
# =====================================================================

def load_geotiff_rasterio(image_path):
    """
    Load GeoTIFF using rasterio (RECOMMENDED for satellite data).

    Citation: Rasterio is the standard library for geospatial raster data
    https://rasterio.readthedocs.io/
    """
    try:
        import rasterio

        print(f"Loading with rasterio...")
        with rasterio.open(image_path) as src:
            # Read all bands
            image = src.read()  # Shape: (bands, height, width)

            # Get metadata
            n_bands = src.count
            height = src.height
            width = src.width
            dtype = src.dtypes[0]

            print(f"Loaded GeoTIFF:")
            print(f"    - Bands: {n_bands}")
            print(f"    - Size: {width} x {height}")
            print(f"    - Data type: {dtype}")
            print(f"    - CRS: {src.crs}")

            # Transpose to (height, width, bands) for easier handling
            image = np.moveaxis(image, 0, -1)

            return image, True

    except ImportError:
        print("rasterio not available.")
        return None, False
    except Exception as e:
        print(f"rasterio error: {type(e).__name__}: {e}") # More detailed error
        return None, False


def load_geotiff_gdal(image_path):
    """
    Load GeoTIFF using GDAL (fallback method).

    Citation: GDAL is the fundamental geospatial library
    https://gdal.org/
    """
    try:
        from osgeo import gdal

        print(f"  Loading with GDAL...")
        dataset = gdal.Open(image_path, gdal.GA_ReadOnly)

        if dataset is None:
            print("GDAL failed to open dataset (returned None).") # More specific message
            return None, False

        # Get dimensions
        n_bands = dataset.RasterCount
        width = dataset.RasterXSize
        height = dataset.RasterYSize

        print(f"Loaded GeoTIFF:")
        print(f"    - Bands: {n_bands}")
        print(f"    - Size: {width} x {height}")

        # Read all bands
        bands = []
        for i in range(1, n_bands + 1):
            band = dataset.GetRasterBand(i)
            bands.append(band.ReadAsArray())

        # Stack into (height, width, bands)
        image = np.stack(bands, axis=-1)

        dataset = None  # Close dataset
        return image, True

    except ImportError:
        print("GDAL not available.")
        return None, False
    except Exception as e:
        print(f"GDAL error: {type(e).__name__}: {e}") # More detailed error
        return None, False


def load_geotiff_tifffile(image_path):
    """
    Load TIFF using tifffile library (works for some TIFFs).

    Citation: tifffile - Read and write TIFF files
    https://github.com/cgohlke/tifffile
    """
    try:
        import tifffile

        print(f"  Loading with tifffile...")
        image = tifffile.imread(image_path)

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        # Handle different organizations
        if len(image.shape) == 3:
            # Could be (bands, height, width) or (height, width, bands)
            if image.shape[0] < image.shape[2]:
                # Likely (bands, height, width) - transpose
                image = np.moveaxis(image, 0, -1)

        return image, True

    except Exception as e: # Add except block to catch potential errors
        print(f"tifffile error: {type(e).__name__}: {e}") # More detailed error
        return None, False


def load_geotiff_opencv(image_path):
    """
    Load TIFF using OpenCV (limited multi-band support).
    """
    try:
        import cv2

        print(f"  Loading with OpenCV...")
        image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)

        if image is None:
            print("OpenCV failed to load image (returned None).") # More specific message
            return None, False

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        # Convert BGR to RGB if needed
        if len(image.shape) == 3 and image.shape[2] == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        return image, True

    except Exception as e: # Add except block
        print(f"OpenCV error: {type(e).__name__}: {e}") # More detailed error
        return None, False

def load_satellite_image(image_path):
    """
    Try multiple methods to load GeoTIFF satellite image.

    Priority order:
    1. rasterio (best for GeoTIFF)
    2. GDAL (robust fallback)
    3. tifffile (general TIFF)
    4. OpenCV (limited support)
    """
    import os # Import os to check file existence

    print(f"\n[1] Loading satellite image from:")
    print(f"    {image_path}")

    # Check if file exists before attempting to load
    if not os.path.exists(image_path):
        print(f"Error: File not found at {image_path}")
        return None # Return None if file does not exist


    # Method 1: Rasterio (RECOMMENDED)
    image, success = load_geotiff_rasterio(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 2: GDAL
    image, success = load_geotiff_gdal(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 3: tifffile
    image, success = load_geotiff_tifffile(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 4: OpenCV
    image, success = load_geotiff_opencv(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # All methods failed
    # If file existed but couldn't be loaded, raise a specific error
    if os.path.exists(image_path):
        raise RuntimeError(
            f"Could not load image from {image_path} with any available library.\n"
            "Please check the file format and consider installing rasterio: pip install rasterio"
        )
    else:
        # This case should be handled by the os.path.exists check above, but as a fallback
         raise FileNotFoundError(f"Image file not found at: {image_path}")


# =====================================================================
# HELPER FUNCTIONS
# =====================================================================

def normalize_to_uint8(image_array):
    """Normalize to 0-255 for display using percentile clipping."""
    if image_array.size == 0: # Handle empty array case
        return np.empty_like(image_array, dtype=np.uint8)

    p2, p98 = np.percentile(image_array, (2, 98))
    # Handle case where p2 and p98 are the same (flat image)
    if p2 == p98:
        return np.full_like(image_array, 127, dtype=np.uint8) # Return a mid-gray image

    image_clipped = np.clip(image_array, p2, p98)
    image_normalized = ((image_clipped - p2) / (p98 - p2) * 255)
    return np.clip(image_normalized, 0, 255).astype(np.uint8)


def select_rgb_bands(image_array, rgb_indices=None):
    """Extract RGB bands from multi-band image."""
    print(f"\n[2] Extracting RGB bands...")
    print(f"    Image shape: {image_array.shape}")

    if len(image_array.shape) == 2:
        print("    Grayscale detected - converting to RGB")
        return np.stack([image_array] * 3, axis=-1)

    elif len(image_array.shape) == 3 and image_array.shape[2] == 3:
        print("    RGB image detected")
        return image_array

    elif len(image_array.shape) == 3 and image_array.shape[2] > 3:
        # Multi-band - auto-detect or use specified indices
        if rgb_indices is None:
            # Try common satellite RGB combinations
            n_bands = image_array.shape[2]

            print(f"    Multi-band image: {n_bands} bands detected")
            print("    Trying common RGB band combinations...")

            # For Sentinel-2: RGB = bands 4,3,2 (0-indexed: 3,2,1)
            # For Landsat: RGB = bands 4,3,2 (0-indexed: 3,2,1)
            if n_bands >= 4:
                print("    Using bands [3,2,1] for RGB (Sentinel-2/Landsat style)")
                rgb_indices = (3, 2, 1)
            else:
                print("    Using first 3 bands [0,1,2]")
                rgb_indices = (0, 1, 2)

        # Ensure indices are within valid range
        max_band_index = image_array.shape[2] - 1
        valid_indices = [idx for idx in rgb_indices if 0 <= idx <= max_band_index]

        if len(valid_indices) != 3:
            print(f"Warning: Could not find 3 valid RGB bands from indices {rgb_indices}. Using first 3 available bands.")
            valid_indices = list(range(min(3, image_array.shape[2])))
            if len(valid_indices) < 3:
                 raise ValueError(f"Image only has {image_array.shape[2]} bands, cannot select 3 for RGB.")

        print(f"    Selected RGB indices: {valid_indices}")
        return image_array[:, :, valid_indices]


    else:
        raise ValueError(f"Unexpected image shape: {image_array.shape}")


def add_gaussian_noise(image, sigma=25.0):
    """Add Gaussian noise."""
    # Ensure image is float for noise addition
    image_float = image.astype(np.float64)
    noise = np.random.normal(0, sigma, image_float.shape)
    noisy_image = image_float + noise
    return noisy_image


def apply_mean_filter_multichannel(image, kernel_size=5):
    """Apply mean filter to each channel separately."""
    # Ensure image is float for filtering
    image_float = image.astype(np.float64)

    if len(image_float.shape) == 2:
        return uniform_filter(image_float, size=kernel_size)

    filtered_channels = []
    for i in range(image_float.shape[2]):
        filtered_channels.append(
            uniform_filter(image_float[:, :, i], size=kernel_size)
        )
    return np.stack(filtered_channels, axis=-1)


def calculate_metrics(original, processed, data_range=None):
    """Calculate PSNR and SSIM."""
    # Determine data range if not provided
    if data_range is None:
        # If original and processed have different min/max, use a common range like 0-255
        # or the full possible range based on dtype
        min_val = min(original.min(), processed.min())
        max_val = max(original.max(), processed.max())
        # Adjust data_range calculation based on actual value range
        # A common approach is to use the range of the original image, or 255 for 8-bit
        if original.dtype == np.uint8:
            data_range = 255.0
        elif original.dtype == np.float64 or original.dtype == np.float32:
             # Assume float images are normalized, e.g., to 0-1 or 0-max_sensor_value
             # If normalized to 0-1, data_range should be 1.0
             # If not normalized, use the actual range or a typical sensor bit depth range
             # For this case, let's assume the data is scaled to something where the max value is meaningful
             data_range = np.max(original) # Using max value of original as data_range

        # Fallback if data_range is still None (e.g., empty arrays)
        if data_range is None or data_range <= 0:
             data_range = 1.0 # Default to 1.0 for normalized float data


    # Ensure original and processed are float for metrics calculation
    original_float = original.astype(np.float64)
    processed_float = processed.astype(np.float64)

    # Clip processed to the range of original for metric calculation if they differ significantly
    # This prevents large deviations from skewing PSNR for images with different value ranges after processing
    processed_clipped = np.clip(processed_float, original_float.min(), original_float.max())


    try:
        # Ensure data_range is a positive scalar
        if not isinstance(data_range, (int, float)) or data_range <= 0:
            print(f"Warning: Invalid data_range ({data_range}) for PSNR calculation. Using max value of original.")
            data_range = np.max(original_float)
            if data_range <= 0: # Fallback if max is also non-positive
                 data_range = 1.0


        psnr_value = psnr(original_float, processed_clipped, data_range=data_range)
    except Exception as e:
        print(f"Error calculating PSNR: {e}")
        psnr_value = np.nan # Return Not a Number on error


    try:
        # Ensure win_size is appropriate for image dimensions
        min_dim = min(original_float.shape[0], original_float.shape[1])
        win_size = min(7, min_dim)
        if win_size < 1: # Ensure win_size is at least 1 for very small images
            win_size = 1


        if len(original_float.shape) == 3:
            # SSIM requires channel_axis for color images
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range,
                             channel_axis=2, win_size=win_size)
        else:
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range, win_size=win_size)
    except Exception as e:
        print(f"Error calculating SSIM: {e}")
        ssim_value = np.nan # Return Not a Number on error


    return psnr_value, ssim_value


# =====================================================================
# MAIN PROCESSING
# =====================================================================

def main():
    """Complete denoising pipeline."""

    print("="*70)
    print("SATELLITE IMAGE DENOISING - GEOTIFF COMPATIBLE")
    print("="*70)

    # YOUR IMAGE PATH - Updated to use a local file path from cloned repo
    image_path = "/content/all_band_vskp.tif" # Using the local file found


    try:
        # Load image
        image_array_raw = load_satellite_image(image_path)

        if image_array_raw is None:
             # load_satellite_image already prints an error if file not found
             return None, None, None, None, None, None, None, None # Return None if loading failed

        # Extract RGB bands
        # Option: manually specify RGB bands if you know them
        # rgb_indices = (3, 2, 1)  # For Sentinel-2/Landsat
        # rgb_indices = (2, 1, 0) # For typical RGB (band 3=Red, 2=Green, 1=Blue) if available
        rgb_indices = None  # Auto-detect

        # Handle potential grayscale image loaded by load_satellite_image (e.g. from .bmp)
        if len(image_array_raw.shape) == 2:
             # If grayscale, convert to RGB for consistent processing
             image_array = np.stack([image_array_raw]*3, axis=-1)
             print("Grayscale image loaded, converted to RGB.")
        else:
             # Otherwise, select RGB bands from multi-band or use existing RGB
             image_array = select_rgb_bands(image_array_raw, rgb_indices)

        # Ensure image_array is float64 for calculations
        image_array = image_array.astype(np.float64)


        # Normalize data range before adding noise and calculating metrics
        # This is important for consistent noise levels and metric comparisons
        print(f"\n[3] Normalizing data range...")
        print(f"    Original range: [{image_array.min():.2f}, {image_array.max():.2f}]")

        # Normalize the original image data to a common range, e.g., 0-255 or 0-1
        # Normalizing to the original max value might be suitable for some sensor data
        # Let's normalize to 0-max_original_value for this example if max > 255
        max_original_value = np.max(image_array)
        if max_original_value > 0: # Avoid division by zero
             image_array_normalized = image_array / max_original_value * 255.0 # Scale to 0-255 range
        else:
             # Handle case where image is all zeros or negative
             image_array_normalized = image_array # No scaling if max is non-positive


        print(f"    Normalized range: [{image_array_normalized.min():.2f}, {image_array_normalized.max():.2f}]")


        # Add noise to the normalized image
        print(f"\n[4] Adding Gaussian noise...")
        noise_sigma = 40.0
        print(f"    Noise level: σ = {noise_sigma}")

        noisy_image = add_gaussian_noise(image_array_normalized, sigma=noise_sigma)

        # Calculate metrics (PSNR/SSIM) comparing normalized original to noisy image
        # Use the max value of the normalized original image as data_range for metrics
        psnr_noisy, ssim_noisy = calculate_metrics(image_array_normalized, noisy_image, data_range=255.0) # Assuming 0-255 range after normalization

        print(f"    Noisy - PSNR: {psnr_noisy:.2f} dB, SSIM: {ssim_noisy:.4f}")

        # Apply mean filter to the noisy image
        print(f"\n[5] Applying mean filter...")
        kernel_size = 5
        print(f"    Kernel size: {kernel_size}x{kernel_size}")

        filtered_image = apply_mean_filter_multichannel(noisy_image, kernel_size)

        # Calculate metrics comparing normalized original to filtered image
        psnr_filtered, ssim_filtered = calculate_metrics(image_array_normalized, filtered_image, data_range=255.0) # Assuming 0-255 range

        print(f"    Filtered - PSNR: {psnr_filtered:.2f} dB, SSIM: {ssim_filtered:.4f}")
        print(f"    Improvement: +{psnr_filtered - psnr_noisy:.2f} dB")

        # Visualize
        print(f"\n[6] Creating visualization...")

        fig, axes = plt.subplots(1, 3, figsize=(15, 5))

        axes[0].imshow(normalize_to_uint8(image_array_normalized))
        axes[0].set_title('Original (Clean)', fontsize=12, fontweight='bold')
        axes[0].axis('off')

        axes[1].imshow(normalize_to_uint8(noisy_image))
        axes[1].set_title(f'Noisy (σ={noise_sigma})\nPSNR: {psnr_noisy:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[1].axis('off')

        axes[2].imshow(normalize_to_uint8(filtered_image))
        axes[2].set_title(f'Mean Filtered ({kernel_size}x{kernel_size})\n'
                         f'PSNR: {psnr_filtered:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[2].axis('off')

        plt.tight_layout()
        plt.savefig('satellite_denoising_results.png', dpi=150, bbox_inches='tight')
        print(f"    ✓ Saved: satellite_denoising_results.png")
        plt.show()

        # Summary
        print("\n" + "="*70)
        print("RESULTS SUMMARY")
        print("="*70)
        print(f"  Image shape: {image_array_normalized.shape}")
        print(f"  Noise level: σ = {noise_sigma}")
        print(f"\n  {'Method':<20} {'PSNR (dB)':<12} {'SSIM':<10}")
        print(f"  {'-'*45}")
        print(f"  {'Noisy':<20} {psnr_noisy:<12.2f} {ssim_noisy:<10.4f}")
        print(f"  {'Mean Filter':<20} {psnr_filtered:<12.2f} {ssim_filtered:<10.4f}")
        print(f"  {'Improvement':<20} {f'+{psnr_filtered-psnr_noisy:.2f}':<12} {f'+{ssim_filtered-ssim_noisy:.4f}':<10}")
        print("="*70)

        # Return the key images and metrics for potential further use in other cells
        return image_array_normalized, noisy_image, filtered_image, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma


    except FileNotFoundError:
        print(f"\n❌ File not found: {image_path}")
        print("\nTroubleshooting:")
        print("  1. Check if the file path is correct")
        print("  2. Check file permissions")
        return None, None, None, None, None, None, None, None


    except Exception as e:
        print(f"\n❌ Error: {type(e).__name__}")
        print(f"   {str(e)}")
        import traceback
        traceback.print_exc()
        return None, None, None, None, None, None, None, None


# =====================================================================
# RUN
# =====================================================================

if __name__ == "__main__":
    # First, install rasterio if not already installed
    print("Checking dependencies...")
    try:
        import rasterio
        print("✓ rasterio is installed")
    except ImportError:
        print("⚠ rasterio not found. Installing...")
        try:
            import subprocess
            import sys
            subprocess.check_call([sys.executable, "-m", "pip", "install", "rasterio", "-q"])
            print("✓ rasterio installed")
        except Exception as e:
            print(f"Error installing rasterio: {e}")


    # Run main pipeline
    # Store the results in variables that can be used in subsequent cells
    # Assigning the results to global variables
    global original_image_arr, noisy_image_arr, mean_filtered_image_arr, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma
    original_image_arr, noisy_image_arr, mean_filtered_image_arr, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma = main()
```

```python
"""
GeoTIFF Satellite Image Denoising - FIXED FOR GEOTIFF FILES
Handles multi-band satellite imagery with proper libraries

Citations:
[1] GDAL/Rasterio: Gillies, S., et al. (2013-2024) "Rasterio: geospatial raster I/O for Python"
    https://github.com/rasterio/rasterio
[2] GeoTIFF format: Ritter, N., Ruth, M. (1997) "GeoTIFF Format Specification"
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import uniform_filter
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import warnings
warnings.filterwarnings('ignore')


def install_dependencies():
    """Install required libraries for GeoTIFF processing."""
    import subprocess
    import sys

    print("Installing required libraries...")
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install",
                              "rasterio", "-q"])
        print("Successfully installed rasterio")
    except:
        print("Could not install rasterio. Trying alternative method...")


# =====================================================================
# STEP 1: LOAD GEOTIFF IMAGE (MULTIPLE METHODS)
# =====================================================================

def load_geotiff_rasterio(image_path):
    """
    Load GeoTIFF using rasterio (RECOMMENDED for satellite data).

    Citation: Rasterio is the standard library for geospatial raster data
    https://rasterio.readthedocs.io/
    """
    try:
        import rasterio

        print(f"Loading with rasterio...")
        with rasterio.open(image_path) as src:
            # Read all bands
            image = src.read()  # Shape: (bands, height, width)

            # Get metadata
            n_bands = src.count
            height = src.height
            width = src.width
            dtype = src.dtypes[0]

            print(f"Loaded GeoTIFF:")
            print(f"    - Bands: {n_bands}")
            print(f"    - Size: {width} x {height}")
            print(f"    - Data type: {dtype}")
            print(f"    - CRS: {src.crs}")

            # Transpose to (height, width, bands) for easier handling
            image = np.moveaxis(image, 0, -1)

            return image, True

    except ImportError:
        print("rasterio not available")
        return None, False
    except Exception as e:
        print(f"rasterio error: {e}")
        return None, False


def load_geotiff_gdal(image_path):
    """
    Load GeoTIFF using GDAL (fallback method).

    Citation: GDAL is the fundamental geospatial library
    https://gdal.org/
    """
    try:
        from osgeo import gdal

        print(f"  Loading with GDAL...")
        dataset = gdal.Open(image_path, gdal.GA_ReadOnly)

        if dataset is None:
            return None, False

        # Get dimensions
        n_bands = dataset.RasterCount
        width = dataset.RasterXSize
        height = dataset.RasterYSize

        print(f"Loaded GeoTIFF:")
        print(f"    - Bands: {n_bands}")
        print(f"    - Size: {width} x {height}")

        # Read all bands
        bands = []
        for i in range(1, n_bands + 1):
            band = dataset.GetRasterBand(i)
            bands.append(band.ReadAsArray())

        # Stack into (height, width, bands)
        image = np.stack(bands, axis=-1)

        dataset = None  # Close dataset
        return image, True

    except ImportError:
        print("GDAL not available")
        return None, False
    except Exception as e:
        print(f"GDAL error: {e}")
        return None, False


def load_geotiff_tifffile(image_path):
    """
    Load TIFF using tifffile library (works for some TIFFs).

    Citation: tifffile - Read and write TIFF files
    https://github.com/cgohlke/tifffile
    """
    try:
        import tifffile

        print(f"  Loading with tifffile...")
        image = tifffile.imread(image_path)

        print(f"  ✓ Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        # Handle different organizations
        if len(image.shape) == 3:
            # Could be (bands, height, width) or (height, width, bands)
            if image.shape[0] < image.shape[2]:
                # Likely (bands, height, width) - transpose
                image = np.moveaxis(image, 0, -1)

        return image, True

    except Exception as e: # Add except block to catch potential errors
        print(f"tifffile error: {e}")
        return None, False


def load_geotiff_opencv(image_path):
    """
    Load TIFF using OpenCV (limited multi-band support).
    """
    try:
        import cv2

        print(f"  Loading with OpenCV...")
        image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)

        if image is None:
            return None, False

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        # Convert BGR to RGB if needed
        if len(image.shape) == 3 and image.shape[2] == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        return image, True

    except Exception as e: # Add except block
        print(f"OpenCV error: {e}")
        return None, False

def load_satellite_image(image_path):
    """
    Try multiple methods to load GeoTIFF satellite image.

    Priority order:
    1. rasterio (best for GeoTIFF)
    2. GDAL (robust fallback)
    3. tifffile (general TIFF)
    4. OpenCV (limited support)
    """
    print(f"\n[1] Loading satellite image from:")
    print(f"    {image_path}")

    # Method 1: Rasterio (RECOMMENDED)
    image, success = load_geotiff_rasterio(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 2: GDAL
    image, success = load_geotiff_gdal(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 3: tifffile
    image, success = load_geotiff_tifffile(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 4: OpenCV
    image, success = load_geotiff_opencv(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # All methods failed
    raise RuntimeError(
        "❌ Could not load image with any available library.\n"
        "Please install rasterio: pip install rasterio"
    )


# =====================================================================
# HELPER FUNCTIONS
# =====================================================================

def normalize_to_uint8(image_array):
    """Normalize to 0-255 for display using percentile clipping."""
    if image_array.size == 0: # Handle empty array case
        return np.empty_like(image_array, dtype=np.uint8)

    p2, p98 = np.percentile(image_array, (2, 98))
    # Handle case where p2 and p98 are the same (flat image)
    if p2 == p98:
        return np.full_like(image_array, 127, dtype=np.uint8) # Return a mid-gray image

    image_clipped = np.clip(image_array, p2, p98)
    image_normalized = ((image_clipped - p2) / (p98 - p2) * 255)
    return np.clip(image_normalized, 0, 255).astype(np.uint8)


def select_rgb_bands(image_array, rgb_indices=None):
    """Extract RGB bands from multi-band image."""
    print(f"\n[2] Extracting RGB bands...")
    print(f"    Image shape: {image_array.shape}")

    if len(image_array.shape) == 2:
        print("    Grayscale detected - converting to RGB")
        return np.stack([image_array] * 3, axis=-1)

    elif len(image_array.shape) == 3 and image_array.shape[2] == 3:
        print("    RGB image detected")
        return image_array

    elif len(image_array.shape) == 3 and image_array.shape[2] > 3:
        # Multi-band - auto-detect or use specified indices
        if rgb_indices is None:
            # Try common satellite RGB combinations
            n_bands = image_array.shape[2]

            print(f"    Multi-band image: {n_bands} bands detected")
            print("    Trying common RGB band combinations...")

            # For Sentinel-2: RGB = bands 4,3,2 (0-indexed: 3,2,1)
            # For Landsat: RGB = bands 4,3,2 (0-indexed: 3,2,1)
            if n_bands >= 4:
                print("    Using bands [3,2,1] for RGB (Sentinel-2/Landsat style)")
                rgb_indices = (3, 2, 1)
            else:
                print("    Using first 3 bands [0,1,2]")
                rgb_indices = (0, 1, 2)

        # Ensure indices are within valid range
        max_band_index = image_array.shape[2] - 1
        valid_indices = [idx for idx in rgb_indices if 0 <= idx <= max_band_index]

        if len(valid_indices) != 3:
            print(f"Warning: Could not find 3 valid RGB bands from indices {rgb_indices}. Using first 3 available bands.")
            valid_indices = list(range(min(3, image_array.shape[2])))
            if len(valid_indices) < 3:
                 raise ValueError(f"Image only has {image_array.shape[2]} bands, cannot select 3 for RGB.")

        print(f"    Selected RGB indices: {valid_indices}")
        return image_array[:, :, valid_indices]


    else:
        raise ValueError(f"Unexpected image shape: {image_array.shape}")


def add_gaussian_noise(image, sigma=25.0):
    """Add Gaussian noise."""
    # Ensure image is float for noise addition
    image_float = image.astype(np.float64)
    noise = np.random.normal(0, sigma, image_float.shape)
    noisy_image = image_float + noise
    return noisy_image


def apply_mean_filter_multichannel(image, kernel_size=5):
    """Apply mean filter to each channel separately."""
    # Ensure image is float for filtering
    image_float = image.astype(np.float64)

    if len(image_float.shape) == 2:
        return uniform_filter(image_float, size=kernel_size)

    filtered_channels = []
    for i in range(image_float.shape[2]):
        filtered_channels.append(
            uniform_filter(image_float[:, :, i], size=kernel_size)
        )
    return np.stack(filtered_channels, axis=-1)


def calculate_metrics(original, processed, data_range=None):
    """Calculate PSNR and SSIM."""
    # Determine data range if not provided
    if data_range is None:
        # If original and processed have different min/max, use a common range like 0-255
        # or the full possible range based on dtype
        min_val = min(original.min(), processed.min())
        max_val = max(original.max(), processed.max())
        # Adjust data_range calculation based on actual value range
        # A common approach is to use the range of the original image, or 255 for 8-bit
        if original.dtype == np.uint8:
            data_range = 255.0
        elif original.dtype == np.float64 or original.dtype == np.float32:
             # Assume float images are normalized, e.g., to 0-1 or 0-max_sensor_value
             # If normalized to 0-1, data_range should be 1.0
             # If not normalized, use the actual range or a typical sensor bit depth range
             # For this case, let's assume the data is scaled to something where the max value is meaningful
             data_range = np.max(original) # Using max value of original as data_range

        # Fallback if data_range is still None (e.g., empty arrays)
        if data_range is None or data_range <= 0:
             data_range = 1.0 # Default to 1.0 for normalized float data


    # Ensure original and processed are float for metrics calculation
    original_float = original.astype(np.float64)
    processed_float = processed.astype(np.float64)

    # Clip processed to the range of original for metric calculation if they differ significantly
    # This prevents large deviations from skewing PSNR for images with different value ranges after processing
    processed_clipped = np.clip(processed_float, original_float.min(), original_float.max())


    try:
        # Ensure data_range is a positive scalar
        if not isinstance(data_range, (int, float)) or data_range <= 0:
            print(f"Warning: Invalid data_range ({data_range}) for PSNR calculation. Using max value of original.")
            data_range = np.max(original_float)
            if data_range is None or data_range <= 0: # Fallback if max is also non-positive or None
                 data_range = 1.0


        psnr_value = psnr(original_float, processed_clipped, data_range=data_range)
    except Exception as e:
        print(f"Error calculating PSNR: {e}")
        psnr_value = np.nan # Return Not a Number on error


    try:
        # Ensure win_size is appropriate for image dimensions
        min_dim = min(original_float.shape[0], original_float.shape[1])
        win_size = min(7, min_dim)
        if win_size < 1: # Ensure win_size is at least 1 for very small images
            win_size = 1


        if len(original_float.shape) == 3:
            # SSIM requires channel_axis for color images
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range,
                             channel_axis=2, win_size=win_size)
        else:
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range, win_size=win_size)
    except Exception as e:
        print(f"Error calculating SSIM: {e}")
        ssim_value = np.nan # Return Not a Number on error


    return psnr_value, ssim_value


# =====================================================================
# MAIN PROCESSING
# =====================================================================

def main():
    """Complete denoising pipeline."""

    print("="*70)
    print("SATELLITE IMAGE DENOISING - GEOTIFF COMPATIBLE")
    print("="*70)

    # YOUR IMAGE PATH - Updated to use a local file path from cloned repo
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"

    try:
        # Load image
        # Added check for file existence
        import os
        if not os.path.exists(image_path):
             raise FileNotFoundError(f"Image file not found at: {image_path}")


        image_array_raw = load_satellite_image(image_path)

        if image_array_raw is None:
             raise RuntimeError("Failed to load image.")

        # Extract RGB bands
        # Option: manually specify RGB bands if you know them
        # rgb_indices = (3, 2, 1)  # For Sentinel-2/Landsat
        # rgb_indices = (2, 1, 0) # For typical RGB (band 3=Red, 2=Green, 1=Blue) if available
        rgb_indices = None  # Auto-detect

        # Handle potential grayscale image loaded by load_satellite_image (e.g. from .bmp)
        if len(image_array_raw.shape) == 2:
             # If grayscale, convert to RGB for consistent processing
             image_array = np.stack([image_array_raw]*3, axis=-1)
             print("Grayscale image loaded, converted to RGB.")
        else:
             # Otherwise, select RGB bands from multi-band or use existing RGB
             image_array = select_rgb_bands(image_array_raw, rgb_indices)

        # Ensure image_array is float64 for calculations
        image_array = image_array.astype(np.float64)


        # Normalize data range before adding noise and calculating metrics
        # This is important for consistent noise levels and metric comparisons
        print(f"\n[3] Normalizing data range...")
        print(f"    Original range: [{image_array.min():.2f}, {image_array.max():.2f}]")

        # Normalize the original image data to a common range, e.g., 0-255 or 0-1
        # Normalizing to the original max value might be suitable for some sensor data
        # Let's normalize to 0-max_original_value for this example if max > 255
        max_original_value = np.max(image_array)
        if max_original_value > 0: # Avoid division by zero
             image_array_normalized = image_array / max_original_value * 255.0 # Scale to 0-255 range
        else:
             # Handle case where image is all zeros or negative
             image_array_normalized = image_array # No scaling if max is non-positive


        print(f"    Normalized range: [{image_array_normalized.min():.2f}, {image_array_normalized.max():.2f}]")


        # Add noise to the normalized image
        print(f"\n[4] Adding Gaussian noise...")
        noise_sigma = 40.0
        print(f"    Noise level: σ = {noise_sigma}")

        noisy_image = add_gaussian_noise(image_array_normalized, sigma=noise_sigma)

        # Calculate metrics (PSNR/SSIM) comparing normalized original to noisy image
        # Use the max value of the normalized original image as data_range for metrics
        psnr_noisy, ssim_noisy = calculate_metrics(image_array_normalized, noisy_image, data_range=255.0) # Assuming 0-255 range after normalization

        print(f"    Noisy - PSNR: {psnr_noisy:.2f} dB, SSIM: {ssim_noisy:.4f}")

        # Apply mean filter to the noisy image
        print(f"\n[5] Applying mean filter...")
        kernel_size = 5
        print(f"    Kernel size: {kernel_size}x{kernel_size}")

        filtered_image = apply_mean_filter_multichannel(noisy_image, kernel_size)

        # Calculate metrics comparing normalized original to filtered image
        psnr_filtered, ssim_filtered = calculate_metrics(image_array_normalized, filtered_image, data_range=255.0) # Assuming 0-255 range

        print(f"    Filtered - PSNR: {psnr_filtered:.2f} dB, SSIM: {ssim_filtered:.4f}")
        print(f"    Improvement: +{psnr_filtered - psnr_noisy:.2f} dB")

        # Visualize
        print(f"\n[6] Creating visualization...")

        fig, axes = plt.subplots(1, 3, figsize=(15, 5))

        axes[0].imshow(normalize_to_uint8(image_array_normalized))
        axes[0].set_title('Original (Clean)', fontsize=12, fontweight='bold')
        axes[0].axis('off')

        axes[1].imshow(normalize_to_uint8(noisy_image))
        axes[1].set_title(f'Noisy (σ={noise_sigma})\nPSNR: {psnr_noisy:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[1].axis('off')

        axes[2].imshow(normalize_to_uint8(filtered_image))
        axes[2].set_title(f'Mean Filtered ({kernel_size}x{kernel_size})\n'
                         f'PSNR: {psnr_filtered:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[2].axis('off')

        plt.tight_layout()
        plt.savefig('satellite_denoising_results.png', dpi=150, bbox_inches='tight')
        print(f"    ✓ Saved: satellite_denoising_results.png")
        plt.show()

        # Summary
        print("\n" + "="*70)
        print("RESULTS SUMMARY")
        print("="*70)
        print(f"  Image shape: {image_array_normalized.shape}")
        print(f"  Noise level: σ = {noise_sigma}")
        print(f"\n  {'Method':<20} {'PSNR (dB)':<12} {'SSIM':<10}")
        print(f"  {'-'*45}")
        print(f"  {'Noisy':<20} {psnr_noisy:<12.2f} {ssim_noisy:<10.4f}")
        print(f"  {'Mean Filter':<20} {psnr_filtered:<12.2f} {ssim_filtered:<10.4f}")
        print(f"  {'Improvement':<20} {f'+{psnr_filtered-psnr_noisy:.2f}':<12} {f'+{ssim_filtered-ssim_noisy:.4f}':<10}")
        print("="*70)

        # Return the key images and metrics for potential further use in other cells
        return image_array_normalized, noisy_image, filtered_image, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma


    except FileNotFoundError:
        print(f"\n❌ File not found: {image_path}")
        print("\nTroubleshooting:")
        print("  1. Check if Google Drive is mounted:")
        print("     from google.colab import drive")
        print("     drive.mount('/content/drive')")
        print("  2. Verify the file path is correct")
        print("  3. Check file permissions")

    except Exception as e:
        print(f"\n❌ Error: {type(e).__name__}")
        print(f"   {str(e)}")
        import traceback
        traceback.print_exc()
        return None, None, None, None, None, None, None, None


# =====================================================================
# RUN
# =====================================================================

if __name__ == "__main__":
    # First, install rasterio if not already installed
    print("Checking dependencies...")
    try:
        import rasterio
        print("✓ rasterio is installed")
    except ImportError:
        print("⚠ rasterio not found. Installing...")
        try:
            import subprocess
            import sys
            subprocess.check_call([sys.executable, "-m", "pip", "install", "rasterio", "-q"])
            print("✓ rasterio installed")
        except Exception as e:
            print(f"Error installing rasterio: {e}")


    # Run main pipeline
    # Store the results in variables that can be used in subsequent cells
    original_image_arr, noisy_image_arr, mean_filtered_image_arr, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma = main()
```

```python
# Run cell oyemYZjfY-k8
# This cell implements SVD + Wavelet denoising and demonstrates its usage.
# Ensure necessary libraries (numpy, matplotlib, pywt, cv2, skimage, sklearn) are installed.
# This cell's main function uses a hardcoded image path or sample data.
# The output will include visualizations and quality metrics for the SVD + Wavelet method.

# Since the code is already in cell oyemYZjfY-k8, you can execute it directly.
# The content of cell oyemYZjfY-k8 is replicated below for completeness if needed,
# but executing cell oyemYZjfY-k8 is the intended action.

# Replicating content of cell oyemYZjfY-k8:
# (This section is for reference; executing cell oyemYZjfY-k8 is the primary action)

#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            if file_path.lower().endswith(('.tif', '.tiff')):
                # For GeoTIFF files
                try:
                    from osgeo import gdal
                    dataset = gdal.Open(file_path)
                    data_array = dataset.ReadAsArray()

                    # Handle multi-band data
                    if len(data_array.shape) == 3:
                        data_array = np.mean(data_array, axis=0)  # Convert to grayscale

                    # Normalize to [0, 1]
                    noisy_data = (data_array - data_array.min()) / (data_array.max() - data_array.min())

                except ImportError:
                    print("GDAL not available. Using OpenCV...")
                    img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                    if img is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    noisy_data = img_as_float(img)

            elif file_path.lower().endswith('.npy'):
                noisy_data = np.load(file_path)

            else:
                # Standard image formats
                img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                if img is None:
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None
                noisy_data = img_as_float(img)

            print(f"Loaded satellite data from: {file_path}")
            print(f"Data shape: {noisy_data.shape}")

        self.original_data = noisy_data
        return noisy_data

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()

    # Example for processing your own data:
    # results = process_your_satellite_data(
    #     "path/to/your/satellite_data.tif",
    #     "path/to/output/denoised_data.npy"
    # )
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D
from tensorflow.keras.models import Model
import numpy as np
import matplotlib.pyplot as plt

# Assuming original_image_arr and noisy_image_arr are available from previous steps.
# We'll need to prepare data (create patches) before training the model.
# This cell focuses on defining the model architecture.

print("Defining Autoencoder model...")

# Define the input shape (assuming grayscale images for simplicity, adjust for color)
# If your images are color (e.g., RGB), change the last dimension to 3: (image_height, image_width, 3)
# For this example, let's assume the image will be processed as grayscale or the model
# will be adapted for the channel dimension of the input data.
# We need to know the shape of the input image. Let's assume the input patches
# will have a shape like (patch_size, patch_size, num_channels).
# For now, we'll define a placeholder input shape and clarify the data preparation step.

# Placeholder input shape - will be determined by the data preparation step (Step 1)
# Assuming we'll create patches of size 32x32 with 3 channels (if using the RGB image from ecc1aab3)
input_shape = (32, 32, 3)
print(f"Assuming input patch shape for the model: {input_shape}")


# Encoder
input_img = Input(shape=input_shape)

x = Conv2D(32, (3, 3), activation='relu', padding='same')(input_img)
x = MaxPooling2D((2, 2), padding='same')(x)
x = Conv2D(32, (3, 3), activation='relu', padding='same')(x)
encoded = MaxPooling2D((2, 2), padding='same')(x)

# At this point, the representation is (8, 8, 32) assuming 32x32 input

# Decoder
x = Conv2D(32, (3, 3), activation='relu', padding='same')(encoded)
x = UpSampling2D((2, 2))(x)
x = Conv2D(32, (3, 3), activation='relu', padding='same')(x)
x = UpSampling2D((2, 2))(x)
# Output layer should match the input shape's channel dimension
# If input is grayscale (1 channel), output should be Conv2D(1, ...)
# If input is RGB (3 channels), output should be Conv2D(3, ...)
# Assuming the input patches have 3 channels based on the likely input from ecc1aab3
decoded = Conv2D(input_shape[-1], (3, 3), activation='sigmoid', padding='same')(x) # Use sigmoid for output range 0-1 if data is normalized


# Autoencoder Model
autoencoder = Model(input_img, decoded)

# Compile the model
# Use mean squared error as the loss function for denoising
# Use Adam optimizer
autoencoder.compile(optimizer='adam', loss='mean_squared_error')

print("Autoencoder model defined and compiled.")
print(autoencoder.summary())

# Note: This cell only defines the model architecture.
# You need to run the data preparation step (Step 1 in the plan)
# and the training step (Step 3 in the plan) before using this model for denoising.
```

```python
import numpy as np
# Assuming original_image_arr and noisy_image_arr are available from a previous cell (e.g., ecc1aab3)

print("Preparing data for Autoencoder (creating patches)...")

# Define patch size and overlap
# The patch size should match the input shape defined in the autoencoder model (cell QkimXJteZP2Q)
patch_size = 32
overlap = 16 # Overlap between patches

# Get image dimensions
# Assuming original_image_arr and noisy_image_arr have the same shape
h, w, c = original_image_arr.shape

# Calculate number of patches along height and width
# Ensure we cover the entire image, potentially with padding if needed later
# For simplicity, let's use strides considering overlap
stride = patch_size - overlap
num_patches_h = (h - patch_size) // stride + 1
num_patches_w = (w - patch_size) // stride + 1

# Initialize lists to store patches
original_patches = []
noisy_patches = []

print(f"Extracting {num_patches_h * num_patches_w} patches of size {patch_size}x{patch_size} with {overlap} overlap...")

# Extract overlapping patches
for i in range(num_patches_h):
    for j in range(num_patches_w):
        # Calculate starting coordinates for the patch
        start_h = i * stride
        start_w = j * stride

        # Extract patches from both images
        original_patch = original_image_arr[start_h : start_h + patch_size, start_w : start_w + patch_size, :]
        noisy_patch = noisy_image_arr[start_h : start_h + patch_size, start_w : start_w + patch_size, :]

        # Append patches to lists
        original_patches.append(original_patch)
        noisy_patches.append(noisy_patch)

# Convert lists of patches to NumPy arrays
# The shape will be (num_patches, patch_size, patch_size, num_channels)
original_patches_arr = np.array(original_patches)
noisy_patches_arr = np.array(noisy_patches)

print("Data preparation complete.")
print(f"Shape of original patches array: {original_patches_arr.shape}")
print(f"Shape of noisy patches array: {noisy_patches_arr.shape}")

# You can now use original_patches_arr and noisy_patches_arr for training the autoencoder.
# Typically, you would split these into training and validation sets.
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pywt
from skimage import img_as_float, data # Import data module from skimage
from sklearn.metrics import mean_squared_error
import cv2

class SatelliteWaveletProcessor:
    def __init__(self, wavelet='db4', mode='symmetric'):
        """
        Initialize the wavelet processor

        Parameters:
        wavelet: str, wavelet type (e.g., 'db4', 'db8', 'haar', 'bior2.2')
        mode: str, boundary condition mode
        """
        self.wavelet = wavelet
        self.mode = mode

    def load_satellite_data(self, file_path='/content/sample_data/mnist_train_small.csv', use_sample=False):
        """
        Load satellite data (replace with your actual data loading)
        """
        if use_sample:
            # Using sample data for demonstration
            # Replace this with your actual satellite data loading
            image_data = img_as_float(data.astronaut()) # Use skimage.data
            if len(image_data.shape) == 3:
                image_data = np.mean(image_data, axis=2)  # Convert to grayscale
            return image_data
        else:
          if file_path is None:
            # Provide a default path or raise an error if no file_path is given when not using sample data
            print("No file_path provided when use_sample is False.")
            return None
          # Use cv2.imread to load the image
          image_data = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
          if image_data is None:
            print(f"Error: Unable to load image from {file_path}")
            return None
          image_data = img_as_float(image_data) # Convert to float for processing
          return image_data


    def add_noise(self, data, noise_type='gaussian', noise_level=0.1):
        """
        Add noise to simulate noisy satellite data
        """
        if noise_type == 'gaussian':
            noise = np.random.normal(0, noise_level, data.shape)
        elif noise_type == 'salt_pepper':
            noise = np.random.random(data.shape)
            noise[noise < noise_level/2] = -1
            noise[noise > 1 - noise_level/2] = 1
            noise[(noise >= noise_level/2) & (noise <= 1 - noise_level/2)] = 0
            return data + noise

        return data + noise

    def wavelet_decompose(self, data, levels=3):
        """
        Perform wavelet decomposition
        """
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)
        return coeffs

    def estimate_noise_sigma(self, coeffs):
        """
        Estimate noise standard deviation using the robust median estimator
        """
        # Use the finest detail coefficients (HH at level 1)
        hh_coeffs = coeffs[-1][-1]  # Last level, HH component
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745
        return sigma

    def soft_threshold(self, coeffs, threshold):
        """
        Apply soft thresholding to wavelet coefficients
        """
        def soft_thresh(x, thresh):
            return np.sign(x) * np.maximum(np.abs(x) - thresh, 0)

        # Apply thresholding to detail coefficients only (not approximation)
        thresholded_coeffs = list(coeffs)
        for i in range(1, len(coeffs)):
            if isinstance(coeffs[i], tuple):
                # Detail coefficients (LH, HL, HH)
                thresholded_coeffs[i] = tuple([
                    soft_thresh(detail_coeff, threshold)
                    for detail_coeff in coeffs[i]
                ])

        return thresholded_coeffs

    def hard_threshold(self, coeffs, threshold):
        """
        Apply hard thresholding to wavelet coefficients
        """
        def hard_thresh(x, thresh):
            return x * (np.abs(x) > thresh)

        # Apply thresholding to detail coefficients only
        thresholded_coeffs = list(coeffs)
        for i in range(1, len(coeffs)):
            if isinstance(coeffs[i], tuple):
                thresholded_coeffs[i] = tuple([
                    hard_thresh(detail_coeff, threshold)
                    for detail_coeff in coeffs[i]
                ])

        return thresholded_coeffs

    def bayes_threshold(self, coeffs):
        """
        Apply BayesShrink thresholding (adaptive thresholding)
        """
        sigma = self.estimate_noise_sigma(coeffs)

        thresholded_coeffs = list(coeffs)
        for i in range(1, len(coeffs)):
            if isinstance(coeffs[i], tuple):
                new_detail_coeffs = []
                for detail_coeff in coeffs[i]:
                    # Estimate signal variance
                    sigma_y_sq = np.var(detail_coeff)
                    sigma_x_sq = max(0, sigma_y_sq - sigma**2)

                    if sigma_x_sq > 0:
                        threshold = sigma**2 / np.sqrt(sigma_x_sq)
                    else:
                        threshold = np.max(np.abs(detail_coeff))

                    # Apply soft thresholding
                    thresholded = np.sign(detail_coeff) * np.maximum(
                        np.abs(detail_coeff) - threshold, 0
                    )
                    new_detail_coeffs.append(thresholded)

                thresholded_coeffs[i] = tuple(new_detail_coeffs)

        return thresholded_coeffs

    def wavelet_reconstruct(self, coeffs):
        """
        Reconstruct data from wavelet coefficients
        """
        return pywt.waverec2(coeffs, self.wavelet, mode=self.mode)

    def calculate_metrics(self, original, denoised):
        """
        Calculate quality metrics
        """
        mse = mean_squared_error(original, denoised)
        psnr = 20 * np.log10(np.max(original) / np.sqrt(mse)) if mse > 0 and np.max(original) > 0 else float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        snr = 10 * np.log10(signal_power / noise_power) if noise_power > 0 else float('inf')

        return {'MSE': mse, 'PSNR': psnr, 'SNR': snr}

    def process_satellite_data(self, data, method='soft', threshold=None, levels=3):
        """
        Complete wavelet thresholding pipeline
        """
        # Decompose
        coeffs = self.wavelet_decompose(data, levels)

        # Apply thresholding
        if method == 'soft':
            if threshold is None:
                sigma = self.estimate_noise_sigma(coeffs)
                threshold = sigma * np.sqrt(2 * np.log(data.size))
            thresholded_coeffs = self.soft_threshold(coeffs, threshold)

        elif method == 'hard':
            if threshold is None:
                sigma = self.estimate_noise_sigma(coeffs)
                threshold = sigma * np.sqrt(2 * np.log(data.size))
            thresholded_coeffs = self.hard_threshold(coeffs, threshold)

        elif method == 'bayes':
            thresholded_coeffs = self.bayes_threshold(coeffs)

        else:
            raise ValueError("Method must be 'soft', 'hard', or 'bayes'")

        # Reconstruct
        denoised_data = self.wavelet_reconstruct(thresholded_coeffs)

        return denoised_data, coeffs, thresholded_coeffs

    def visualize_results(self, original, noisy, denoised, save_path=None):
        """
        Visualize original, noisy, and denoised satellite data
        """
        fig, axes = plt.subplots(1, 4, figsize=(16, 4))

        # Original
        axes[0].imshow(original, cmap='gray')
        axes[0].set_title('Original Satellite Data')
        axes[0].axis('off')

        # Noisy
        axes[1].imshow(noisy, cmap='gray')
        axes[1].set_title('Noisy Data')
        axes[1].axis('off')

        # Denoised
        axes[2].imshow(denoised, cmap='gray')
        axes[2].set_title('Denoised Data')
        axes[2].axis('off')

        # Difference
        diff = np.abs(original - denoised)
        axes[3].imshow(diff, cmap='hot')
        axes[3].set_title('Absolute Difference')
        axes[3].axis('off')

        plt.tight_layout()

        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')

        plt.show()

# Example usage and demonstration
def main():
    # Initialize processor
    processor = SatelliteWaveletProcessor(wavelet='db8', mode='symmetric')

    # Load satellite data (using sample data for demonstration)
    original_data = processor.load_satellite_data(file_path="/content/Sat_Image_Denoising/2_A.bmp",use_sample=False)

    if original_data is None:
      print("Exiting due to inability to load image data.")
      return

    # Add noise to simulate noisy satellite data
    noisy_data = processor.add_noise(original_data, noise_type='gaussian', noise_level=0.05)

    print("Processing satellite data with different thresholding methods...")

    # Test different methods
    methods = ['soft', 'hard', 'bayes']
    results = {}

    for method in methods:
        print(f"\nProcessing with {method} thresholding...")

        denoised, original_coeffs, thresholded_coeffs = processor.process_satellite_data(
            noisy_data, method=method, levels=4
        )

        # Calculate metrics
        metrics = processor.calculate_metrics(original_data, denoised)
        results[method] = {'denoised': denoised, 'metrics': metrics}

        print(f"{method.upper()} Thresholding Results:")
        print(f"  PSNR: {metrics['PSNR']:.2f} dB")
        print(f"  SNR: {metrics['SNR']:.2f} dB")
        print(f"  MSE: {metrics['MSE']:.6f}")

    # Find best method
    best_method = max(results.keys(), key=lambda k: results[k]['metrics']['PSNR'])
    print(f"\nBest method based on PSNR: {best_method.upper()}")

    # Visualize results for the best method
    processor.visualize_results(
        original_data,
        noisy_data,
        results[best_method]['denoised']
    )

    # Compare wavelets
    print("\nComparing different wavelet families...")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        proc = SatelliteWaveletProcessor(wavelet=wav)
        denoised, _, _ = proc.process_satellite_data(noisy_data, method='soft', levels=3)
        metrics = proc.calculate_metrics(original_data, denoised)
        print(f"{wav}: PSNR = {metrics['PSNR']:.2f} dB")

if __name__ == "__main__":
    main()

# Additional utility functions for real satellite data

def load_geotiff_data(file_path):
    """
    Load GeoTIFF satellite data using GDAL
    Note: requires gdal package (pip install gdal)
    """
    try:
        from osgeo import gdal
        dataset = gdal.Open(file_path)
        data = dataset.ReadAsArray()
        if len(data.shape) == 3:
            # Multi-band data - you might want to process each band separately
            # or combine them in a specific way
            data = np.mean(data, axis=0)  # Simple average
        return data
    except ImportError:
        print("GDAL not available. Install with: pip install gdal")
        return None

def save_processed_data(data, output_path, format='numpy'):
    """
    Save processed satellite data
    """
    if format == 'numpy':
        np.save(output_path, data)
    elif format == 'image':
        # Normalize to 0-255 for image saving
        normalized = ((data - data.min()) / (data.max() - data.min()) * 255).astype(np.uint8)
        cv2.imwrite(output_path, normalized)
    else:
        raise ValueError("Format must be 'numpy' or 'image'")

# Batch processing function
def batch_process_satellite_files(input_dir, output_dir, processor_params=None):
    """
    Process multiple satellite files in batch
    """
    import os
    import glob

    if processor_params is None:
        processor_params = {'wavelet': 'db8', 'mode': 'symmetric'}

    processor = SatelliteWaveletProcessor(**processor_params)

    # Find all files in input directory
    file_patterns = ['*.tif', '*.tiff', '*.npy', '*.png', '*.jpg','*.bmp']
    files = []
    for pattern in file_patterns:
        files.extend(glob.glob(os.path.join(input_dir, pattern)))

    print(f"Found {len(files)} files to process")

    for file_path in files:
        try:
            # Load data (you'll need to modify this based on your file format)
            if file_path.endswith(('.tif', '.tiff')):
                data = load_geotiff_data(file_path)
            elif file_path.endswith('.npy'):
                data = np.load(file_path)
            else:
                data = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                data = img_as_float(data)

            if data is not None:
                # Process
                denoised, _, _ = processor.process_satellite_data(
                    data, method='soft', levels=4
                )

                # Save
                filename = os.path.basename(file_path)
                name, ext = os.path.splitext(filename)
                output_path = os.path.join(output_dir, f"{name}_denoised{ext}")
                save_processed_data(denoised, output_path, format='numpy')

                print(f"Processed: {filename}")

        except Exception as e:
            print(f"Error processing {file_path}: {str(e)}")
```

```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            if file_path.lower().endswith(('.tif', '.tiff')):
                # For GeoTIFF files
                try:
                    from osgeo import gdal
                    dataset = gdal.Open(file_path)
                    data_array = dataset.ReadAsArray()

                    # Handle multi-band data
                    if len(data_array.shape) == 3:
                        data_array = np.mean(data_array, axis=0)  # Convert to grayscale

                    # Normalize to [0, 1]
                    noisy_data = (data_array - data_array.min()) / (data_array.max() - data_array.min())

                except ImportError:
                    print("GDAL not available. Using OpenCV...")
                    img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                    if img is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    noisy_data = img_as_float(img)

            elif file_path.lower().endswith('.npy'):
                noisy_data = np.load(file_path)

            else:
                # Standard image formats
                img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                if img is None:
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None
                noisy_data = img_as_float(img)

            print(f"Loaded satellite data from: {file_path}")
            print(f"Data shape: {noisy_data.shape}")

        self.original_data = noisy_data
        return noisy_data

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        psnr = 20 * np.log10(1.0 / np.sqrt(mse)) if mse > 0 else float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        snr = 10 * np.log10(signal_power / noise_power) if noise_power > 0 else float('inf')

        # Structural Similarity (simplified)
        ssim_score = 1 - mse  # Simplified SSIM approximation

        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM_approx': ssim_score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Original noisy data
        axes[0, 0].imshow(original, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            axes[1, 0].imshow(svd_diff, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            coeffs_array, slices = pywt.coeffs_to_array(coeffs)
            axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
            axes[1, 2].set_title('Wavelet Coefficients')
            axes[1, 2].axis('off')

        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM (approx): {metrics['SSIM_approx']:.4f}")

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  MSE: {metrics['MSE']:.6f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method
        best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'])
        print(f"\nBest method: {best_method['config']}")
        print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")

        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    satellite_data = denoiser.load_satellite_data(file_path="/content/Sat_Image_Denoising/2_A.bmp", use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
        results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
            satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
        )
        metrics_wav = denoiser_wav.calculate_quality_metrics(
            satellite_data, results_wav['final_denoised']
        )
        print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your own satellite data file

    Parameters:
    file_path: str, path to your satellite data
    output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()

    # Example for processing your own data:
    # results = process_your_satellite_data(
    #     "path/to/your/satellite_data.tif",
    #     "path/to/output/denoised_data.npy"
    # )
```

```python
#Weiner Filter + Wavelet Threshold for Satellite Data
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal, ndimage
from scipy.fft import fft2, ifft2, fftshift, ifftshift
import pywt
from skimage import data, filters
from sklearn.datasets import make_classification
import warnings
warnings.filterwarnings('ignore')

class WienerWaveletDenoiser:
    def __init__(self, file_path="/content/Sat_Image_Denoising/2_A.bmp"):
        self.file_path = file_path
        self.wiener_noise_var = None
        self.signal_var = None

    def estimate_noise_variance(self, data):
        """Estimate noise variance using robust median absolute deviation"""
        if data.ndim == 1:
            # For 1D signals, use high-pass filtering
            diff = np.diff(data)
            noise_var = np.median(np.abs(diff)) / 0.6745
        else:
            # For 2D images, use Laplacian method
            laplacian = np.array([[0, -1, 0], [-1, 4, -1], [0, -1, 0]])
            filtered = ndimage.convolve(data, laplacian)
            noise_var = np.var(filtered) * 0.5
        return noise_var ** 2

    def wiener_filter_1d(self, signal_data, noise_var=None):
        """Apply Wiener filter to 1D signal"""
        if noise_var is None:
            noise_var = self.estimate_noise_variance(signal_data)

        # FFT of the signal
        signal_fft = np.fft.fft(signal_data)
        power_spectrum = np.abs(signal_fft) ** 2

        # Estimate signal power spectrum (assuming signal >> noise in low frequencies)
        signal_power = np.maximum(power_spectrum - noise_var, 0.1 * power_spectrum)

        # Wiener filter transfer function
        wiener_filter = signal_power / (signal_power + noise_var)

        # Apply filter
        filtered_fft = signal_fft * wiener_filter
        filtered_signal = np.real(np.fft.ifft(filtered_fft))

        return filtered_signal, wiener_filter

    def wiener_filter_2d(self, image, noise_var=None):
        """Apply Wiener filter to 2D image"""
        if noise_var is None:
            noise_var = self.estimate_noise_variance(image)

        # FFT of the image
        image_fft = fft2(image)
        power_spectrum = np.abs(image_fft) ** 2

        # Estimate signal power spectrum
        signal_power = np.maximum(power_spectrum - noise_var, 0.1 * power_spectrum)

        # Wiener filter transfer function
        wiener_filter = signal_power / (signal_power + noise_var)

        # Apply filter
        filtered_fft = image_fft * wiener_filter
        filtered_image = np.real(ifft2(filtered_fft))

        return filtered_image, wiener_filter

    def wavelet_threshold_denoise(self, data, wavelet='db8', mode='soft', method='bayes'):
        """
        Apply wavelet denoising with various thresholding methods

        Parameters:
        - data: input signal/image
        - wavelet: wavelet type ('db8', 'haar', 'coif2', etc.)
        - mode: thresholding mode ('soft' or 'hard')
        - method: threshold estimation method ('bayes', 'sure', 'minimax')
        """

        if data.ndim == 1:
            return self._wavelet_denoise_1d(data, wavelet, mode, method)
        else:
            return self._wavelet_denoise_2d(data, wavelet, mode, method)

    def _wavelet_denoise_1d(self, signal_data, wavelet, mode, method):
        """Wavelet denoising for 1D signals"""
        # Decompose
        coeffs = pywt.wavedec(signal_data, wavelet, level=None)

        # Estimate noise level from finest detail coefficients
        sigma = np.median(np.abs(coeffs[-1])) / 0.6745

        # Calculate threshold based on method
        N = len(signal_data)
        if method == 'bayes':
            threshold = sigma * np.sqrt(2 * np.log(N))
        elif method == 'sure':
            threshold = sigma * np.sqrt(2 * np.log(N))  # Simplified SURE
        elif method == 'minimax':
            threshold = sigma * (0.3936 + 0.1829 * np.log2(N))
        else:
            threshold = sigma * np.sqrt(2 * np.log(N))

        # Apply thresholding to detail coefficients
        coeffs_thresh = list(coeffs)
        coeffs_thresh[1:] = [pywt.threshold(detail, threshold, mode=mode)
                            for detail in coeffs[1:]]

        # Reconstruct
        denoised = pywt.waverec(coeffs_thresh, wavelet)

        return denoised, threshold, coeffs, coeffs_thresh

    def _wavelet_denoise_2d(self, image, wavelet, mode, method):
        """Wavelet denoising for 2D images"""
        # Decompose
        coeffs = pywt.wavedec2(image, wavelet, level=None)

        # Estimate noise level from finest detail coefficients
        sigma = np.median(np.abs(coeffs[-1][-1])) / 0.6745

        # Calculate threshold
        N = image.size
        if method == 'bayes':
            threshold = sigma * np.sqrt(2 * np.log(N))
        elif method == 'sure':
            threshold = sigma * np.sqrt(2 * np.log(N))
        elif method == 'minimax':
            threshold = sigma * (0.3936 + 0.1829 * np.log2(N))
        else:
            threshold = sigma * np.sqrt(2 * np.log(N))

        # Apply thresholding to detail coefficients
        coeffs_thresh = list(coeffs)
        coeffs_thresh[0] = coeffs[0]  # Keep approximation coefficients
        for i in range(1, len(coeffs)):
            coeffs_thresh[i] = tuple([pywt.threshold(detail, threshold, mode=mode)
                                    for detail in coeffs[i]])

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresh, wavelet)

        return denoised, threshold, coeffs, coeffs_thresh

    def combined_denoise(self, data, wavelet='db8', threshold_mode='soft',
                        threshold_method='bayes', noise_var=None):
        """
        Combined Wiener filter + Wavelet denoising
        """
        print("Applying Wiener filter...")

        if data.ndim == 1:
            wiener_filtered, wiener_tf = self.wiener_filter_1d(data, noise_var)
        else:
            wiener_filtered, wiener_tf = self.wiener_filter_2d(data, noise_var)

        print("Applying wavelet denoising...")

        if data.ndim == 1:
            final_denoised, threshold, coeffs, coeffs_thresh = \
                self._wavelet_denoise_1d(wiener_filtered, wavelet, threshold_mode, threshold_method)
        else:
            final_denoised, threshold, coeffs, coeffs_thresh = \
                self._wavelet_denoise_2d(wiener_filtered, wavelet, threshold_mode, threshold_method)

        return {
            'original': data,
            'wiener_filtered': wiener_filtered,
            'final_denoised': final_denoised,
            'wiener_transfer_function': wiener_tf,
            'wavelet_threshold': threshold,
            'wavelet_coeffs_original': coeffs,
            'wavelet_coeffs_thresholded': coeffs_thresh
        }

# Example usage and demonstrations
def create_noisy_signal_1d(length=1000, noise_level=0.3):
    """Create a noisy 1D test signal"""
    t = np.linspace(0, 1, length)
    # Clean signal: combination of sine waves
    clean = (np.sin(2 * np.pi * 5 * t) +
             0.5 * np.sin(2 * np.pi * 20 * t) +
             0.3 * np.sin(2 * np.pi * 50 * t))
    # Add noise
    noise = noise_level * np.random.randn(length)
    noisy = clean + noise
    return t, clean, noisy

def create_noisy_image_2d(size=256, noise_level=0.2):
    """Create a noisy 2D test image"""
    # Create synthetic image with geometric patterns
    x, y = np.meshgrid(np.linspace(-1, 1, size), np.linspace(-1, 1, size))
    clean = (np.exp(-(x**2 + y**2)/0.3) +
             0.5 * np.sin(10 * x) * np.cos(10 * y) +
             0.3 * (x**2 + y**2 < 0.5))
    # Add noise
    noise = noise_level * np.random.randn(size, size)
    noisy = clean + noise
    return clean, noisy

def demonstrate_1d_denoising():
    """Demonstrate 1D signal denoising"""
    print("=== 1D Signal Denoising Demo ===")

    # Create test signal
    t, clean_signal, noisy_signal = create_noisy_signal_1d()

    # Apply denoising
    denoiser = WienerWaveletDenoiser()
    results = denoiser.combined_denoise(noisy_signal, wavelet='db8',
                                       threshold_mode='soft', threshold_method='bayes')

    # Calculate metrics
    mse_original = np.mean((noisy_signal - clean_signal)**2)
    mse_wiener = np.mean((results['wiener_filtered'] - clean_signal)**2)
    mse_final = np.mean((results['final_denoised'] - clean_signal)**2)

    print(f"MSE - Original: {mse_original:.6f}")
    print(f"MSE - After Wiener: {mse_wiener:.6f}")
    print(f"MSE - After Wavelet: {mse_final:.6f}")
    print(f"Improvement: {(mse_original/mse_final):.2f}x")

    # Plot results
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))

    # Time domain signals
    axes[0,0].plot(t, clean_signal, 'g-', label='Clean', linewidth=2)
    axes[0,0].plot(t, noisy_signal, 'r-', alpha=0.7, label='Noisy')
    axes[0,0].set_title('Original Signals')
    axes[0,0].legend()
    axes[0,0].grid(True)

    axes[0,1].plot(t, clean_signal, 'g-', label='Clean', linewidth=2)
    axes[0,1].plot(t, results['wiener_filtered'], 'b-', alpha=0.8, label='Wiener Filtered')
    axes[0,1].plot(t, results['final_denoised'], 'm-', alpha=0.8, label='Final Denoised')
    axes[0,1].set_title('Denoising Results')
    axes[0,1].legend()
    axes[0,1].grid(True)

    # Frequency domain
    freqs = np.fft.fftfreq(len(noisy_signal), t[1]-t[0])
    axes[1,0].semilogy(freqs[:len(freqs)//2],
                       np.abs(np.fft.fft(noisy_signal))[:len(freqs)//2],
                       'r-', alpha=0.7, label='Noisy')
    axes[1,0].semilogy(freqs[:len(freqs)//2],
                       np.abs(np.fft.fft(results['final_denoised']))[:len(freqs)//2],
                       'm-', label='Denoised')
    axes[1,0].set_title('Frequency Spectrum')
    axes[1,0].set_xlabel('Frequency (Hz)')
    axes[1,0].legend()
    axes[1,0].grid(True)

    # Wiener filter transfer function
    axes[1,1].plot(results['wiener_transfer_function'])
    axes[1,1].set_title('Wiener Filter Transfer Function')
    axes[1,1].set_xlabel('Frequency Bin')
    axes[1,1].set_ylabel('Filter Response')
    axes[1,1].grid(True)

    plt.tight_layout()
    plt.show()

    return results

def demonstrate_2d_denoising():
    """Demonstrate 2D image denoising"""
    print("\n=== 2D Image Denoising Demo ===")

    # Create test image
    clean_image, noisy_image = create_noisy_image_2d(size=128)

    # Apply denoising
    denoiser = WienerWaveletDenoiser()
    results = denoiser.combined_denoise(noisy_image, wavelet='db4',
                                       threshold_mode='soft', threshold_method='bayes')

    # Calculate metrics
    mse_original = np.mean((noisy_image - clean_image)**2)
    mse_wiener = np.mean((results['wiener_filtered'] - clean_image)**2)
    mse_final = np.mean((results['final_denoised'] - clean_image)**2)

    print(f"MSE - Original: {mse_original:.6f}")
    print(f"MSE - After Wiener: {mse_wiener:.6f}")
    print(f"MSE - After Wavelet: {mse_final:.6f}")
    print(f"Improvement: {(mse_original/mse_final):.2f}x")

    # Plot results
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))

    im1 = axes[0,0].imshow(clean_image, cmap='gray')
    axes[0,0].set_title('Clean Image')
    axes[0,0].axis('off')
    plt.colorbar(im1, ax=axes[0,0], shrink=0.8)

    im2 = axes[0,1].imshow(noisy_image, cmap='gray')
    axes[0,1].set_title('Noisy Image')
    axes[0,1].axis('off')
    plt.colorbar(im2, ax=axes[0,1], shrink=0.8)

    im3 = axes[0,2].imshow(results['wiener_filtered'], cmap='gray')
    axes[0,2].set_title('Wiener Filtered')
    axes[0,2].axis('off')
    plt.colorbar(im3, ax=axes[0,2], shrink=0.8)

    im4 = axes[1,0].imshow(results['final_denoised'], cmap='gray')
    axes[1,0].set_title('Final Denoised')
    axes[1,0].axis('off')
    plt.colorbar(im4, ax=axes[1,0], shrink=0.8)

    # Show difference images
    diff_wiener = np.abs(results['wiener_filtered'] - clean_image)
    diff_final = np.abs(results['final_denoised'] - clean_image)

    im5 = axes[1,1].imshow(diff_wiener, cmap='hot')
    axes[1,1].set_title('Wiener Error')
    axes[1,1].axis('off')
    plt.colorbar(im5, ax=axes[1,1], shrink=0.8)

    im6 = axes[1,2].imshow(diff_final, cmap='hot')
    axes[1,2].set_title('Final Error')
    axes[1,2].axis('off')
    plt.colorbar(im6, ax=axes[1,2], shrink=0.8)

    plt.tight_layout()
    plt.show()

    return results

def compare_wavelets_and_methods():
    """Compare different wavelets and thresholding methods"""
    print("\n=== Wavelet and Method Comparison ===")

    # Create test signal
    t, clean_signal, noisy_signal = create_noisy_signal_1d(length=512)

    wavelets = ['db4', 'db8', 'haar', 'coif2', 'bior2.2']
    methods = ['bayes', 'sure', 'minimax']
    modes = ['soft', 'hard']

    denoiser = WienerWaveletDenoiser()
    results_comparison = {}

    print("\nMSE Comparison:")
    print("Wavelet\tMethod\tMode\tMSE")
    print("-" * 40)

    for wavelet in wavelets:
        for method in methods:
            for mode in modes:
                try:
                    result = denoiser.combined_denoise(noisy_signal,
                                                     wavelet=wavelet,
                                                     threshold_mode=mode,
                                                     threshold_method=method)
                    mse = np.mean((result['final_denoised'] - clean_signal)**2)
                    results_comparison[f"{wavelet}_{method}_{mode}"] = mse
                    print(f"{wavelet}\t{method}\t{mode}\t{mse:.6f}")
                except:
                    print(f"{wavelet}\t{method}\t{mode}\tFailed")

    # Find best combination
    best_combo = min(results_comparison.items(), key=lambda x: x[1])
    print(f"\nBest combination: {best_combo[0]} with MSE: {best_combo[1]:.6f}")

    return results_comparison

# Main execution
if __name__ == "__main__":
    # Run demonstrations
    print("Starting Wiener Filter + Wavelet Denoising Demonstrations...")

    # 1D signal denoising
    results_1d = demonstrate_1d_denoising()

    # 2D image denoising
    results_2d = demonstrate_2d_denoising()

    # Compare different wavelets and methods
    comparison = compare_wavelets_and_methods()

    print("\n=== Usage Examples ===")
    print("""
    # Basic usage:
    denoiser = WienerWaveletDenoiser()

    # For 1D signals:
    results = denoiser.combined_denoise(your_signal, wavelet='db8',
                                       threshold_mode='soft', threshold_method='bayes')

    # For 2D images:
    results = denoiser.combined_denoise(your_image, wavelet='db4',
                                       threshold_mode='soft', threshold_method='bayes')

    # Access results:
    denoised_data = results['final_denoised']
    wiener_filtered = results['wiener_filtered']

    # Different wavelets to try: 'db4', 'db8', 'haar', 'coif2', 'bior2.2'
    # Threshold methods: 'bayes', 'sure', 'minimax'
    # Threshold modes: 'soft', 'hard'
    """)
```

```python
#Adaptive SVD Selection+Enhanced Wavelet Processing+Advanced Noise Handling
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal, ndimage, optimize
from scipy.fft import fft2, ifft2, fftshift, ifftshift
import pywt
from skimage import filters, restoration, measure, morphology, exposure
from sklearn.decomposition import TruncatedSVD
import cv2
import warnings
warnings.filterwarnings('ignore')

class EnhancedSatelliteDenoiser:
    def __init__(self):
        self.noise_profile = None
        self.optimal_components = None

    def estimate_noise_profile(self, image):
        """Advanced noise estimation for satellite imagery"""
        # Multi-scale noise estimation
        noise_estimates = []

        # Method 1: Robust median absolute deviation
        laplacian = np.array([[0, -1, 0], [-1, 4, -1], [0, -1, 0]])
        filtered = ndimage.convolve(image.astype(float), laplacian)
        noise_mad = np.median(np.abs(filtered - np.median(filtered))) / 0.6745
        noise_estimates.append(noise_mad)

        # Method 2: High-frequency analysis
        f_transform = fft2(image)
        h, w = f_transform.shape
        # Create a high-pass filter mask
        rows, cols = image.shape
        center_row, center_col = rows // 2, cols // 2
        mask = np.ones((rows, cols), dtype=np.uint8)
        # Define the low-frequency region to exclude
        r = min(rows, cols) // 8  # Adjust radius as needed
        cv2.circle(mask, (center_col, center_row), r, 0, -1)
        # Apply the mask in the frequency domain
        f_transform_shifted = fftshift(f_transform)
        high_freq_power = np.mean(np.abs(f_transform_shifted * mask)**2)
        noise_estimates.append(np.sqrt(high_freq_power) * 0.1)


        # Method 3: Wavelet-based estimation
        coeffs = pywt.dwt2(image, 'db4')
        _, (cH, cV, cD) = coeffs
        noise_wavelet = np.median(np.abs(cD)) / 0.6745
        noise_estimates.append(noise_wavelet)

        # Robust average
        self.noise_profile = np.median(noise_estimates)
        return self.noise_profile

    def adaptive_svd_denoising(self, image, max_components=None):
        """Improved SVD denoising with automatic rank selection"""
        # Reshape image for SVD
        original_shape = image.shape
        if len(original_shape) == 2:
            # For 2D image, create patches or use direct SVD
            h, w = original_shape

            # Method 1: Block-wise SVD for large images
            if h > 512 or w > 512:
                return self._block_svd_denoise(image, max_components)

            # Method 2: Direct SVD for smaller images
            U, s, Vt = np.linalg.svd(image, full_matrices=False)

            # Automatic rank selection using noise profile
            if max_components is None:
                noise_level = self.estimate_noise_profile(image)
                # Keep components above noise threshold
                signal_threshold = noise_level * np.sqrt(min(h, w))
                optimal_rank = np.sum(s > signal_threshold)
                optimal_rank = max(optimal_rank, min(h, w) // 10)  # At least 10% of components
                optimal_rank = min(optimal_rank, min(h, w) // 2)   # At most 50% of components
            else:
                optimal_rank = min(max_components, len(s))

            self.optimal_components = optimal_rank

            # Reconstruct with optimal rank
            s_filtered = s.copy()
            s_filtered[optimal_rank:] = 0
            denoised = U @ np.diag(s_filtered) @ Vt

            return denoised, optimal_rank, s

        return image, 0, None

    def _block_svd_denoise(self, image, max_components=None, block_size=256):
        """Block-wise SVD for large images"""
        h, w = image.shape
        denoised = np.zeros_like(image)
        overlap = block_size // 4

        for i in range(0, h - block_size + 1, block_size - overlap):
            for j in range(0, w - block_size + 1, block_size - overlap):
                # Extract block
                end_i = min(i + block_size, h)
                end_j = min(j + block_size, w)
                block = image[i:end_i, j:end_j]

                # Apply SVD to block
                U, s, Vt = np.linalg.svd(block, full_matrices=False)

                # Determine optimal rank for this block
                if max_components is None:
                    noise_level = np.median(np.abs(block.flatten() - np.median(block.flatten()))) / 0.6745
                    signal_threshold = noise_level * np.sqrt(min(block.shape))
                    optimal_rank = np.sum(s > signal_threshold)
                    optimal_rank = max(optimal_rank, len(s) // 10)
                    optimal_rank = min(optimal_rank, len(s) // 2)
                else:
                    optimal_rank = min(max_components, len(s))

                # Reconstruct block
                s_filtered = s.copy()
                s_filtered[optimal_rank:] = 0
                block_denoised = U @ np.diag(s_filtered) @ Vt

                # Blend with existing result (for overlapping regions)
                if i == 0 and j == 0:
                    denoised[i:end_i, j:end_j] = block_denoised
                else:
                    # Simple averaging for overlaps
                    mask = denoised[i:end_i, j:end_j] != 0
                    denoised[i:end_i, j:end_j] = np.where(mask,
                                                         (denoised[i:end_i, j:end_j] + block_denoised) / 2,
                                                         block_denoised)

        return denoised, optimal_rank, None

    def enhanced_wiener_filter(self, image, noise_var=None, psf=None):
        """Enhanced Wiener filter with PSF estimation"""
        if noise_var is None:
            noise_var = (self.estimate_noise_profile(image)) ** 2

        # If no PSF provided, estimate motion blur or use identity
        if psf is None:
            psf = np.array([[1]])  # Identity PSF (no blur)

        # Apply Wiener deconvolution
        if psf.size > 1:
            denoised = restoration.wiener(image, psf, noise_var, clip=False)
        else:
            # Simple Wiener filtering in frequency domain
            image_fft = fft2(image)
            power_spectrum = np.abs(image_fft) ** 2
            signal_power = np.maximum(power_spectrum - noise_var, 0.1 * power_spectrum)
            wiener_filter = signal_power / (power_spectrum + noise_var) # Corrected denominator
            filtered_fft = image_fft * wiener_filter
            denoised = np.real(ifft2(filtered_fft))

        return denoised

    def adaptive_wavelet_denoising(self, image, wavelet='auto', levels=None):
        """Enhanced wavelet denoising with adaptive thresholding"""
        # Determine optimal decomposition level
        if levels is None:
            min_dim = min(image.shape)
            levels = int(np.log2(min_dim)) - 2
            levels = max(1, min(levels, 6))

        # Multi-wavelet approach: try different wavelets and combine
        wavelets_to_try = ['db4', 'db8', 'coif2', 'bior2.2'] if wavelet == 'auto' else [wavelet]
        denoised_results = []

        for wvlt in wavelets_to_try:
            try:
                # Decompose
                coeffs = pywt.wavedec2(image, wvlt, level=levels)

                # Adaptive thresholding for each level
                coeffs_thresh = [coeffs[0]]  # Keep approximation coefficients

                for i in range(1, len(coeffs)):
                    detail_coeffs = coeffs[i]
                    thresh_coeffs = []

                    for detail in detail_coeffs:
                        # Level-specific threshold
                        sigma = np.median(np.abs(detail)) / 0.6745

                        # Bayes shrink threshold
                        if sigma > 0:
                            threshold = sigma**2 / np.sqrt(np.maximum(np.var(detail) - sigma**2, 0) + 1e-10)
                        else:
                            threshold = np.max(np.abs(detail)) # Use max value as threshold if sigma is zero

                        # Apply soft thresholding
                        thresh_detail = pywt.threshold(detail, threshold, mode='soft')
                        thresh_coeffs.append(thresh_detail)

                    coeffs_thresh.append(tuple(thresh_coeffs))

                # Reconstruct
                denoised = pywt.waverec2(coeffs_thresh, wvlt)

                # Ensure proper size
                if denoised.shape != image.shape:
                    denoised = denoised[:image.shape[0], :image.shape[1]]

                denoised_results.append(denoised)

            except Exception as e:
                print(f"Warning: Wavelet {wvlt} failed: {e}")
                continue

        if not denoised_results:
            return image

        # If multiple wavelets used, combine results
        if len(denoised_results) == 1:
            return denoised_results[0]
        else:
            # Weighted combination based on local variance
            final_result = np.zeros_like(image)
            total_weight = np.zeros_like(image)

            for result in denoised_results:
                # Local variance as weight (higher variance = more detail preserved)
                local_var = ndimage.generic_filter(result, np.var, size=5)
                weight = local_var / (np.sum([ndimage.generic_filter(r, np.var, size=5)
                                            for r in denoised_results], axis=0) + 1e-10)
                final_result += weight * result
                total_weight += weight

            # Normalize and clip to [0, 1]
            combined_result = final_result / (total_weight + 1e-10)
            return np.clip(combined_result, 0, 1)


    def edge_preserving_filter(self, image, iterations=1):
        """Apply edge-preserving smoothing to reduce remaining noise"""
        # Bilateral filter for edge preservation
        # Ensure image is in the correct format and range for cv2.bilateralFilter
        if image.dtype != np.uint8:
            # Scale to 0-255 for cv2.bilateralFilter and convert to uint8
            image_norm = exposure.rescale_intensity(image, out_range=(0, 255)).astype(np.uint8)
            filtered = cv2.bilateralFilter(image_norm, 9, 75, 75)
            # Scale back to the original range of the input float image
            # Assuming the input float image was in the range [0, 1]
            filtered = exposure.rescale_intensity(filtered.astype(float), out_range=(0.0, 1.0))
        else:
             # If already uint8, just apply the filter
             filtered = cv2.bilateralFilter(image, 9, 75, 75)
             # Convert to float and scale to [0, 1] if the original was float
             filtered = filtered.astype(float) / 255.0


        return filtered


    def enhance_contrast(self, image, method='adaptive'):
        """Post-processing contrast enhancement"""
        # Ensure image is in the correct format and range for scikit-image exposure functions
        if image.dtype != np.float64 or np.min(image) < 0 or np.max(image) > 1:
           image = img_as_float(image) # Convert to float and scale to [0, 1]

        if method == 'adaptive':
            # Adaptive histogram equalization
            return exposure.equalize_adapthist(image, clip_limit=0.01)
        elif method == 'gamma':
            # Gamma correction
            return exposure.adjust_gamma(image, gamma=0.8)
        elif method == 'sigmoid':
            # Sigmoid correction
            return exposure.adjust_sigmoid(image, cutoff=0.5, gain=10)
        else:
            return image

    def comprehensive_denoise(self, image, method='auto', enhance_contrast=True):
        """
        Comprehensive denoising pipeline

        Parameters:
        - image: input noisy image
        - method: 'svd_wavelet', 'wiener_wavelet', 'all', or 'auto'
        - enhance_contrast: apply post-processing contrast enhancement
        """
        print("Starting comprehensive satellite image denoising...")

        # Ensure float type and scale to [0, 1]
        if image.dtype != np.float64:
            image = img_as_float(image) # Convert to float and scale to [0, 1]

        # Store original for comparison
        original = image.copy()

        # Step 1: Initial noise estimation
        noise_level = self.estimate_noise_profile(image)
        print(f"Estimated noise level: {noise_level:.4f}")

        results = {'original': original}

        if method in ['svd_wavelet', 'all', 'auto']:
            print("Applying SVD denoising...")
            svd_denoised, optimal_rank, singular_values = self.adaptive_svd_denoising(image)
            print(f"Optimal SVD rank: {optimal_rank}")

            print("Applying wavelet denoising to SVD result...")
            svd_wavelet_denoised = self.adaptive_wavelet_denoising(svd_denoised, wavelet='auto')

            results['svd_denoised'] = svd_denoised
            results['svd_wavelet_denoised'] = svd_wavelet_denoised
            results['optimal_rank'] = optimal_rank

        if method in ['wiener_wavelet', 'all', 'auto']:
            print("Applying Wiener filter...")
            wiener_denoised = self.enhanced_wiener_filter(image, noise_var=noise_level**2)

            print("Applying wavelet denoising to Wiener result...")
            wiener_wavelet_denoised = self.adaptive_wavelet_denoising(wiener_denoised, wavelet='auto')

            results['wiener_denoised'] = wiener_denoised
            results['wiener_wavelet_denoised'] = wiener_wavelet_denoised

        # Choose best result based on method
        if method == 'svd_wavelet':
            final_denoised = results['svd_wavelet_denoised']
        elif method == 'wiener_wavelet':
            final_denoised = results['wiener_wavelet_denoised']
        elif method in ['all', 'auto']:
            # Combine both methods using local quality metrics
            if 'svd_wavelet_denoised' in results and 'wiener_wavelet_denoised' in results:
                svd_result = results['svd_wavelet_denoised']
                wiener_result = results['wiener_wavelet_denoised']

                # Local gradient preservation as quality metric
                original_grad = np.sqrt(np.gradient(original)[0]**2 + np.gradient(original)[1]**2)
                svd_grad = np.sqrt(np.gradient(svd_result)[0]**2 + np.gradient(svd_result)[1]**2)
                wiener_grad = np.sqrt(np.gradient(wiener_result)[0]**2 + np.gradient(wiener_result)[1]**2)

                svd_quality = 1 / (1 + np.abs(original_grad - svd_grad))
                wiener_quality = 1 / (1 + np.abs(original_grad - wiener_grad))

                # Smooth the quality maps
                svd_quality = ndimage.gaussian_filter(svd_quality, sigma=2)
                wiener_quality = ndimage.gaussian_filter(wiener_quality, sigma=2)

                # Normalize weights
                total_quality = svd_quality + wiener_quality + 1e-10
                svd_weight = svd_quality / total_quality
                wiener_weight = wiener_quality / total_quality

                final_denoised = svd_weight * svd_result + wiener_weight * wiener_result
            else:
                final_denoised = results.get('svd_wavelet_denoised',
                                           results.get('wiener_wavelet_denoised', image))

        # Ensure final denoised image is clipped to [0, 1] before post-processing
        final_denoised = np.clip(final_denoised, 0, 1)

        # Post-processing
        print("Applying edge-preserving filter...")
        final_denoised = self.edge_preserving_filter(final_denoised)

        if enhance_contrast:
            print("Enhancing contrast...")
            final_denoised = self.enhance_contrast(final_denoised, method='adaptive')

        results['final_denoised'] = final_denoised
        results['noise_level'] = noise_level

        # Calculate improvement metrics
        if np.std(original) > 0:  # Avoid division by zero
            # Calculate PSNR between original and denoised
            mse = np.mean((original - final_denoised)**2)
            if mse > 0:
                psnr = 10 * np.log10(1.0 / mse) # Assuming image range [0, 1]
                results['psnr'] = psnr
                print(f"PSNR: {psnr:.2f} dB")
            else:
                results['psnr'] = float('inf')
                print("PSNR: Inf dB (Perfect reconstruction)")


        print("Denoising complete!")
        return results

    def visualize_results(self, results):
        """Comprehensive visualization of denoising results"""
        n_results = len(results) - sum(key in ['noise_level', 'psnr', 'optimal_rank'] for key in results.keys()) # Exclude metrics
        cols = min(3, n_results)
        rows = (n_results + cols - 1) // cols

        fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))
        if rows == 1 and cols > 1:
            axes = axes.reshape(1, -1)
        elif rows > 1 and cols == 1:
             axes = axes.reshape(-1, 1)
        elif rows == 1 and cols == 1:
             axes = np.array([[axes]])


        plot_idx = 0
        for key, image in results.items():
            if key in ['noise_level', 'psnr', 'optimal_rank']:
                continue

            row = plot_idx // cols
            col = plot_idx % cols

            if isinstance(image, np.ndarray) and image.ndim == 2:
                im = axes[row, col].imshow(image, cmap='gray', vmin=image.min(), vmax=image.max())
                axes[row, col].set_title(key.replace('_', ' ').title())
                axes[row, col].axis('off')
                plt.colorbar(im, ax=axes[row, col], shrink=0.8)

            plot_idx += 1

        # Hide empty subplots
        for i in range(plot_idx, rows * cols):
            row = i // cols
            col = i % cols
            axes[row, col].axis('off')


        plt.tight_layout()
        plt.show()

def main_satellite_denoising_demo():
    """Demo with synthetic satellite-like imagery"""
    # Create synthetic satellite image with various features
    size = 256
    x, y = np.meshgrid(np.linspace(0, 1, size), np.linspace(0, 1, size))

    # Simulate different terrain features
    fields = 0.5 + 0.3 * np.sin(20 * x) * np.cos(20 * y)  # Agricultural fields
    forest = 0.3 * np.exp(-((x-0.2)**2 + (y-0.8)**2) / 0.1)  # Forest area
    urban = 0.8 * ((x > 0.6) & (x < 0.9) & (y > 0.1) & (y < 0.4))  # Urban area
    water = 0.1 * ((x-0.8)**2 + (y-0.7)**2 < 0.05)  # Water body

    clean_image = fields + forest + urban + water

    # Add realistic noise (sensor noise + atmospheric effects)
    noise_level = 0.15
    gaussian_noise = noise_level * np.random.randn(size, size)
    salt_pepper = np.random.random((size, size))
    clean_image[salt_pepper < 0.01] = 0  # Salt noise
    clean_image[salt_pepper > 0.99] = 1  # Pepper noise

    noisy_image = clean_image + gaussian_noise
    noisy_image = np.clip(noisy_image, 0, 1)

    # Apply denoising
    denoiser = EnhancedSatelliteDenoiser()
    results = denoiser.comprehensive_denoise(noisy_image, method='auto')

    # Visualize
    denoiser.visualize_results(results)

    return results

# Usage example for your satellite data
def process_your_satellite_data(image_array):
    """
    Process your actual satellite image

    Parameters:
    - image_array: numpy array of your satellite image
    """
    denoiser = EnhancedSatelliteDenoiser()

    # Method 1: SVD + Wavelet (similar to your current approach)
    results_svd = denoiser.comprehensive_denoise(image_array, method='svd_wavelet')

    # Method 2: Wiener + Wavelet (alternative approach)
    results_wiener = denoiser.comprehensive_denoise(image_array, method='wiener_wavelet')

    # Method 3: Combined approach (recommended)
    results_combined = denoiser.comprehensive_denoise(image_array, method='auto')

    return results_svd, results_wiener, results_combined

if __name__ == "__main__":
    print("Enhanced Satellite Image Denoising Demo")
    print("=" * 50)

    # Run demo
    demo_results = main_satellite_denoising_demo()

    print("\nTo use with your satellite data:")
    print("results = process_your_satellite_data(your_image_array)")
    print("best_result = results[2]['final_denoised']  # Combined approach")
```

```python
"""
GeoTIFF Satellite Image Denoising - FIXED FOR GEOTIFF FILES
Handles multi-band satellite imagery with proper libraries

Citations:
[1] GDAL/Rasterio: Gillies, S., et al. (2013-2024) "Rasterio: geospatial raster I/O for Python"
    https://github.com/rasterio/rasterio
[2] GeoTIFF format: Ritter, N., Ruth, M. (1997) "GeoTIFF Format Specification"
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import uniform_filter
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import warnings
warnings.filterwarnings('ignore')


def install_dependencies():
    """Install required libraries for GeoTIFF processing."""
    import subprocess
    import sys

    print("Installing required libraries...")
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install",
                              "rasterio", "-q"])
        print("Successfully installed rasterio")
    except:
        print("Could not install rasterio. Trying alternative method...")

# The call to install_dependencies() was outside the main execution block,
# which could cause issues if it's not run in the correct environment.
# Moving it inside __main__ or removing it if packages are already installed
# is a better practice. Since the prompt indicated this cell failed,
# I'll try to fix the syntax error first and assume dependencies are handled
# by previous cells or the environment setup.
# install_dependencies()


# =====================================================================
# STEP 1: LOAD GEOTIFF IMAGE (MULTIPLE METHODS)
# =====================================================================

def load_geotiff_rasterio(image_path):
    """
    Load GeoTIFF using rasterio (RECOMMENDED for satellite data).

    Citation: Rasterio is the standard library for geospatial raster data
    https://rasterio.readthedocs.io/
    """
    try:
        import rasterio

        print(f"Loading with rasterio...")
        with rasterio.open(image_path) as src:
            # Read all bands
            image = src.read()  # Shape: (bands, height, width)

            # Get metadata
            n_bands = src.count
            height = src.height
            width = src.width
            dtype = src.dtypes[0]

            print(f"Loaded GeoTIFF:")
            print(f"    - Bands: {n_bands}")
            print(f"    - Size: {width} x {height}")
            print(f"    - Data type: {dtype}")
            print(f"    - CRS: {src.crs}")

            # Transpose to (height, width, bands) for easier handling
            image = np.moveaxis(image, 0, -1)

            return image, True

    except ImportError:
        print("rasterio not available")
        return None, False
    except Exception as e:
        print(f"rasterio error: {e}")
        return None, False


def load_geotiff_gdal(image_path):
    """
    Load GeoTIFF using GDAL (fallback method).

    Citation: GDAL is the fundamental geospatial library
    https://gdal.org/
    """
    try:
        from osgeo import gdal

        print(f"  Loading with GDAL...")
        dataset = gdal.Open(image_path, gdal.GA_ReadOnly)

        if dataset is None:
            return None, False

        # Get dimensions
        n_bands = dataset.RasterCount
        width = dataset.RasterXSize
        height = dataset.RasterYSize

        print(f"Loaded GeoTIFF:")
        print(f"    - Bands: {n_bands}")
        print(f"    - Size: {width} x {height}")

        # Read all bands
        bands = []
        for i in range(1, n_bands + 1):
            band = dataset.GetRasterBand(i)
            bands.append(band.ReadAsArray())

        # Stack into (height, width, bands)
        image = np.stack(bands, axis=-1)

        dataset = None  # Close dataset
        return image, True

    except ImportError:
        print("GDAL not available")
        return None, False
    except Exception as e:
        print(f"GDAL error: {e}")
        return None, False


def load_geotiff_tifffile(image_path):
    """
    Load TIFF using tifffile library (works for some TIFFs).

    Citation: tifffile - Read and write TIFF files
    https://github.com/cgohlke/tifffile
    """
    try:
        import tifffile

        print(f"  Loading with tifffile...")
        image = tifffile.imread(image_path)

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        # Handle different organizations
        if len(image.shape) == 3:
            # Could be (bands, height, width) or (height, width, bands)
            if image.shape[0] < image.shape[2]:
                # Likely (bands, height, width) - transpose
                image = np.moveaxis(image, 0, -1)

        return image, True

    except Exception as e: # Add except block to catch potential errors
        print(f"tifffile error: {e}")
        return None, False


def load_geotiff_opencv(image_path):
    """
    Load TIFF using OpenCV (limited multi-band support).
    """
    try:
        import cv2

        print(f"  Loading with OpenCV...")
        image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)

        if image is None:
            return None, False

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        # Convert BGR to RGB if needed
        if len(image.shape) == 3 and image.shape[2] == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        return image, True

    except Exception as e: # Add except block
        print(f"OpenCV error: {e}")
        return None, False

def load_satellite_image(image_path):
    """
    Try multiple methods to load GeoTIFF satellite image.

    Priority order:
    1. rasterio (best for GeoTIFF)
    2. GDAL (robust fallback)
    3. tifffile (general TIFF)
    4. OpenCV (limited support)
    """
    print(f"\n[1] Loading satellite image from:")
    print(f"    {image_path}")

    # Method 1: Rasterio (RECOMMENDED)
    image, success = load_geotiff_rasterio(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 2: GDAL
    image, success = load_geotiff_gdal(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 3: tifffile
    image, success = load_geotiff_tifffile(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # Method 4: OpenCV
    image, success = load_geotiff_opencv(image_path)
    if success and image is not None: # Added check for image being None
        return image

    # All methods failed
    raise RuntimeError(
        "Could not load image with any available library.\n"
        "Please install rasterio: pip install rasterio"
    )


# =====================================================================
# HELPER FUNCTIONS
# =====================================================================

def normalize_to_uint8(image_array):
    """Normalize to 0-255 for display using percentile clipping."""
    if image_array.size == 0: # Handle empty array case
        return np.empty_like(image_array, dtype=np.uint8)

    p2, p98 = np.percentile(image_array, (2, 98))
    # Handle case where p2 and p98 are the same (flat image)
    if p2 == p98:
        return np.full_like(image_array, 127, dtype=np.uint8) # Return a mid-gray image

    image_clipped = np.clip(image_array, p2, p98)
    image_normalized = ((image_clipped - p2) / (p98 - p2) * 255)
    return np.clip(image_normalized, 0, 255).astype(np.uint8)


def select_rgb_bands(image_array, rgb_indices=None):
    """Extract RGB bands from multi-band image."""
    print(f"\n[2] Extracting RGB bands...")
    print(f"    Image shape: {image_array.shape}")

    if len(image_array.shape) == 2:
        print("    Grayscale detected - converting to RGB")
        return np.stack([image_array] * 3, axis=-1)

    elif len(image_array.shape) == 3 and image_array.shape[2] == 3:
        print("    RGB image detected")
        return image_array

    elif len(image_array.shape) == 3 and image_array.shape[2] > 3:
        # Multi-band - auto-detect or use specified indices
        if rgb_indices is None:
            # Try common satellite RGB combinations
            n_bands = image_array.shape[2]

            print(f"    Multi-band image: {n_bands} bands detected")
            print("    Trying common RGB band combinations...")

            # For Sentinel-2: RGB = bands 4,3,2 (0-indexed: 3,2,1)
            # For Landsat: RGB = bands 4,3,2 (0-indexed: 3,2,1)
            if n_bands >= 4:
                print("    Using bands [3,2,1] for RGB (Sentinel-2/Landsat style)")
                rgb_indices = (3, 2, 1)
            else:
                print("    Using first 3 bands [0,1,2]")
                rgb_indices = (0, 1, 2)

        # Ensure indices are within valid range
        max_band_index = image_array.shape[2] - 1
        valid_indices = [idx for idx in rgb_indices if 0 <= idx <= max_band_index]

        if len(valid_indices) != 3:
            print(f"Warning: Could not find 3 valid RGB bands from indices {rgb_indices}. Using first 3 available bands.")
            valid_indices = list(range(min(3, image_array.shape[2])))
            if len(valid_indices) < 3:
                 raise ValueError(f"Image only has {image_array.shape[2]} bands, cannot select 3 for RGB.")

        print(f"    Selected RGB indices: {valid_indices}")
        return image_array[:, :, valid_indices]


    else:
        raise ValueError(f"Unexpected image shape: {image_array.shape}")


def add_gaussian_noise(image, sigma=25.0):
    """Add Gaussian noise."""
    # Ensure image is float for noise addition
    image_float = image.astype(np.float64)
    noise = np.random.normal(0, sigma, image_float.shape)
    noisy_image = image_float + noise
    return noisy_image


def apply_mean_filter_multichannel(image, kernel_size=5):
    """Apply mean filter to each channel separately."""
    # Ensure image is float for filtering
    image_float = image.astype(np.float64)

    if len(image_float.shape) == 2:
        return uniform_filter(image_float, size=kernel_size)

    filtered_channels = []
    for i in range(image_float.shape[2]):
        filtered_channels.append(
            uniform_filter(image_float[:, :, i], size=kernel_size)
        )
    return np.stack(filtered_channels, axis=-1)


def calculate_metrics(original, processed, data_range=None):
    """Calculate PSNR and SSIM."""
    # Determine data range if not provided
    if data_range is None:
        # If original and processed have different min/max, use a common range like 0-255
        # or the full possible range based on dtype
        min_val = min(original.min(), processed.min())
        max_val = max(original.max(), processed.max())
        # Adjust data_range calculation based on actual value range
        # A common approach is to use the range of the original image, or 255 for 8-bit
        if original.dtype == np.uint8:
            data_range = 255.0
        elif original.dtype == np.float64 or original.dtype == np.float32:
             # Assume float images are normalized, e.g., to 0-1 or 0-max_sensor_value
             # If normalized to 0-1, data_range should be 1.0
             # If not normalized, use the actual range or a typical sensor bit depth range
             # For this case, let's assume the data is scaled to something where the max value is meaningful
             # A robust approach is to use the max value of the original image
            data_range = np.max(original) # Using max value of original as data_range

        # Fallback if data_range is still None (e.g., empty arrays)
        if data_range is None or data_range <= 0:
             data_range = 1.0 # Default to 1.0 for normalized float data


    # Ensure original and processed are float for metrics calculation
    original_float = original.astype(np.float64)
    processed_float = processed.astype(np.float64)

    # Clip processed to the range of original for metric calculation if they differ significantly
    # This prevents large deviations from skewing PSNR for images with different value ranges after processing
    processed_clipped = np.clip(processed_float, original_float.min(), original_float.max())


    try:
        psnr_value = psnr(original_float, processed_clipped, data_range=data_range)
    except Exception as e:
        print(f"Error calculating PSNR: {e}")
        psnr_value = np.nan # Return Not a Number on error


    try:
        if len(original_float.shape) == 3:
            # SSIM requires channel_axis for color images
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range,
                             channel_axis=2,
                             win_size=min(7, original_float.shape[0], original_float.shape[1])) # Adjust win_size
        else:
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range,
                             win_size=min(7, original_float.shape[0], original_float.shape[1])) # Adjust win_size
    except Exception as e:
        print(f"Error calculating SSIM: {e}")
        ssim_value = np.nan # Return Not a Number on error


    return psnr_value, ssim_value


# =====================================================================
# MAIN PROCESSING
# =====================================================================

def main():
    """Complete denoising pipeline."""

    print("="*70)
    print("SATELLITE IMAGE DENOISING - GEOTIFF COMPATIBLE")
    print("="*70)

    # YOUR IMAGE PATH
    # image_path = "/content/drive/MyDrive/Satellite Project with Debojyoti Banerjee/all_band_vskp.tif"
    # Using a local file path assuming the previous git clone cell has been run
    # Adjust the path to the actual location of your image file within the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"


    try:
        # Load image
        # Added check for file existence
        import os
        if not os.path.exists(image_path):
             raise FileNotFoundError(f"Image file not found at: {image_path}")


        image_array_raw = load_satellite_image(image_path)

        if image_array_raw is None:
             raise RuntimeError("Failed to load image.")

        # Extract RGB bands
        # Option: manually specify RGB bands if you know them
        # rgb_indices = (3, 2, 1)  # For Sentinel-2/Landsat
        # rgb_indices = (2, 1, 0) # For typical RGB (band 3=Red, 2=Green, 1=Blue) if available
        rgb_indices = None  # Auto-detect

        # Handle potential grayscale image loaded by load_satellite_image (e.g. from .bmp)
        if len(image_array_raw.shape) == 2:
             # If grayscale, convert to RGB for consistent processing
             image_array = np.stack([image_array_raw]*3, axis=-1)
             print("Grayscale image loaded, converted to RGB.")
        else:
             # Otherwise, select RGB bands from multi-band or use existing RGB
             image_array = select_rgb_bands(image_array_raw, rgb_indices)

        # Ensure image_array is float64 for calculations
        image_array = image_array.astype(np.float64)


        # Normalize data range before adding noise and calculating metrics
        # This is important for consistent noise levels and metric comparisons
        print(f"\n[3] Normalizing data range...")
        print(f"    Original range: [{image_array.min():.2f}, {image_array.max():.2f}]")

        # Normalize the original image data to a common range, e.g., 0-255 or 0-1
        # Normalizing to the original max value might be suitable for some sensor data
        # Let's normalize to 0-max_original_value for this example if max > 255
        max_original_value = np.max(image_array)
        if max_original_value > 0: # Avoid division by zero
             image_array_normalized = image_array / max_original_value * 255.0 # Scale to 0-255 range


        print(f"    Normalized range: [{image_array_normalized.min():.2f}, {image_array_normalized.max():.2f}]")


        # Add noise to the normalized image
        print(f"\n[4] Adding Gaussian noise...")
        noise_sigma = 40.0
        print(f"    Noise level: σ = {noise_sigma}")

        noisy_image = add_gaussian_noise(image_array_normalized, sigma=noise_sigma)

        # Calculate metrics (PSNR/SSIM) comparing normalized original to noisy image
        # Use the max value of the normalized original image as data_range for metrics
        psnr_noisy, ssim_noisy = calculate_metrics(image_array_normalized, noisy_image, data_range=255.0) # Assuming 0-255 range after normalization

        print(f"    Noisy - PSNR: {psnr_noisy:.2f} dB, SSIM: {ssim_noisy:.4f}")

        # Apply mean filter to the noisy image
        print(f"\n[5] Applying mean filter...")
        kernel_size = 5
        print(f"    Kernel size: {kernel_size}x{kernel_size}")

        filtered_image = apply_mean_filter_multichannel(noisy_image, kernel_size)

        # Calculate metrics comparing normalized original to filtered image
        psnr_filtered, ssim_filtered = calculate_metrics(image_array_normalized, filtered_image, data_range=255.0) # Assuming 0-255 range

        print(f"    Filtered - PSNR: {psnr_filtered:.2f} dB, SSIM: {ssim_filtered:.4f}")
        print(f"    Improvement: +{psnr_filtered - psnr_noisy:.2f} dB")

        # Visualize
        print(f"\n[6] Creating visualization...")

        fig, axes = plt.subplots(1, 3, figsize=(15, 5))

        # Normalize images to uint8 for display
        axes[0].imshow(normalize_to_uint8(image_array_normalized))
        axes[0].set_title('Original (Clean)', fontsize=12, fontweight='bold')
        axes[0].axis('off')

        axes[1].imshow(normalize_to_uint8(noisy_image))
        axes[1].set_title(f'Noisy (σ={noise_sigma})\nPSNR: {psnr_noisy:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[1].axis('off')

        axes[2].imshow(normalize_to_uint8(filtered_image))
        axes[2].set_title(f'Mean Filtered ({kernel_size}x{kernel_size})\n'
                         f'PSNR: {psnr_filtered:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[2].axis('off')

        plt.tight_layout()
        plt.savefig('satellite_denoising_results.png', dpi=150, bbox_inches='tight')
        print(f"    ✓ Saved: satellite_denoising_results.png")
        plt.show()

        # Summary
        print("\n" + "="*70)
        print("RESULTS SUMMARY")
        print("="*70)
        print(f"  Image shape: {image_array_normalized.shape}")
        print(f"  Noise level: σ = {noise_sigma}")
        print(f"\n  {'Method':<20} {'PSNR (dB)':<12} {'SSIM':<10}")
        print(f"  {'-'*45}")
        print(f"  {'Noisy':<20} {psnr_noisy:<12.2f} {ssim_noisy:<10.4f}")
        print(f"  {'Mean Filter':<20} {psnr_filtered:<12.2f} {ssim_filtered:<10.4f}")
        print(f"  {'Improvement':<20} {f'+{psnr_filtered-psnr_noisy:.2f}':<12} {f'+{ssim_filtered-ssim_noisy:.4f}':<10}")
        print("="*70)

        # Return the key images for potential further use in other cells
        return image_array_normalized, noisy_image, filtered_image


    except FileNotFoundError as e:
        print(f"Error: {e}")
        print("Please ensure the image path is correct.")
        return None, None, None # Return None on error
    except Exception as e:
        print(f"An error occurred during processing: {e}")
        import traceback
        traceback.print_exc() # Print traceback for debugging
        return None, None, None # Return None on error


# =====================================================================
# RUN
# =====================================================================

if __name__ == "__main__":
    # Check and install rasterio if needed (this was the cause of the SyntaxError)
    # The original error was a SyntaxError likely due to incorrect structure
    # around the __main__ block or incomplete except/finally blocks.
    # I've moved the dependency check inside __main__ and ensured proper structure.
    print("Checking dependencies...")
    try:
        import rasterio
        print("rasterio is installed")
    except ImportError:
        print("rasterio not found. Installing...")
        try:
            import subprocess
            import sys
            subprocess.check_call([sys.executable, "-m", "pip", "install", "rasterio", "-q"])
            print("rasterio installed")
        except Exception as e:
            print(f"Error installing rasterio: {e}")


    # Run main pipeline
    # Store the results in variables that can be used in other cells
    original_image_arr, noisy_image_arr, mean_filtered_image_arr = main()

    # You can now access these variables in subsequent cells
    # Example: print(original_image_arr.shape)
```

## Explore svd + wavelet denoising

### Subtask:
Explore svd + wavelet denoising by executing cell `oyemYZjfY-k8` which implements a hybrid SVD + Wavelet denoising method.


**Reasoning**:
The subtask is to explore SVD + Wavelet denoising by executing cell `oyemYZjfY-k8`. I will execute this cell to run the demonstration code which loads data, applies the SVD + Wavelet method, visualizes results, and calculates metrics.



```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            if file_path.lower().endswith(('.tif', '.tiff')):
                # For GeoTIFF files
                try:
                    from osgeo import gdal
                    dataset = gdal.Open(file_path)
                    data_array = dataset.ReadAsArray()

                    # Handle multi-band data
                    if len(data_array.shape) == 3:
                        data_array = np.mean(data_array, axis=0)  # Convert to grayscale

                    # Normalize to [0, 1]
                    noisy_data = (data_array - data_array.min()) / (data_array.max() - data_array.min())

                except ImportError:
                    print("GDAL not available. Using OpenCV...")
                    img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                    if img is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    noisy_data = img_as_float(img)

            elif file_path.lower().endswith('.npy'):
                noisy_data = np.load(file_path)

            else:
                # Standard image formats
                img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                if img is None:
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None
                noisy_data = img_as_float(img)

            print(f"Loaded satellite data from: {file_path}")
            print(f"Data shape: {noisy_data.shape}")

        self.original_data = noisy_data
        return noisy_data

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()

    # Example for processing your own data:
    # results = process_your_satellite_data(
    #     "path/to/your/satellite_data.tif",
    #     "path/to/output/denoised_data.npy"
    # )
```

```python
#Adaptive SVD Selection+Enhanced Wavelet Processing+Advanced Noise Handling
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal, ndimage, optimize
from scipy.fft import fft2, ifft2, fftshift, ifftshift
import pywt
from skimage import filters, restoration, measure, morphology, exposure
from sklearn.decomposition import TruncatedSVD
import cv2
import warnings
warnings.filterwarnings('ignore')

class EnhancedSatelliteDenoiser:
    def __init__(self):
        self.noise_profile = None
        self.optimal_components = None

    def estimate_noise_profile(self, image):
        """Advanced noise estimation for satellite imagery"""
        # Multi-scale noise estimation
        noise_estimates = []

        # Method 1: Robust median absolute deviation
        laplacian = np.array([[0, -1, 0], [-1, 4, -1], [0, -1, 0]])
        filtered = ndimage.convolve(image.astype(float), laplacian)
        noise_mad = np.median(np.abs(filtered - np.median(filtered))) / 0.6745
        noise_estimates.append(noise_mad)

        # Method 2: High-frequency analysis
        f_transform = fft2(image)
        h, w = f_transform.shape
        # Create a high-pass filter mask
        rows, cols = image.shape
        center_row, center_col = rows // 2, cols // 2
        mask = np.ones((rows, cols), dtype=np.uint8)
        # Define the low-frequency region to exclude
        r = min(rows, cols) // 8  # Adjust radius as needed
        cv2.circle(mask, (center_col, center_row), r, 0, -1)
        # Apply the mask in the frequency domain
        f_transform_shifted = fftshift(f_transform)
        high_freq_power = np.mean(np.abs(f_transform_shifted * mask)**2)
        noise_estimates.append(np.sqrt(high_freq_power) * 0.1)


        # Method 3: Wavelet-based estimation
        coeffs = pywt.dwt2(image, 'db4')
        _, (cH, cV, cD) = coeffs
        noise_wavelet = np.median(np.abs(cD)) / 0.6745
        noise_estimates.append(noise_wavelet)

        # Robust average
        self.noise_profile = np.median(noise_estimates)
        return self.noise_profile

    def adaptive_svd_denoising(self, image, max_components=None):
        """Improved SVD denoising with automatic rank selection"""
        # Reshape image for SVD
        original_shape = image.shape
        if len(original_shape) == 2:
            # For 2D image, create patches or use direct SVD
            h, w = original_shape

            # Method 1: Block-wise SVD for large images
            if h > 512 or w > 512:
                return self._block_svd_denoise(image, max_components)

            # Method 2: Direct SVD for smaller images
            U, s, Vt = np.linalg.svd(image, full_matrices=False)

            # Automatic rank selection using noise profile
            if max_components is None:
                noise_level = self.estimate_noise_profile(image)
                # Keep components above noise threshold
                signal_threshold = noise_level * np.sqrt(min(h, w))
                optimal_rank = np.sum(s > signal_threshold)
                optimal_rank = max(optimal_rank, min(h, w) // 10)  # At least 10% of components
                optimal_rank = min(optimal_rank, min(h, w) // 2)   # At most 50% of components
            else:
                optimal_rank = min(max_components, len(s))

            self.optimal_components = optimal_rank

            # Reconstruct with optimal rank
            s_filtered = s.copy()
            s_filtered[optimal_rank:] = 0
            denoised = U @ np.diag(s_filtered) @ Vt

            return denoised, optimal_rank, s

        return image, 0, None

    def _block_svd_denoise(self, image, max_components=None, block_size=256):
        """Block-wise SVD for large images"""
        h, w = image.shape
        denoised = np.zeros_like(image)
        overlap = block_size // 4

        for i in range(0, h - block_size + 1, block_size - overlap):
            for j in range(0, w - block_size + 1, block_size - overlap):
                # Extract block
                end_i = min(i + block_size, h)
                end_j = min(j + block_size, w)
                block = image[i:end_i, j:end_j]

                # Apply SVD to block
                U, s, Vt = np.linalg.svd(block, full_matrices=False)

                # Determine optimal rank for this block
                if max_components is None:
                    noise_level = np.median(np.abs(block.flatten() - np.median(block.flatten()))) / 0.6745
                    signal_threshold = noise_level * np.sqrt(min(block.shape))
                    optimal_rank = np.sum(s > signal_threshold)
                    optimal_rank = max(optimal_rank, len(s) // 10)
                    optimal_rank = min(optimal_rank, len(s) // 2)
                else:
                    optimal_rank = min(max_components, len(s))

                # Reconstruct block
                s_filtered = s.copy()
                s_filtered[optimal_rank:] = 0
                block_denoised = U @ np.diag(s_filtered) @ Vt

                # Blend with existing result (for overlapping regions)
                if i == 0 and j == 0:
                    denoised[i:end_i, j:end_j] = block_denoised
                else:
                    # Simple averaging for overlaps
                    mask = denoised[i:end_i, j:end_j] != 0
                    denoised[i:end_i, j:end_j] = np.where(mask,
                                                         (denoised[i:end_i, j:end_j] + block_denoised) / 2,
                                                         block_denoised)

        return denoised, optimal_rank, None

    def enhanced_wiener_filter(self, image, noise_var=None, psf=None):
        """Enhanced Wiener filter with PSF estimation"""
        if noise_var is None:
            noise_var = (self.estimate_noise_profile(image)) ** 2

        # If no PSF provided, estimate motion blur or use identity
        if psf is None:
            psf = np.array([[1]])  # Identity PSF (no blur)

        # Apply Wiener deconvolution
        if psf.size > 1:
            denoised = restoration.wiener(image, psf, noise_var, clip=False)
        else:
            # Simple Wiener filtering in frequency domain
            image_fft = fft2(image)
            power_spectrum = np.abs(image_fft) ** 2
            signal_power = np.maximum(power_spectrum - noise_var, 0.1 * power_spectrum)
            wiener_filter = signal_power / (power_spectrum + noise_var) # Corrected denominator
            filtered_fft = image_fft * wiener_filter
            denoised = np.real(ifft2(filtered_fft))

        return denoised

    def adaptive_wavelet_denoising(self, image, wavelet='auto', levels=None):
        """Enhanced wavelet denoising with adaptive thresholding"""
        # Determine optimal decomposition level
        if levels is None:
            min_dim = min(image.shape)
            levels = int(np.log2(min_dim)) - 2
            levels = max(1, min(levels, 6))

        # Multi-wavelet approach: try different wavelets and combine
        wavelets_to_try = ['db4', 'db8', 'coif2', 'bior2.2'] if wavelet == 'auto' else [wavelet]
        denoised_results = []

        for wvlt in wavelets_to_try:
            try:
                # Decompose
                coeffs = pywt.wavedec2(image, wvlt, level=levels)

                # Adaptive thresholding for each level
                coeffs_thresh = [coeffs[0]]  # Keep approximation coefficients

                for i in range(1, len(coeffs)):
                    detail_coeffs = coeffs[i]
                    thresh_coeffs = []

                    for detail in detail_coeffs:
                        # Level-specific threshold
                        sigma = np.median(np.abs(detail)) / 0.6745

                        # Bayes shrink threshold
                        if sigma > 0:
                            threshold = sigma**2 / np.sqrt(np.maximum(np.var(detail) - sigma**2, 0) + 1e-10)
                        else:
                            threshold = np.max(np.abs(detail)) # Use max value as threshold if sigma is zero

                        # Apply soft thresholding
                        thresh_detail = pywt.threshold(detail, threshold, mode='soft')
                        thresh_coeffs.append(thresh_detail)

                    coeffs_thresh.append(tuple(thresh_coeffs))

                # Reconstruct
                denoised = pywt.waverec2(coeffs_thresh, wvlt)

                # Ensure proper size
                if denoised.shape != image.shape:
                    denoised = denoised[:image.shape[0], :image.shape[1]]

                denoised_results.append(denoised)

            except Exception as e:
                print(f"Warning: Wavelet {wvlt} failed: {e}")
                continue

        if not denoised_results:
            return image

        # If multiple wavelets used, combine results
        if len(denoised_results) == 1:
            return denoised_results[0]
        else:
            # Weighted combination based on local variance
            final_result = np.zeros_like(image)
            total_weight = np.zeros_like(image)

            for result in denoised_results:
                # Local variance as weight (higher variance = more detail preserved)
                local_var = ndimage.generic_filter(result, np.var, size=5)
                weight = local_var / (np.sum([ndimage.generic_filter(r, np.var, size=5)
                                            for r in denoised_results], axis=0) + 1e-10)
                final_result += weight * result
                total_weight += weight

            # Normalize and clip to [0, 1]
            combined_result = final_result / (total_weight + 1e-10)
            return np.clip(combined_result, 0, 1)


    def edge_preserving_filter(self, image, iterations=1):
        """Apply edge-preserving smoothing to reduce remaining noise"""
        # Bilateral filter for edge preservation
        # Ensure image is in the correct format and range for cv2.bilateralFilter
        if image.dtype != np.uint8:
            # Scale to 0-255 for cv2.bilateralFilter and convert to uint8
            image_norm = exposure.rescale_intensity(image, out_range=(0, 255)).astype(np.uint8)
            filtered = cv2.bilateralFilter(image_norm, 9, 75, 75)
            # Scale back to the original range of the input float image
            # Assuming the input float image was in the range [0, 1]
            filtered = exposure.rescale_intensity(filtered.astype(float), out_range=(0.0, 1.0))
        else:
             # If already uint8, just apply the filter
             filtered = cv2.bilateralFilter(image, 9, 75, 75)
             # Convert to float and scale to [0, 1] if the original was float
             filtered = filtered.astype(float) / 255.0


        return filtered


    def enhance_contrast(self, image, method='adaptive'):
        """Post-processing contrast enhancement"""
        # Ensure image is in the correct format and range for scikit-image exposure functions
        if image.dtype != np.float64 or np.min(image) < 0 or np.max(image) > 1:
           image = img_as_float(image) # Convert to float and scale to [0, 1]

        if method == 'adaptive':
            # Adaptive histogram equalization
            return exposure.equalize_adapthist(image, clip_limit=0.01)
        elif method == 'gamma':
            # Gamma correction
            return exposure.adjust_gamma(image, gamma=0.8)
        elif method == 'sigmoid':
            # Sigmoid correction
            return exposure.adjust_sigmoid(image, cutoff=0.5, gain=10)
        else:
            return image

    def comprehensive_denoise(self, image, method='auto', enhance_contrast=True):
        """
        Comprehensive denoising pipeline

        Parameters:
        - image: input noisy image
        - method: 'svd_wavelet', 'wiener_wavelet', 'all', or 'auto'
        - enhance_contrast: apply post-processing contrast enhancement
        """
        print("Starting comprehensive satellite image denoising...")

        # Ensure float type and scale to [0, 1]
        if image.dtype != np.float64:
            image = img_as_float(image) # Convert to float and scale to [0, 1]

        # Store original for comparison
        original = image.copy()

        # Step 1: Initial noise estimation
        noise_level = self.estimate_noise_profile(image)
        print(f"Estimated noise level: {noise_level:.4f}")

        results = {'original': original}

        if method in ['svd_wavelet', 'all', 'auto']:
            print("Applying SVD denoising...")
            svd_denoised, optimal_rank, singular_values = self.adaptive_svd_denoising(image)
            print(f"Optimal SVD rank: {optimal_rank}")

            print("Applying wavelet denoising to SVD result...")
            svd_wavelet_denoised = self.adaptive_wavelet_denoising(svd_denoised, wavelet='auto')

            results['svd_denoised'] = svd_denoised
            results['svd_wavelet_denoised'] = svd_wavelet_denoised
            results['optimal_rank'] = optimal_rank

        if method in ['wiener_wavelet', 'all', 'auto']:
            print("Applying Wiener filter...")
            wiener_denoised = self.enhanced_wiener_filter(image, noise_var=noise_level**2)

            print("Applying wavelet denoising to Wiener result...")
            wiener_wavelet_denoised = self.adaptive_wavelet_denoising(wiener_denoised, wavelet='auto')

            results['wiener_denoised'] = wiener_denoised
            results['wiener_wavelet_denoised'] = wiener_wavelet_denoised

        # Choose best result based on method
        if method == 'svd_wavelet':
            final_denoised = results['svd_wavelet_denoised']
        elif method == 'wiener_wavelet':
            final_denoised = results['wiener_wavelet_denoised']
        elif method in ['all', 'auto']:
            # Combine both methods using local quality metrics
            if 'svd_wavelet_denoised' in results and 'wiener_wavelet_denoised' in results:
                svd_result = results['svd_wavelet_denoised']
                wiener_result = results['wiener_wavelet_denoised']

                # Local gradient preservation as quality metric
                original_grad = np.sqrt(np.gradient(original)[0]**2 + np.gradient(original)[1]**2)
                svd_grad = np.sqrt(np.gradient(svd_result)[0]**2 + np.gradient(svd_result)[1]**2)
                wiener_grad = np.sqrt(np.gradient(wiener_result)[0]**2 + np.gradient(wiener_result)[1]**2)

                svd_quality = 1 / (1 + np.abs(original_grad - svd_grad))
                wiener_quality = 1 / (1 + np.abs(original_grad - wiener_grad))

                # Smooth the quality maps
                svd_quality = ndimage.gaussian_filter(svd_quality, sigma=2)
                wiener_quality = ndimage.gaussian_filter(wiener_quality, sigma=2)

                # Normalize weights
                total_quality = svd_quality + wiener_quality + 1e-10
                svd_weight = svd_quality / total_quality
                wiener_weight = wiener_quality / total_quality

                final_denoised = svd_weight * svd_result + wiener_weight * wiener_result
            else:
                final_denoised = results.get('svd_wavelet_denoised',
                                           results.get('wiener_wavelet_denoised', image))

        # Ensure final denoised image is clipped to [0, 1] before post-processing
        final_denoised = np.clip(final_denoised, 0, 1)

        # Post-processing
        print("Applying edge-preserving filter...")
        final_denoised = self.edge_preserving_filter(final_denoised)

        if enhance_contrast:
            print("Enhancing contrast...")
            final_denoised = self.enhance_contrast(final_denoised, method='adaptive')

        results['final_denoised'] = final_denoised
        results['noise_level'] = noise_level

        # Calculate improvement metrics
        if np.std(original) > 0:  # Avoid division by zero
            # Calculate PSNR between original and denoised
            mse = np.mean((original - final_denoised)**2)
            if mse > 0:
                psnr = 10 * np.log10(1.0 / mse) # Assuming image range [0, 1]
                results['psnr'] = psnr
                print(f"PSNR: {psnr:.2f} dB")
            else:
                results['psnr'] = float('inf')
                print("PSNR: Inf dB (Perfect reconstruction)")


        print("Denoising complete!")
        return results

    def visualize_results(self, results):
        """Comprehensive visualization of denoising results"""
        # Adjust subplot layout based on available results, excluding metrics
        display_keys = [key for key in results.keys() if key not in ['noise_level', 'psnr', 'optimal_rank']]
        n_results = len(display_keys)
        cols = min(3, n_results)
        rows = (n_results + cols - 1) // cols

        if n_results == 0:
            print("No images to display in results.")
            return

        fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))

        # Handle cases where axes might be a single object or a 1D array
        if n_results == 1:
            axes = np.array([[axes]])
        elif rows == 1:
            axes = axes.reshape(1, -1)
        elif cols == 1:
             axes = axes.reshape(-1, 1)


        plot_idx = 0
        for key in display_keys:
            image = results[key]
            if isinstance(image, np.ndarray) and image.ndim >= 2: # Ensure it's an image array
                row = plot_idx // cols
                col = plot_idx % cols

                # Determine appropriate color map and range for display
                cmap = 'gray' if image.ndim == 2 else None # Use gray for 2D, default for color
                vmin = np.min(image)
                vmax = np.max(image)
                # If the image is likely a difference map, adjust cmap and range
                if 'diff' in key.lower() or 'noise' in key.lower() or 'error' in key.lower():
                    cmap = 'hot'
                    vmin = 0 # Difference/Noise maps usually start from 0
                    # vmax could be set to a small value to highlight subtle differences
                    # Let's use a percentile or a fixed small value
                    vmax = np.percentile(image, 99) if image.size > 0 and np.percentile(image, 99) > 1e-6 else 0.1 # Adjust vmax

                im = axes[row, col].imshow(image, cmap=cmap, vmin=vmin, vmax=vmax)
                axes[row, col].set_title(key.replace('_', ' ').title())
                axes[row, col].axis('off')
                # Add colorbar only if it's a grayscale or difference map
                if cmap == 'gray' or cmap == 'hot':
                    plt.colorbar(im, ax=axes[row, col], shrink=0.8)

                plot_idx += 1

        # Hide empty subplots
        for i in range(plot_idx, rows * cols):
            row = i // cols
            col = i % cols
            axes[row, col].axis('off')


        plt.tight_layout()
        plt.show()

def main_satellite_denoising_demo():
    """Demo with synthetic satellite-like imagery"""
    # Create synthetic satellite image with various features
    size = 256
    x, y = np.meshgrid(np.linspace(0, 1, size), np.linspace(0, 1, size))

    # Simulate different terrain features
    fields = 0.5 + 0.3 * np.sin(20 * x) * np.cos(20 * y)  # Agricultural fields
    forest = 0.3 * np.exp(-((x-0.2)**2 + (y-0.8)**2) / 0.1)  # Forest area
    urban = 0.8 * ((x > 0.6) & (x < 0.9) & (y > 0.1) & (y < 0.4))  # Urban area
    water = 0.1 * ((x-0.8)**2 + (y-0.7)**2 < 0.05)  # Water body

    clean_image = fields + forest + urban + water

    # Add realistic noise (sensor noise + atmospheric effects)
    noise_level = 0.15
    gaussian_noise = noise_level * np.random.randn(size, size)
    salt_pepper = np.random.random((size, size))
    clean_image[salt_pepper < 0.01] = 0  # Salt noise
    clean_image[salt_pepper > 0.99] = 1  # Pepper noise

    noisy_image = clean_image + gaussian_noise
    noisy_image = np.clip(noisy_image, 0, 1)

    # Apply denoising
    denoiser = EnhancedSatelliteDenoiser()
    # Run the demo with the 'auto' method which includes Wiener + Wavelet
    results = denoiser.comprehensive_denoise(noisy_image, method='auto')

    # Visualize
    denoiser.visualize_results(results)

    return results

# Usage example for your satellite data
def process_your_satellite_data(image_array):
    """
    Process your actual satellite image

    Parameters:
    - image_array: numpy array of your satellite image
    """
    denoiser = EnhancedSatelliteDenoiser()

    # Method 1: SVD + Wavelet (similar to your current approach)
    results_svd = denoiser.comprehensive_denoise(image_array, method='svd_wavelet')

    # Method 2: Wiener + Wavelet (alternative approach)
    results_wiener = denoiser.comprehensive_denoise(image_array, method='wiener_wavelet')

    # Method 3: Combined approach (recommended)
    results_combined = denoiser.comprehensive_denoise(image_array, method='auto')

    return results_svd, results_wiener, results_combined

if __name__ == "__main__":
    print("Enhanced Satellite Image Denoising Demo")
    print("=" * 50)

    # Run demo which includes Wiener + Wavelet
    demo_results = main_satellite_denoising_demo()

    print("\nTo use with your satellite data:")
    print("results = process_your_satellite_data(your_image_array)")
    print("best_result = results[2]['final_denoised']  # Combined approach")
```

```python
#Weiner Filter + Wavelet Threshold for Satellite Data
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal, ndimage
from scipy.fft import fft2, ifft2, fftshift, ifftshift
import pywt
from skimage import data, filters
from sklearn.datasets import make_classification
import warnings
warnings.filterwarnings('ignore')

class WienerWaveletDenoiser:
    def __init__(self, file_path="/content/Sat_Image_Denoising/2_A.bmp"):
        self.file_path = file_path
        self.wiener_noise_var = None
        self.signal_var = None

    def estimate_noise_variance(self, data):
        """Estimate noise variance using robust median absolute deviation"""
        if data.ndim == 1:
            # For 1D signals, use high-pass filtering
            diff = np.diff(data)
            noise_var = np.median(np.abs(diff)) / 0.6745
        else:
            # For 2D images, use Laplacian method
            laplacian = np.array([[0, -1, 0], [-1, 4, -1], [0, -1, 0]])
            filtered = ndimage.convolve(data, laplacian)
            noise_var = np.var(filtered) * 0.5
        return noise_var ** 2

    def wiener_filter_1d(self, signal_data, noise_var=None):
        """Apply Wiener filter to 1D signal"""
        if noise_var is None:
            noise_var = self.estimate_noise_variance(signal_data)

        # FFT of the signal
        signal_fft = np.fft.fft(signal_data)
        power_spectrum = np.abs(signal_fft) ** 2

        # Estimate signal power spectrum (assuming signal >> noise in low frequencies)
        signal_power = np.maximum(power_spectrum - noise_var, 0.1 * power_spectrum)

        # Wiener filter transfer function
        wiener_filter = signal_power / (signal_power + noise_var)

        # Apply filter
        filtered_fft = signal_fft * wiener_filter
        filtered_signal = np.real(np.fft.ifft(filtered_fft))

        return filtered_signal, wiener_filter

    def wiener_filter_2d(self, image, noise_var=None):
        """Apply Wiener filter to 2D image"""
        if noise_var is None:
            noise_var = self.estimate_noise_variance(image)

        # FFT of the image
        image_fft = fft2(image)
        power_spectrum = np.abs(image_fft) ** 2

        # Estimate signal power spectrum
        signal_power = np.maximum(power_spectrum - noise_var, 0.1 * power_spectrum)

        # Wiener filter transfer function
        wiener_filter = signal_power / (power_spectrum + noise_var)

        # Apply filter
        filtered_fft = image_fft * wiener_filter
        filtered_image = np.real(ifft2(filtered_fft))

        return filtered_image, wiener_filter

    def wavelet_threshold_denoise(self, data, wavelet='db8', mode='soft', method='bayes'):
        """
        Apply wavelet denoising with various thresholding methods

        Parameters:
        - data: input signal/image
        - wavelet: wavelet type ('db8', 'haar', 'coif2', etc.)
        - mode: thresholding mode ('soft' or 'hard')
        - method: threshold estimation method ('bayes', 'sure', 'minimax')
        """

        if data.ndim == 1:
            return self._wavelet_denoise_1d(data, wavelet, mode, method)
        else:
            return self._wavelet_denoise_2d(data, wavelet, mode, method)

    def _wavelet_denoise_1d(self, signal_data, wavelet, mode, method):
        """Wavelet denoising for 1D signals"""
        # Decompose
        coeffs = pywt.wavedec(signal_data, wavelet, level=None)

        # Estimate noise level from finest detail coefficients
        sigma = np.median(np.abs(coeffs[-1])) / 0.6745

        # Calculate threshold based on method
        N = len(signal_data)
        if method == 'bayes':
            threshold = sigma * np.sqrt(2 * np.log(N))
        elif method == 'sure':
            threshold = sigma * np.sqrt(2 * np.log(N))  # Simplified SURE
        elif method == 'minimax':
            threshold = sigma * (0.3936 + 0.1829 * np.log2(N))
        else:
            threshold = sigma * np.sqrt(2 * np.log(N))

        # Apply thresholding to detail coefficients
        coeffs_thresh = list(coeffs)
        coeffs_thresh[1:] = [pywt.threshold(detail, threshold, mode=mode)
                            for detail in coeffs[1:]]

        # Reconstruct
        denoised = pywt.waverec(coeffs_thresh, wavelet)

        return denoised, threshold, coeffs, coeffs_thresh

    def _wavelet_denoise_2d(self, image, wavelet, mode, method):
        """Wavelet denoising for 2D images"""
        # Decompose
        coeffs = pywt.wavedec2(image, wavelet, level=None)

        # Estimate noise level from finest detail coefficients
        sigma = np.median(np.abs(coeffs[-1][-1])) / 0.6745

        # Calculate threshold
        N = image.size
        if method == 'bayes':
            threshold = sigma * np.sqrt(2 * np.log(N))
        elif method == 'sure':
            threshold = sigma * np.sqrt(2 * np.log(N))
        elif method == 'minimax':
            threshold = sigma * (0.3936 + 0.1829 * np.log2(N))
        else:
            threshold = sigma * np.sqrt(2 * np.log(N))

        # Apply thresholding to detail coefficients
        coeffs_thresh = list(coeffs)
        coeffs_thresh[0] = coeffs[0]  # Keep approximation coefficients
        for i in range(1, len(coeffs)):
            coeffs_thresh[i] = tuple([pywt.threshold(detail, threshold, mode=mode)
                                    for detail in coeffs[i]])

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresh, wavelet)

        return denoised, threshold, coeffs, coeffs_thresh

    def combined_denoise(self, data, wavelet='db8', threshold_mode='soft',
                        threshold_method='bayes', noise_var=None):
        """
        Combined Wiener filter + Wavelet denoising
        """
        print("Applying Wiener filter...")

        if data.ndim == 1:
            wiener_filtered, wiener_tf = self.wiener_filter_1d(data, noise_var)
        else:
            wiener_filtered, wiener_tf = self.wiener_filter_2d(data, noise_var)

        print("Applying wavelet denoising...")

        if data.ndim == 1:
            final_denoised, threshold, coeffs, coeffs_thresh = \
                self._wavelet_denoise_1d(wiener_filtered, wavelet, threshold_mode, threshold_method)
        else:
            final_denoised, threshold, coeffs, coeffs_thresh = \
                self._wavelet_denoise_2d(wiener_filtered, wavelet, threshold_mode, threshold_method)

        return {
            'original': data,
            'wiener_filtered': wiener_filtered,
            'final_denoised': final_denoised,
            'wiener_transfer_function': wiener_tf,
            'wavelet_threshold': threshold,
            'wavelet_coeffs_original': coeffs,
            'wavelet_coeffs_thresholded': coeffs_thresh
        }

# Example usage and demonstrations
def create_noisy_signal_1d(length=1000, noise_level=0.3):
    """Create a noisy 1D test signal"""
    t = np.linspace(0, 1, length)
    # Clean signal: combination of sine waves
    clean = (np.sin(2 * np.pi * 5 * t) +
             0.5 * np.sin(2 * np.pi * 20 * t) +
             0.3 * np.sin(2 * np.pi * 50 * t))
    # Add noise
    noise = noise_level * np.random.randn(length)
    noisy = clean + noise
    return t, clean, noisy

def create_noisy_image_2d(size=256, noise_level=0.2):
    """Create a noisy 2D test image"""
    # Create synthetic image with geometric patterns
    x, y = np.meshgrid(np.linspace(-1, 1, size), np.linspace(-1, 1, size))
    clean = (np.exp(-(x**2 + y**2)/0.3) +
             0.5 * np.sin(10 * x) * np.cos(10 * y) +
             0.3 * (x**2 + y**2 < 0.5))
    # Add noise
    noise = noise_level * np.random.randn(size, size)
    noisy = clean + noise
    return clean, noisy

def demonstrate_1d_denoising():
    """Demonstrate 1D signal denoising"""
    print("=== 1D Signal Denoising Demo ===")

    # Create test signal
    t, clean_signal, noisy_signal = create_noisy_signal_1d()

    # Apply denoising
    denoiser = WienerWaveletDenoiser()
    results = denoiser.combined_denoise(noisy_signal, wavelet='db8',
                                       threshold_mode='soft', threshold_method='bayes')

    # Calculate metrics
    mse_original = np.mean((noisy_signal - clean_signal)**2)
    mse_wiener = np.mean((results['wiener_filtered'] - clean_signal)**2)
    mse_final = np.mean((results['final_denoised'] - clean_signal)**2)

    print(f"MSE - Original: {mse_original:.6f}")
    print(f"MSE - After Wiener: {mse_wiener:.6f}")
    print(f"MSE - After Wavelet: {mse_final:.6f}")
    print(f"Improvement: {(mse_original/mse_final):.2f}x")

    # Plot results
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))

    # Time domain signals
    axes[0,0].plot(t, clean_signal, 'g-', label='Clean', linewidth=2)
    axes[0,0].plot(t, noisy_signal, 'r-', alpha=0.7, label='Noisy')
    axes[0,0].set_title('Original Signals')
    axes[0,0].legend()
    axes[0,0].grid(True)

    axes[0,1].plot(t, clean_signal, 'g-', label='Clean', linewidth=2)
    axes[0,1].plot(t, results['wiener_filtered'], 'b-', alpha=0.8, label='Wiener Filtered')
    axes[0,1].plot(t, results['final_denoised'], 'm-', alpha=0.8, label='Final Denoised')
    axes[0,1].set_title('Denoising Results')
    axes[0,1].legend()
    axes[0,1].grid(True)

    # Frequency domain
    freqs = np.fft.fftfreq(len(noisy_signal), t[1]-t[0])
    axes[1,0].semilogy(freqs[:len(freqs)//2],
                       np.abs(np.fft.fft(noisy_signal))[:len(freqs)//2],
                       'r-', alpha=0.7, label='Noisy')
    axes[1,0].semilogy(freqs[:len(freqs)//2],
                       np.abs(np.fft.fft(results['final_denoised']))[:len(freqs)//2],
                       'm-', label='Denoised')
    axes[1,0].set_title('Frequency Spectrum')
    axes[1,0].set_xlabel('Frequency (Hz)')
    axes[1,0].legend()
    axes[1,0].grid(True)

    # Wiener filter transfer function
    axes[1,1].plot(results['wiener_transfer_function'])
    axes[1,1].set_title('Wiener Filter Transfer Function')
    axes[1,1].set_xlabel('Frequency Bin')
    axes[1,1].set_ylabel('Filter Response')
    axes[1,1].grid(True)

    plt.tight_layout()
    plt.show()

    return results

def demonstrate_2d_denoising():
    """Demonstrate 2D image denoising"""
    print("\n=== 2D Image Denoising Demo ===")

    # Create test image
    clean_image, noisy_image = create_noisy_image_2d(size=128)

    # Apply denoising
    denoiser = WienerWaveletDenoiser()
    results = denoiser.combined_denoise(noisy_image, wavelet='db4',
                                       threshold_mode='soft', threshold_method='bayes')

    # Calculate metrics
    mse_original = np.mean((noisy_image - clean_image)**2)
    mse_wiener = np.mean((results['wiener_filtered'] - clean_image)**2)
    mse_final = np.mean((results['final_denoised'] - clean_image)**2)

    print(f"MSE - Original: {mse_original:.6f}")
    print(f"MSE - After Wiener: {mse_wiener:.6f}")
    print(f"MSE - After Wavelet: {mse_final:.6f}")
    print(f"Improvement: {(mse_original/mse_final):.2f}x")

    # Plot results
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))

    im1 = axes[0,0].imshow(clean_image, cmap='gray')
    axes[0,0].set_title('Clean Image')
    axes[0,0].axis('off')
    plt.colorbar(im1, ax=axes[0,0], shrink=0.8)

    im2 = axes[0,1].imshow(noisy_image, cmap='gray')
    axes[0,1].set_title('Noisy Image')
    axes[0,1].axis('off')
    plt.colorbar(im2, ax=axes[0,1], shrink=0.8)

    im3 = axes[0,2].imshow(results['wiener_filtered'], cmap='gray')
    axes[0,2].set_title('Wiener Filtered')
    axes[0,2].axis('off')
    plt.colorbar(im3, ax=axes[0,2], shrink=0.8)

    im4 = axes[1,0].imshow(results['final_denoised'], cmap='gray')
    axes[1,0].set_title('Final Denoised')
    axes[1,0].axis('off')
    plt.colorbar(im4, ax=axes[1,0], shrink=0.8)

    # Show difference images
    diff_wiener = np.abs(results['wiener_filtered'] - clean_image)
    diff_final = np.abs(results['final_denoised'] - clean_image)

    im5 = axes[1,1].imshow(diff_wiener, cmap='hot')
    axes[1,1].set_title('Wiener Error')
    axes[1,1].axis('off')
    plt.colorbar(im5, ax=axes[1,1], shrink=0.8)

    im6 = axes[1,2].imshow(diff_final, cmap='hot')
    axes[1,2].set_title('Final Error')
    axes[1,2].axis('off')
    plt.colorbar(im6, ax=axes[1,2], shrink=0.8)

    plt.tight_layout()
    plt.show()

    return results

def compare_wavelets_and_methods():
    """Compare different wavelets and thresholding methods"""
    print("\n=== Wavelet and Method Comparison ===")

    # Create test signal
    t, clean_signal, noisy_signal = create_noisy_signal_1d(length=512)

    wavelets = ['db4', 'db8', 'haar', 'coif2', 'bior2.2']
    methods = ['bayes', 'sure', 'minimax']
    modes = ['soft', 'hard']

    denoiser = WienerWaveletDenoiser()
    results_comparison = {}

    print("\nMSE Comparison:")
    print("Wavelet\tMethod\tMode\tMSE")
    print("-" * 40)

    for wavelet in wavelets:
        for method in methods:
            for mode in modes:
                try:
                    result = denoiser.combined_denoise(noisy_signal,
                                                     wavelet=wavelet,
                                                     threshold_mode=mode,
                                                     threshold_method=method)
                    mse = np.mean((result['final_denoised'] - clean_signal)**2)
                    results_comparison[f"{wavelet}_{method}_{mode}"] = mse
                    print(f"{wavelet}\t{method}\t{mode}\t{mse:.6f}")
                except Exception as e: # Catch exceptions and print failure message
                    print(f"{wavelet}\t{method}\t{mode}\tFailed ({e})")


    # Find best combination
    # Filter out failed attempts (where MSE is not available)
    successful_results = {k: v for k, v in results_comparison.items() if isinstance(v, (int, float))}

    if successful_results:
        best_combo = min(successful_results.items(), key=lambda x: x[1])
        print(f"\nBest combination: {best_combo[0]} with MSE: {best_combo[1]:.6f}")
    else:
        print("\nNo successful denoising combinations found for comparison.")


    return results_comparison

# Main execution
if __name__ == "__main__":
    print("Starting Wiener Filter + Wavelet Denoising Demonstrations...")

    # Run demonstrations
    # 1D signal denoising
    results_1d = demonstrate_1d_denoising()

    # 2D image denoising
    results_2d = demonstrate_2d_denoising()

    # Compare different wavelets and methods (on 1D signal)
    comparison = compare_wavelets_and_methods()


    print("\n=== Usage Examples ===")
    print("""
    # Basic usage:
    denoiser = WienerWaveletDenoiser()

    # For 1D signals:
    results = denoiser.combined_denoise(your_signal, wavelet='db8',
                                       threshold_mode='soft', threshold_method='bayes')

    # For 2D images:
    results = denoiser.combined_denoise(your_image, wavelet='db4',
                                       threshold_mode='soft', threshold_method='bayes')

    # Access results:
    denoised_data = results['final_denoised']
    wiener_filtered = results['wiener_filtered']

    # Different wavelets to try: 'db4', 'db8', 'haar', 'coif2', 'bior2.2'
    # Threshold methods: 'bayes', 'sure', 'minimax'
    # Threshold modes: 'soft', 'hard'
    """)
```

## Explore enhanced satellite denoising

### Subtask:
Explore enhanced satellite denoising by executing cell `pc2xZ2kdVLJP` which provides a more advanced denoising pipeline including adaptive SVD and enhanced wavelet processing. This method is designed to minimize changes in resolution.


**Reasoning**:
Execute the provided code cell to explore the enhanced satellite denoising method, which includes adaptive SVD and enhanced wavelet processing, and observe its performance on a synthetic image.



```python
#Adaptive SVD Selection+Enhanced Wavelet Processing+Advanced Noise Handling
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal, ndimage, optimize
from scipy.fft import fft2, ifft2, fftshift, ifftshift
import pywt
from skimage import filters, restoration, measure, morphology, exposure, img_as_float, data
from sklearn.decomposition import TruncatedSVD
import cv2
import warnings
warnings.filterwarnings('ignore')

class EnhancedSatelliteDenoiser:
    def __init__(self):
        self.noise_profile = None
        self.optimal_components = None

    def estimate_noise_profile(self, image):
        """Advanced noise estimation for satellite imagery"""
        # Multi-scale noise estimation
        noise_estimates = []

        # Method 1: Robust median absolute deviation
        laplacian = np.array([[0, -1, 0], [-1, 4, -1], [0, -1, 0]])
        filtered = ndimage.convolve(image.astype(float), laplacian)
        noise_mad = np.median(np.abs(filtered - np.median(filtered))) / 0.6745
        noise_estimates.append(noise_mad)

        # Method 2: High-frequency analysis
        f_transform = fft2(image)
        h, w = f_transform.shape
        # Create a high-pass filter mask
        rows, cols = image.shape
        center_row, center_col = rows // 2, cols // 2
        mask = np.ones((rows, cols), dtype=np.uint8)
        # Define the low-frequency region to exclude
        r = min(rows, cols) // 8  # Adjust radius as needed
        cv2.circle(mask, (center_col, center_row), r, 0, -1)
        # Apply the mask in the frequency domain
        f_transform_shifted = fftshift(f_transform)
        high_freq_power = np.mean(np.abs(f_transform_shifted * mask)**2)
        noise_estimates.append(np.sqrt(high_freq_power) * 0.1)


        # Method 3: Wavelet-based estimation
        coeffs = pywt.dwt2(image, 'db4')
        _, (cH, cV, cD) = coeffs
        noise_wavelet = np.median(np.abs(cD)) / 0.6745
        noise_estimates.append(noise_wavelet)

        # Robust average
        self.noise_profile = np.median(noise_estimates)
        return self.noise_profile

    def adaptive_svd_denoising(self, image, max_components=None):
        """Improved SVD denoising with automatic rank selection"""
        # Reshape image for SVD
        original_shape = image.shape
        if len(original_shape) == 2:
            # For 2D image, create patches or use direct SVD
            h, w = original_shape

            # Method 1: Block-wise SVD for large images
            if h > 512 or w > 512:
                return self._block_svd_denoise(image, max_components)

            # Method 2: Direct SVD for smaller images
            U, s, Vt = np.linalg.svd(image, full_matrices=False)

            # Automatic rank selection using noise profile
            if max_components is None:
                noise_level = self.estimate_noise_profile(image)
                # Keep components above noise threshold
                signal_threshold = noise_level * np.sqrt(min(h, w))
                optimal_rank = np.sum(s > signal_threshold)
                optimal_rank = max(optimal_rank, min(h, w) // 10)  # At least 10% of components
                optimal_rank = min(optimal_rank, min(h, w) // 2)   # At most 50% of components
            else:
                optimal_rank = min(max_components, len(s))

            self.optimal_components = optimal_rank

            # Reconstruct with optimal rank
            s_filtered = s.copy()
            s_filtered[optimal_rank:] = 0
            denoised = U @ np.diag(s_filtered) @ Vt

            return denoised, optimal_rank, s

        return image, 0, None

    def _block_svd_denoise(self, image, max_components=None, block_size=256):
        """Block-wise SVD for large images"""
        h, w = image.shape
        denoised = np.zeros_like(image)
        overlap = block_size // 4

        for i in range(0, h - block_size + 1, block_size - overlap):
            for j in range(0, w - block_size + 1, block_size - overlap):
                # Extract block
                end_i = min(i + block_size, h)
                end_j = min(j + block_size, w)
                block = image[i:end_i, j:end_j]

                # Apply SVD to block
                U, s, Vt = np.linalg.svd(block, full_matrices=False)

                # Determine optimal rank for this block
                if max_components is None:
                    noise_level = np.median(np.abs(block.flatten() - np.median(block.flatten()))) / 0.6745
                    signal_threshold = noise_level * np.sqrt(min(block.shape))
                    optimal_rank = np.sum(s > signal_threshold)
                    optimal_rank = max(optimal_rank, len(s) // 10)
                    optimal_rank = min(optimal_rank, len(s) // 2)
                else:
                    optimal_rank = min(max_components, len(s))

                # Reconstruct block
                s_filtered = s.copy()
                s_filtered[optimal_rank:] = 0
                block_denoised = U @ np.diag(s_filtered) @ Vt

                # Blend with existing result (for overlapping regions)
                if i == 0 and j == 0:
                    denoised[i:end_i, j:end_j] = block_denoised
                else:
                    # Simple averaging for overlaps
                    mask = denoised[i:end_i, j:end_j] != 0
                    denoised[i:end_i, j:end_j] = np.where(mask,
                                                         (denoised[i:end_i, j:end_j] + block_denoised) / 2,
                                                         block_denoised)

        return denoised, optimal_rank, None

    def enhanced_wiener_filter(self, image, noise_var=None, psf=None):
        """Enhanced Wiener filter with PSF estimation"""
        if noise_var is None:
            noise_var = (self.estimate_noise_profile(image)) ** 2

        # If no PSF provided, estimate motion blur or use identity
        if psf is None:
            psf = np.array([[1]])  # Identity PSF (no blur)

        # Apply Wiener deconvolution
        if psf.size > 1:
            denoised = restoration.wiener(image, psf, noise_var, clip=False)
        else:
            # Simple Wiener filtering in frequency domain
            image_fft = fft2(image)
            power_spectrum = np.abs(image_fft) ** 2
            signal_power = np.maximum(power_spectrum - noise_var, 0.1 * power_spectrum)
            wiener_filter = signal_power / (power_spectrum + noise_var) # Corrected denominator
            filtered_fft = image_fft * wiener_filter
            denoised = np.real(ifft2(filtered_fft))

        return denoised

    def adaptive_wavelet_denoising(self, image, wavelet='auto', levels=None):
        """Enhanced wavelet denoising with adaptive thresholding"""
        # Determine optimal decomposition level
        if levels is None:
            min_dim = min(image.shape)
            levels = int(np.log2(min_dim)) - 2
            levels = max(1, min(levels, 6))

        # Multi-wavelet approach: try different wavelets and combine
        wavelets_to_try = ['db4', 'db8', 'coif2', 'bior2.2'] if wavelet == 'auto' else [wavelet]
        denoised_results = []

        for wvlt in wavelets_to_try:
            try:
                # Decompose
                coeffs = pywt.wavedec2(image, wvlt, level=levels)

                # Adaptive thresholding for each level
                coeffs_thresh = [coeffs[0]]  # Keep approximation coefficients

                for i in range(1, len(coeffs)):
                    detail_coeffs = coeffs[i]
                    thresh_coeffs = []

                    for detail in detail_coeffs:
                        # Level-specific threshold
                        sigma = np.median(np.abs(detail)) / 0.6745

                        # Bayes shrink threshold
                        if sigma > 0:
                            threshold = sigma**2 / np.sqrt(np.maximum(np.var(detail) - sigma**2, 0) + 1e-10)
                        else:
                            threshold = np.max(np.abs(detail)) # Use max value as threshold if sigma is zero

                        # Apply soft thresholding
                        thresh_detail = pywt.threshold(detail, threshold, mode='soft')
                        thresh_coeffs.append(thresh_detail)

                    coeffs_thresh.append(tuple(thresh_coeffs))

                # Reconstruct
                denoised = pywt.waverec2(coeffs_thresh, wvlt)

                # Ensure proper size
                if denoised.shape != image.shape:
                    denoised = denoised[:image.shape[0], :image.shape[1]]

                denoised_results.append(denoised)

            except Exception as e:
                print(f"Warning: Wavelet {wvlt} failed: {e}")
                continue

        if not denoised_results:
            return image

        # If multiple wavelets used, combine results
        if len(denoised_results) == 1:
            return denoised_results[0]
        else:
            # Weighted combination based on local variance
            final_result = np.zeros_like(image)
            total_weight = np.zeros_like(image)

            for result in denoised_results:
                # Local variance as weight (higher variance = more detail preserved)
                local_var = ndimage.generic_filter(result, np.var, size=5)
                weight = local_var / (np.sum([ndimage.generic_filter(r, np.var, size=5)
                                            for r in denoised_results], axis=0) + 1e-10)
                final_result += weight * result
                total_weight += weight

            # Normalize and clip to [0, 1]
            combined_result = final_result / (total_weight + 1e-10)
            return np.clip(combined_result, 0, 1)


    def edge_preserving_filter(self, image, iterations=1):
        """Apply edge-preserving smoothing to reduce remaining noise"""
        # Bilateral filter for edge preservation
        # Ensure image is in the correct format and range for cv2.bilateralFilter
        if image.dtype != np.uint8:
            # Scale to 0-255 for cv2.bilateralFilter and convert to uint8
            image_norm = exposure.rescale_intensity(image, out_range=(0, 255)).astype(np.uint8)
            filtered = cv2.bilateralFilter(image_norm, 9, 75, 75)
            # Scale back to the original range of the input float image
            # Assuming the input float image was in the range [0, 1]
            filtered = exposure.rescale_intensity(filtered.astype(float), out_range=(0.0, 1.0))
        else:
             # If already uint8, just apply the filter
             filtered = cv2.bilateralFilter(image, 9, 75, 75)
             # Convert to float and scale to [0, 1] if the original was float
             filtered = filtered.astype(float) / 255.0


        return filtered


    def enhance_contrast(self, image, method='adaptive'):
        """Post-processing contrast enhancement"""
        # Ensure image is in the correct format and range for scikit-image exposure functions
        if image.dtype != np.float64 or np.min(image) < 0 or np.max(image) > 1:
           image = img_as_float(image) # Convert to float and scale to [0, 1]

        if method == 'adaptive':
            # Adaptive histogram equalization
            return exposure.equalize_adapthist(image, clip_limit=0.01)
        elif method == 'gamma':
            # Gamma correction
            return exposure.adjust_gamma(image, gamma=0.8)
        elif method == 'sigmoid':
            # Sigmoid correction
            return exposure.adjust_sigmoid(image, cutoff=0.5, gain=10)
        else:
            return image

    def comprehensive_denoise(self, image, method='auto', enhance_contrast=True):
        """
        Comprehensive denoising pipeline

        Parameters:
        - image: input noisy image
        - method: 'svd_wavelet', 'wiener_wavelet', 'all', or 'auto'
        - enhance_contrast: apply post-processing contrast enhancement
        """
        print("Starting comprehensive satellite image denoising...")

        # Ensure float type and scale to [0, 1]
        if image.dtype != np.float64:
            image = img_as_float(image) # Convert to float and scale to [0, 1]

        # Store original for comparison
        original = image.copy()

        # Step 1: Initial noise estimation
        noise_level = self.estimate_noise_profile(image)
        print(f"Estimated noise level: {noise_level:.4f}")

        results = {'original': original}

        if method in ['svd_wavelet', 'all', 'auto']:
            print("Applying SVD denoising...")
            svd_denoised, optimal_rank, singular_values = self.adaptive_svd_denoising(image)
            print(f"Optimal SVD rank: {optimal_rank}")

            print("Applying wavelet denoising to SVD result...")
            svd_wavelet_denoised = self.adaptive_wavelet_denoising(svd_denoised, wavelet='auto')

            results['svd_denoised'] = svd_denoised
            results['svd_wavelet_denoised'] = svd_wavelet_denoised
            results['optimal_rank'] = optimal_rank

        if method in ['wiener_wavelet', 'all', 'auto']:
            print("Applying Wiener filter...")
            wiener_denoised = self.enhanced_wiener_filter(image, noise_var=noise_level**2)

            print("Applying wavelet denoising to Wiener result...")
            wiener_wavelet_denoised = self.adaptive_wavelet_denoising(wiener_denoised, wavelet='auto')

            results['wiener_denoised'] = wiener_denoised
            results['wiener_wavelet_denoised'] = wiener_wavelet_denoised

        # Choose best result based on method
        if method == 'svd_wavelet':
            final_denoised = results.get('svd_wavelet_denoised', image) # Use .get for safety
        elif method == 'wiener_wavelet':
            final_denoised = results.get('wiener_wavelet_denoised', image) # Use .get for safety
        elif method in ['all', 'auto']:
            # Combine both methods using local quality metrics
            if 'svd_wavelet_denoised' in results and 'wiener_wavelet_denoised' in results:
                svd_result = results['svd_wavelet_denoised']
                wiener_result = results['wiener_wavelet_denoised']

                # Local gradient preservation as quality metric
                original_grad = np.sqrt(np.gradient(original)[0]**2 + np.gradient(original)[1]**2)
                svd_grad = np.sqrt(np.gradient(svd_result)[0]**2 + np.gradient(svd_result)[1]**2)
                wiener_grad = np.sqrt(np.gradient(wiener_result)[0]**2 + np.gradient(wiener_result)[1]**2)

                svd_quality = 1 / (1 + np.abs(original_grad - svd_grad))
                wiener_quality = 1 / (1 + np.abs(original_grad - wiener_grad))

                # Smooth the quality maps
                svd_quality = ndimage.gaussian_filter(svd_quality, sigma=2)
                wiener_quality = ndimage.gaussian_filter(wiener_quality, sigma=2)

                # Normalize weights
                total_quality = svd_quality + wiener_quality + 1e-10
                svd_weight = svd_quality / total_quality
                wiener_weight = wiener_quality / total_quality

                final_denoised = svd_weight * svd_result + wiener_weight * wiener_result
            else:
                # Fallback if combining is not possible (e.g., one method failed)
                final_denoised = results.get('svd_wavelet_denoised',
                                           results.get('wiener_wavelet_denoised', image))

        # Ensure final denoised image is clipped to [0, 1] before post-processing
        final_denoised = np.clip(final_denoised, 0, 1)

        # Post-processing
        print("Applying edge-preserving filter...")
        final_denoised = self.edge_preserving_filter(final_denoised)

        if enhance_contrast:
            print("Enhancing contrast...")
            final_denoised = self.enhance_contrast(final_denoised, method='adaptive')

        results['final_denoised'] = final_denoised
        results['noise_level'] = noise_level

        # Calculate improvement metrics
        # Calculate PSNR between original (clean) and denoised
        # Note: This requires a "clean" version of the image to be meaningful.
        # The synthetic demo creates a clean image. For a real noisy image where
        # the true clean version is unknown, PSNR is not directly calculable this way.
        # Assuming for the demo/synthetic case, 'original' is the clean image.
        if 'original' in results and np.std(results['original']) > 0: # Check if original exists and is not flat
            clean_for_metrics = results['original']
            # Ensure both are in the same range and type for metric calculation if necessary
            # Assuming they are both float [0, 1] after processing/loading
            mse = np.mean((clean_for_metrics - final_denoised)**2)
            if mse > 1e-10: # Avoid log of zero/near-zero
                # PSNR assumes data range. For float [0, 1], data_range is 1.0
                psnr_value = 10 * np.log10(1.0 / mse)
                results['psnr'] = psnr_value
                print(f"PSNR (vs Original): {psnr_value:.2f} dB")
            else:
                results['psnr'] = float('inf')
                print("PSNR (vs Original): Inf dB (Perfect reconstruction)")

        print("Denoising complete!")
        return results

    def visualize_results(self, results):
        """Comprehensive visualization of denoising results"""
        # Adjust subplot layout based on available results, excluding metrics
        display_keys = [key for key in results.keys() if key not in ['noise_level', 'psnr', 'optimal_rank']]
        n_results = len(display_keys)
        cols = min(3, n_results)
        rows = (n_results + cols - 1) // cols

        if n_results == 0:
            print("No images to display in results.")
            return

        fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))

        # Handle cases where axes might be a single object or a 1D array
        if n_results == 1:
            axes = np.array([[axes]])
        elif rows == 1:
            axes = axes.reshape(1, -1)
        elif cols == 1:
             axes = axes.reshape(-1, 1)
        else:
             axes = axes # For multiple rows and columns


        plot_idx = 0
        for key in display_keys:
            image = results[key]
            # Ensure it's an image array with at least 2 dimensions
            if isinstance(image, np.ndarray) and image.ndim >= 2:
                row = plot_idx // cols
                col = plot_idx % cols

                # Determine appropriate color map and range for display
                cmap = 'gray' if image.ndim == 2 else None # Use gray for 2D, default for color
                vmin = np.min(image)
                vmax = np.max(image)
                # If the image is likely a difference map, adjust cmap and range
                if 'diff' in key.lower() or 'noise' in key.lower() or 'error' in key.lower():
                    cmap = 'hot'
                    vmin = 0 # Difference/Noise maps usually start from 0
                    # vmax could be set to a small value to highlight subtle differences
                    # Let's use a percentile or a fixed small value
                    vmax = np.percentile(image, 99) if image.size > 0 and np.percentile(image, 99) > 1e-6 else 0.1 # Adjust vmax

                im = axes[row, col].imshow(image, cmap=cmap, vmin=vmin, vmax=vmax)
                axes[row, col].set_title(key.replace('_', ' ').title())
                axes[row, col].axis('off')
                # Add colorbar only if it's a grayscale or difference map
                if cmap == 'gray' or cmap == 'hot':
                    plt.colorbar(im, ax=axes[row, col], shrink=0.8)

                plot_idx += 1

        # Hide empty subplots
        for i in range(plot_idx, rows * cols):
            row = i // cols
            col = i % cols
            axes[row, col].axis('off')


        plt.tight_layout()
        plt.show()

def main_satellite_denoising_demo():
    """Demo with synthetic satellite-like imagery"""
    # Create synthetic satellite image with various features
    size = 256
    x, y = np.meshgrid(np.linspace(0, 1, size), np.linspace(0, 1, size))

    # Simulate different terrain features
    fields = 0.5 + 0.3 * np.sin(20 * x) * np.cos(20 * y)  # Agricultural fields
    forest = 0.3 * np.exp(-((x-0.2)**2 + (y-0.8)**2) / 0.1)  # Forest area
    urban = 0.8 * ((x > 0.6) & (x < 0.9) & (y > 0.1) & (y < 0.4))  # Urban area
    water = 0.1 * ((x-0.8)**2 + (y-0.7)**2 < 0.05)  # Water body

    clean_image = fields + forest + urban + water

    # Add realistic noise (sensor noise + atmospheric effects)
    noise_level = 0.15
    gaussian_noise = noise_level * np.random.randn(size, size)
    salt_pepper = np.random.random((size, size))
    clean_image[salt_pepper < 0.01] = 0  # Salt noise
    clean_image[salt_pepper > 0.99] = 1  # Pepper noise

    noisy_image = clean_image + gaussian_noise
    noisy_image = np.clip(noisy_image, 0, 1)

    # Apply denoising
    denoiser = EnhancedSatelliteDenoiser()
    # Run the demo with the 'auto' method which includes Wiener + Wavelet
    results = denoiser.comprehensive_denoise(noisy_image, method='auto')

    # Visualize
    denoiser.visualize_results(results)

    return results

# Usage example for your satellite data
def process_your_satellite_data(image_array):
    """
    Process your actual satellite image

    Parameters:
    - image_array: numpy array of your satellite image
    """
    denoiser = EnhancedSatelliteDenoiser()

    # Method 1: SVD + Wavelet (similar to your current approach)
    results_svd = denoiser.comprehensive_denoise(image_array, method='svd_wavelet')

    # Method 2: Wiener + Wavelet (alternative approach)
    results_wiener = denoiser.comprehensive_denoise(image_array, method='wiener_wavelet')

    # Method 3: Combined approach (recommended)
    results_combined = denoiser.comprehensive_denoise(image_array, method='auto')

    return results_svd, results_wiener, results_combined

if __name__ == "__main__":
    print("Enhanced Satellite Image Denoising Demo")
    print("=" * 50)

    # Run demo which includes Wiener + Wavelet
    demo_results = main_satellite_denoising_demo()

    print("\nTo use with your satellite data:")
    print("results = process_your_satellite_data(your_image_array)")
    print("best_result = results[2]['final_denoised']  # Combined approach")
```

```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            if file_path.lower().endswith(('.tif', '.tiff')):
                # For GeoTIFF files
                try:
                    from osgeo import gdal
                    dataset = gdal.Open(file_path)
                    data_array = dataset.ReadAsArray()

                    # Handle multi-band data
                    if len(data_array.shape) == 3:
                        data_array = np.mean(data_array, axis=0)  # Convert to grayscale

                    # Normalize to [0, 1]
                    noisy_data = (data_array - data_array.min()) / (data_array.max() - data_array.min())

                except ImportError:
                    print("GDAL not available. Using OpenCV...")
                    img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                    if img is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    noisy_data = img_as_float(img)

            elif file_path.lower().endswith('.npy'):
                noisy_data = np.load(file_path)

            else:
                # Standard image formats
                img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                if img is None:
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None
                noisy_data = img_as_float(img)

            print(f"Loaded satellite data from: {file_path}")
            print(f"Data shape: {noisy_data.shape}")

        self.original_data = noisy_data
        return noisy_data

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()

    # Example for processing your own data:
    # results = process_your_satellite_data(
    #     "path/to/your/satellite_data.tif",
    #     "path/to/output/denoised_data.npy"
    # )
```

```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            if file_path.lower().endswith(('.tif', '.tiff')):
                # For GeoTIFF files
                try:
                    from osgeo import gdal
                    dataset = gdal.Open(file_path)
                    data_array = dataset.ReadAsArray()

                    # Handle multi-band data
                    if len(data_array.shape) == 3:
                        data_array = np.mean(data_array, axis=0)  # Convert to grayscale

                    # Normalize to [0, 1]
                    noisy_data = (data_array - data_array.min()) / (data_array.max() - data_array.min() + 1e-10)

                except ImportError:
                    print("GDAL not available. Using OpenCV...")
                    img = cv2.imread(file_path)
                    if img is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img.shape) == 3:
                        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
                    noisy_data = img_as_float(img)

            elif file_path.lower().endswith('.npy'):
                noisy_data = np.load(file_path)

            else:
                # Standard image formats (e.g., .bmp, .png, .jpg)
                img = cv2.imread(file_path)
                if img is None:
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None
                # Convert to grayscale if it's a color image
                if len(img.shape) == 3:
                    img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
                noisy_data = img_as_float(img)

            print(f"Loaded satellite data from: {file_path}")
            print(f"Data shape: {noisy_data.shape}")

        self.original_data = noisy_data
        return noisy_data

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()
```

I will make the `load_satellite_data` method more robust by first attempting to load `.bmp` files using GDAL, as it successfully loaded the image in a previous cell (`ecc1aab3`). This prioritizes a more capable geospatial library for image loading before falling back to OpenCV for generic image formats. I will also explicitly import `ssim` from `skimage.metrics` as it is used in `calculate_quality_metrics`.



```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim # Explicitly import ssim
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            noisy_data = None # Initialize noisy_data

            # Try GDAL for GeoTIFFs and potentially other formats like BMP if it's Geo-referenced
            if file_path.lower().endswith(('.tif', '.tiff', '.bmp')):
                try:
                    from osgeo import gdal
                    dataset = gdal.Open(file_path)
                    if dataset: # Check if dataset was successfully opened
                        data_array = dataset.ReadAsArray()
                        if len(data_array.shape) == 3:
                            data_array = np.mean(data_array, axis=0)  # Convert to grayscale
                        noisy_data = (data_array - data_array.min()) / (data_array.max() - data_array.min() + 1e-10)
                        print(f"Loaded satellite data with GDAL from: {file_path}")
                        print(f"Data shape: {noisy_data.shape}")
                        self.original_data = noisy_data
                        return noisy_data
                    else:
                        print(f"GDAL could not open {file_path}.")
                except ImportError:
                    print("GDAL not available.")
                except Exception as e:
                    print(f"GDAL error: {e}")

            # Fallback to tifffile for TIFFs if GDAL failed or not available for .tif
            if noisy_data is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img = tifffile.imread(file_path)
                    if img is not None:
                         if len(img.shape) == 3 and img.shape[0] < img.shape[2]: # Assume (bands, H, W) -> (H, W, bands)
                             img = np.moveaxis(img, 0, -1)
                         if len(img.shape) == 3: # Convert to grayscale if still color
                             img = np.mean(img, axis=-1)
                         noisy_data = img_as_float(img)
                         print(f"Loaded satellite data with tifffile from: {file_path}")
                         print(f"Data shape: {noisy_data.shape}")
                         self.original_data = noisy_data
                         return noisy_data
                except Exception as e:
                    print(f"tifffile error: {e}")

            # Fallback to OpenCV for other standard image formats (including BMP if GDAL fails)
            if noisy_data is None: # Only try OpenCV if previous methods failed
                try:
                    img = cv2.imread(file_path)
                    if img is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img.shape) == 3:
                        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
                    noisy_data = img_as_float(img)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")

                except Exception as e:
                    print(f"OpenCV error: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            self.original_data = noisy_data # Assign once after successful load
            return noisy_data

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()
```

**Reasoning**:
The previous execution failed because neither GDAL nor OpenCV could load the `.bmp` image. This is likely due to how the image data is read or interpreted by these libraries in this specific context. I will modify the `load_satellite_data` function to enhance its robustness for `.bmp` files by ensuring GDAL uses `gdal.GA_ReadOnly` consistently (as it succeeded in a previous cell) and correctly handles multi-band images before converting them to grayscale and normalizing to `[0,1]` using `img_as_float` from `skimage` for better compatibility with the denoising algorithms.



```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim # Explicitly import ssim
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/drive/MyDrive/Satellite Project with Debojyoti Banerjee/all_band_vskp.tif', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            loaded_image_array = None # Raw array before normalization

            # 1. Try GDAL first for GeoTIFFs and common image formats
            try:
                from osgeo import gdal
                dataset = gdal.Open(file_path, gdal.GA_ReadOnly) # Explicitly use GA_ReadOnly
                if dataset:
                    # Read all bands (GDAL ReadAsArray often returns (bands, height, width))
                    raw_array_gdal = dataset.ReadAsArray()

                    if raw_array_gdal is not None:
                        if raw_array_gdal.ndim == 3:
                            # Check typical band order (bands, H, W) vs (H, W, bands)
                            # If the first dimension is small (e.g., 3 or 4) and others are larger,
                            # assume (bands, H, W) and transpose to (H, W, bands)
                            if raw_array_gdal.shape[0] < raw_array_gdal.shape[1] and raw_array_gdal.shape[0] < raw_array_gdal.shape[2]:
                                raw_array_gdal = np.moveaxis(raw_array_gdal, 0, -1) # Convert to (H, W, bands)
                            # Now raw_array_gdal is (H, W, bands). Convert to grayscale.
                            # For simplicity, mean across channels. This will create a 2D array.
                            loaded_image_array = np.mean(raw_array_gdal, axis=-1)
                        elif raw_array_gdal.ndim == 2:
                            # Already grayscale
                            loaded_image_array = raw_array_gdal

                        if loaded_image_array is not None:
                            # Convert to float and normalize to [0, 1] using skimage's utility
                            noisy_data = img_as_float(loaded_image_array)
                            print(f"Loaded satellite data with GDAL from: {file_path}")
                            print(f"Data shape: {noisy_data.shape}")
                            self.original_data = noisy_data
                            return noisy_data
                    print(f"GDAL could not process data from {file_path}.")
                except ImportError:
                    print("GDAL not available.")
                except Exception as e:
                    print(f"GDAL error during loading: {e}")

            except Exception as e: # Catch any errors from the GDAL block attempt
                print(f"Attempt to load with GDAL failed unexpectedly: {e}")

            # 2. Fallback to tifffile for TIFFs if GDAL failed or not available
            # This block is only entered if noisy_data is still None
            if loaded_image_array is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img_tif = tifffile.imread(file_path)
                    if img_tif is not None:
                        if img_tif.ndim == 3:
                            if img_tif.shape[0] < img_tif.shape[2]: # Assume (bands, H, W) -> (H, W, bands)
                                img_tif = np.moveaxis(img_tif, 0, -1)
                            img_tif = np.mean(img_tif, axis=-1) # Convert to grayscale
                        loaded_image_array = img_tif

                        noisy_data = img_as_float(loaded_image_array)
                        print(f"Loaded satellite data with tifffile from: {file_path}")
                        print(f"Data shape: {noisy_data.shape}")
                        self.original_data = noisy_data
                        return noisy_data
                except Exception as e:
                    print(f"tifffile error during loading: {e}")

            # 3. Fallback to OpenCV for other standard image formats (including BMP if previous methods failed)
            if loaded_image_array is None: # Only try OpenCV if previous methods failed
                try:
                    img_cv = cv2.imread(file_path) # Reads as-is (color if it is, grayscale otherwise)
                    if img_cv is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img_cv.shape) == 3: # If it's a color image, convert to grayscale
                        img_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                    loaded_image_array = img_cv

                    noisy_data = img_as_float(loaded_image_array)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")
                    self.original_data = noisy_data
                    return noisy_data
                except Exception as e:
                    print(f"OpenCV error during loading: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            print(f"Failed to load image from {file_path} with any available library.")
            return None

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]%
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()

    # Example for processing your own data:
    # results = process_your_satellite_data(
    #     "path/to/your/satellite_data.tif",
    #     "path/to/output/denoised_data.npy"
    # )
```

```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim # Explicitly import ssim
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            loaded_image_array = None # Raw array before normalization
            noisy_data = None # Final normalized data

            # 1. Try GDAL first for GeoTIFFs and common image formats
            try:
                from osgeo import gdal
                # If GDAL import is successful, try opening the dataset
                dataset = gdal.Open(file_path, gdal.GA_ReadOnly) # Explicitly use GA_ReadOnly
                if dataset:
                    # Read all bands (GDAL ReadAsArray often returns (bands, height, width))
                    raw_array_gdal = dataset.ReadAsArray()

                    if raw_array_gdal is not None:
                        if raw_array_gdal.ndim == 3:
                            # Check typical band order (bands, H, W) vs (H, W, bands)
                            # If the first dimension is small (e.g., 3 or 4) and others are larger,
                            # assume (bands, H, W) and transpose to (H, W, bands)
                            if raw_array_gdal.shape[0] < raw_array_gdal.shape[1] and raw_array_gdal.shape[0] < raw_array_gdal.shape[2]:
                                raw_array_gdal = np.moveaxis(raw_array_gdal, 0, -1) # Convert to (H, W, bands)
                            # Now raw_array_gdal is (H, W, bands). Convert to grayscale.
                            # For simplicity, mean across channels. This will create a 2D array.
                            loaded_image_array = np.mean(raw_array_gdal, axis=-1)
                        elif raw_array_gdal.ndim == 2:
                            # Already grayscale
                            loaded_image_array = raw_array_gdal

                        if loaded_image_array is not None:
                            # Convert to float and normalize to [0, 1] using skimage's utility
                            noisy_data = img_as_float(loaded_image_array)
                            print(f"Loaded satellite data with GDAL from: {file_path}")
                            print(f"Data shape: {noisy_data.shape}")
                            self.original_data = noisy_data
                            return noisy_data
                    print(f"GDAL could not process data from {file_path}.")
                else:
                    print(f"GDAL could not open {file_path}.")
            except ImportError:
                print("GDAL not available.")
            except Exception as e:
                print(f"GDAL error during loading: {e}")

            # 2. Fallback to tifffile for TIFFs if GDAL failed or not available
            # This block is only entered if noisy_data is still None
            if noisy_data is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img_tif = tifffile.imread(file_path)
                    if img_tif is not None:
                         if img_tif.ndim == 3:
                            if img_tif.shape[0] < img_tif.shape[2]: # Assume (bands, H, W) -> (H, W, bands)
                                img_tif = np.moveaxis(img_tif, 0, -1)
                            img_tif = np.mean(img_tif, axis=-1) # Convert to grayscale
                         loaded_image_array = img_tif

                         noisy_data = img_as_float(loaded_image_array)
                         print(f"Loaded satellite data with tifffile from: {file_path}")
                         print(f"Data shape: {noisy_data.shape}")
                         self.original_data = noisy_data
                         return noisy_data
                except Exception as e:
                    print(f"tifffile error during loading: {e}")

            # 3. Fallback to OpenCV for other standard image formats (including BMP if previous methods failed)
            if noisy_data is None: # Only try OpenCV if previous methods failed
                try:
                    img_cv = cv2.imread(file_path) # Reads as-is (color if it is, grayscale otherwise)
                    if img_cv is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img_cv.shape) == 3: # If it's a color image, convert to grayscale
                        img_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                    loaded_image_array = img_cv

                    noisy_data = img_as_float(loaded_image_array)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")
                    self.original_data = noisy_data
                    return noisy_data
                except Exception as e:
                    print(f"OpenCV error during loading: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            print(f"Failed to load image from {file_path} with any available library.")
            return None

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()

    # Example for processing your own data:
    # results = process_your_satellite_data(
    #     "path/to/your/satellite_data.tif",
    #     "path/to/output/denoised_data.npy"
    # )
```

```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim # Explicitly import ssim
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path='/content/Sat_Image_Denoising/2_A.bmp', use_sample=False):
        """
        Load satellite data from various formats

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration

        Returns:
        data: numpy array, satellite data
        """
        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data
            if file_path is None:
                 print("Error: file_path is required when use_sample is False.")
                 return None

            loaded_image_array = None # Raw array before normalization
            noisy_data = None # Final normalized data

            # 1. Try GDAL first for GeoTIFFs and common image formats
            try:
                from osgeo import gdal
                # If GDAL import is successful, try opening the dataset
                dataset = gdal.Open(file_path, gdal.GA_ReadOnly) # Explicitly use GA_ReadOnly
                if dataset:
                    # Read all bands (GDAL ReadAsArray often returns (bands, height, width))
                    raw_array_gdal = dataset.ReadAsArray()

                    if raw_array_gdal is not None:
                        if raw_array_gdal.ndim == 3:
                            # Check typical band order (bands, H, W) vs (H, W, bands)
                            # If the first dimension is small (e.g., 3 or 4) and others are larger,
                            # assume (bands, H, W) and transpose to (H, W, bands)
                            if raw_array_gdal.shape[0] < raw_array_gdal.shape[1] and raw_array_gdal.shape[0] < raw_array_gdal.shape[2]:
                                raw_array_gdal = np.moveaxis(raw_array_gdal, 0, -1) # Convert to (H, W, bands)
                            # Now raw_array_gdal is (H, W, bands). Convert to grayscale.
                            # For simplicity, mean across channels. This will create a 2D array.
                            loaded_image_array = np.mean(raw_array_gdal, axis=-1)
                        elif raw_array_gdal.ndim == 2:
                            # Already grayscale
                            loaded_image_array = raw_array_gdal

                        if loaded_image_array is not None:
                            # Convert to float and normalize to [0, 1] using skimage's utility
                            noisy_data = img_as_float(loaded_image_array)
                            print(f"Loaded satellite data with GDAL from: {file_path}")
                            print(f"Data shape: {noisy_data.shape}")
                            self.original_data = noisy_data
                            return noisy_data
                    print(f"GDAL could not process data from {file_path}.")
                else:
                    print(f"GDAL could not open {file_path}.")
            except ImportError:
                print("GDAL not available.")
            except Exception as e:
                print(f"GDAL error during loading: {e}")

            # 2. Fallback to tifffile for TIFFs if GDAL failed or not available
            # This block is only entered if noisy_data is still None
            if noisy_data is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img_tif = tifffile.imread(file_path)
                    if img_tif is not None:
                         if img_tif.ndim == 3:
                            if img_tif.shape[0] < img_tif.shape[2]: # Assume (bands, H, W) -> (H, W, bands)
                                img_tif = np.moveaxis(img_tif, 0, -1)
                            img_tif = np.mean(img_tif, axis=-1) # Convert to grayscale
                         loaded_image_array = img_tif

                         noisy_data = img_as_float(loaded_image_array)
                         print(f"Loaded satellite data with tifffile from: {file_path}")
                         print(f"Data shape: {noisy_data.shape}")
                         self.original_data = noisy_data
                         return noisy_data
                except Exception as e:
                    print(f"tifffile error during loading: {e}")

            # 3. Fallback to OpenCV for other standard image formats (including BMP if previous methods failed)
            if noisy_data is None: # Only try OpenCV if previous methods failed
                try:
                    img_cv = cv2.imread(file_path) # Reads as-is (color if it is, grayscale otherwise)
                    if img_cv is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img_cv.shape) == 3: # If it's a color image, convert to grayscale
                        img_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                    loaded_image_array = img_cv

                    noisy_data = img_as_float(loaded_image_array)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")
                    self.original_data = noisy_data
                    return noisy_data
                except Exception as e:
                    print(f"OpenCV error during loading: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            print(f"Failed to load image from {file_path} with any available library.")
            return None

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data
    # Use the local path to the image file from the cloned repository
    image_path = "/content/Sat_Image_Denoising/2_A.bmp"
    satellite_data = denoiser.load_satellite_data(file_path=image_path, use_sample=False)

    if satellite_data is None:
        print("Failed to load satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()
```

**Reasoning**:
The previous executions consistently failed to load the image due to issues with `cv2.imread` and `gdal.Open` for the `.bmp` file, even after multiple attempts to make the loading more robust. To ensure the SVD + Wavelet denoising logic can execute and demonstrate its functionality, I will modify the `main` function within the `SVDWaveletDenoiser` class to use the `original_image_arr` which was successfully loaded and preprocessed in previous steps (specifically, the output of the mean filter cell `ecc1aab3`). This approach directly feeds a valid, normalized image into the SVD + Wavelet pipeline, bypassing the problematic file I/O for the `.bmp` file. I will also ensure the image is converted to grayscale, as the SVD + Wavelet algorithms in this class are designed for 2D (grayscale) images.



```python
#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim # Explicitly import ssim
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path=None, use_sample=False, preloaded_image=None):
        """
        Load satellite data from various formats or use a preloaded image.

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration
        preloaded_image: numpy array, an already loaded image array to use directly

        Returns:
        data: numpy array, satellite data
        """
        if preloaded_image is not None:
            # Use the provided preloaded image
            print("Using preloaded image for SVD + Wavelet denoising.")
            # Ensure it's grayscale if it's RGB, and normalize to 0-1
            if preloaded_image.ndim == 3 and preloaded_image.shape[2] == 3:
                # Convert to grayscale by averaging channels, then normalize to 0-1
                gray_image = np.mean(preloaded_image, axis=-1)
                noisy_data = img_as_float(gray_image)
            elif preloaded_image.ndim == 2:
                # Already grayscale, just normalize to 0-1
                noisy_data = img_as_float(preloaded_image)
            else:
                print(f"Warning: Preloaded image has unexpected dimensions {preloaded_image.shape}. Attempting to use as-is.")
                noisy_data = img_as_float(preloaded_image)

            self.original_data = noisy_data
            print(f"Preloaded data shape: {noisy_data.shape}")
            return noisy_data

        if use_sample:
            # Create sample noisy satellite data
            base_data = img_as_float(data.camera())

            # Add various types of noise to simulate real satellite data issues
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0

            # Combine noises
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)

            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            # Load actual satellite data from file
            if file_path is None:
                 print("Error: file_path is required when use_sample is False and no preloaded_image is given.")
                 return None

            loaded_image_array = None # Raw array before normalization
            noisy_data = None # Final normalized data

            # 1. Try GDAL first for GeoTIFFs and common image formats
            try:
                from osgeo import gdal
                # If GDAL import is successful, try opening the dataset
                dataset = gdal.Open(file_path, gdal.GA_ReadOnly) # Explicitly use GA_ReadOnly
                if dataset:
                    # Read all bands (GDAL ReadAsArray often returns (bands, height, width))
                    raw_array_gdal = dataset.ReadAsArray()

                    if raw_array_gdal is not None:
                        if raw_array_gdal.ndim == 3:
                            # Check typical band order (bands, H, W) vs (H, W, bands)
                            # If the first dimension is small (e.g., 3 or 4) and others are larger,
                            # assume (bands, H, W) and transpose to (H, W, bands)
                            if raw_array_gdal.shape[0] < raw_array_gdal.shape[1] and raw_array_gdal.shape[0] < raw_array_gdal.shape[2]:
                                raw_array_gdal = np.moveaxis(raw_array_gdal, 0, -1) # Convert to (H, W, bands)
                            # Now raw_array_gdal is (H, W, bands). Convert to grayscale.
                            # For simplicity, mean across channels. This will create a 2D array.
                            loaded_image_array = np.mean(raw_array_gdal, axis=-1)
                        elif raw_array_gdal.ndim == 2:
                            # Already grayscale
                            loaded_image_array = raw_array_gdal

                        if loaded_image_array is not None:
                            # Convert to float and normalize to [0, 1] using skimage's utility
                            noisy_data = img_as_float(loaded_image_array)
                            print(f"Loaded satellite data with GDAL from: {file_path}")
                            print(f"Data shape: {noisy_data.shape}")
                            self.original_data = noisy_data
                            return noisy_data
                    print(f"GDAL could not process data from {file_path}.")
                else:
                    print(f"GDAL could not open {file_path}.")
            except ImportError:
                print("GDAL not available.")
            except Exception as e:
                print(f"GDAL error during loading: {e}")

            # 2. Fallback to tifffile for TIFFs if GDAL failed or not available
            # This block is only entered if noisy_data is still None
            if noisy_data is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img_tif = tifffile.imread(file_path)
                    if img_tif is not None:
                         if img_tif.ndim == 3:
                            if img_tif.shape[0] < img_tif.shape[2]: # Assume (bands, H, W) -> (H, W, bands)
                                img_tif = np.moveaxis(img_tif, 0, -1)
                            img_tif = np.mean(img_tif, axis=-1) # Convert to grayscale
                         loaded_image_array = img_tif

                         noisy_data = img_as_float(loaded_image_array)
                         print(f"Loaded satellite data with tifffile from: {file_path}")
                         print(f"Data shape: {noisy_data.shape}")
                         self.original_data = noisy_data
                         return noisy_data
                except Exception as e:
                    print(f"tifffile error during loading: {e}")

            # 3. Fallback to OpenCV for other standard image formats (including BMP if previous methods failed)
            if noisy_data is None: # Only try OpenCV if previous methods failed
                try:
                    img_cv = cv2.imread(file_path) # Reads as-is (color if it is, grayscale otherwise)
                    if img_cv is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img_cv.shape) == 3: # If it's a color image, convert to grayscale
                        img_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                    loaded_image_array = img_cv

                    noisy_data = img_as_float(loaded_image_array)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")
                    self.original_data = noisy_data
                    return noisy_data
                except Exception as e:
                    print(f"OpenCV error during loading: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            print(f"Failed to load image from {file_path} with any available library.")
            return None

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        # Use wavelet coefficients at finest scale for noise estimation
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]  # High-frequency diagonal coefficients
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            # Apply SVD to entire image
            U, s, Vt = svd(data, full_matrices=False)

            # Determine rank to keep
            rank = int(len(s) * rank_ratio)

            # Truncate singular values
            s_truncated = s.copy()
            s_truncated[rank:] = 0

            # Reconstruct
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            # Patch-based SVD denoising
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            # Process overlapping blocks
            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    # Extract block
                    block = data[i:i+block_size, j:j+block_size]

                    # Apply SVD
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0

                    # Reconstruct block
                    denoised_block = U @ np.diag(s_truncated) @ Vt

                    # Add to result with weighting
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            # Normalize by weights
            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        # Wavelet decomposition
        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)

        # Estimate noise level
        sigma = self.estimate_noise_level(data)

        # Calculate thresholds for each level
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):  # Skip approximation coefficients
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                # BayesShrink adaptive threshold
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                # SURE threshold
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                # Minimax threshold
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:  # manual
                # Manual threshold based on noise level
                level_thresh = 3 * sigma  # 3-sigma rule
                thresholds.append([level_thresh] * 3)

        # Apply thresholding
        coeffs_thresholded = [coeffs[0].copy()]  # Keep approximation coefficients

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            # Apply thresholding to each subband
            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:  # hard
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        # Reconstruct
        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised  # For consistency in return values

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        # Avoid division by zero or log of zero/negative number
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        # Signal-to-Noise Ratio
        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0: # Ensure signal_power is also > 0
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0: # Perfect denoising
             snr = float('inf')
        else: # No signal or only noise
            snr = np.nan # Not a Number


        # Structural Similarity (simplified) - using skimage.metrics.structural_similarity
        # Ensure inputs are in a valid range and format for SSIM
        # Normalize to 0-1 range if not already
        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        # Check if images are grayscale or color
        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             # Assuming channel last (height, width, channels)
             if original_norm.shape[2] in [1, 3, 4]: # Handle grayscale (1), RGB (3), RGBA (4)
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else: # Handle other multi-channel images by averaging channels or processing band by band
                 # For simplicity, calculate SSIM for each channel and average
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan # Not applicable


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score # Use actual SSIM score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison

        Parameters:
        results: dict, results from hybrid_svd_wavelet_denoise
        show_coefficients: bool, show wavelet coefficients
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        # Ensure images are in a displayable format (e.g., uint8 or float in 0-1)
        # Normalize to 0-1 range for consistent display
        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        # Original noisy data
        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        # SVD denoised
        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        # Final result
        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            # Difference images
            # Calculate differences relative to the original data range
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            # Normalize difference images for display
            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            # Wavelet coefficients visualization
            coeffs = results['wavelet_coeffs']
            # Use pywt.coeffs_to_array safely, handling potential errors
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off') # Turn off axis if visualization fails


        plt.tight_layout()
        plt.show()

        # Calculate and display metrics
        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}") # Use actual SSIM score
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        # Find best method based on PSNR
        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf')) # Handle NaN PSNR
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Example usage and demonstration
def main():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    # Initialize denoiser
    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    # Load satellite data using the globally available 'original_image_arr'
    # Ensure original_image_arr is normalized to 0-1 and is grayscale for this algorithm
    global original_image_arr # Access the global variable

    if 'original_image_arr' not in globals() or original_image_arr is None:
        print("Error: original_image_arr is not available. Please run the mean filter cell first.")
        return

    # Convert original_image_arr to grayscale and normalize to 0-1 for SVDWaveletDenoiser
    # Assuming original_image_arr is (H, W, 3) and values are 0-255.
    if original_image_arr.ndim == 3 and original_image_arr.shape[2] == 3:
        # Convert to grayscale by averaging channels
        grayscale_image = np.mean(original_image_arr, axis=-1)
    else:
        grayscale_image = original_image_arr # Assume it's already grayscale if not 3D color

    # Normalize to 0-1 range, as img_as_float expects
    satellite_data = img_as_float(grayscale_image)

    print(f"Loaded original image (converted to grayscale, normalized) for SVD + Wavelet. Shape: {satellite_data.shape}")

    if satellite_data is None:
        print("Failed to prepare satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(satellite_data)

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try: # Add try-except block for wavelet testing
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                satellite_data, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# Function to process your own satellite data
def process_your_satellite_data(file_path, output_path=None):
    """
    Process your actual satellite image

    Parameters:
    - file_path: str, path to your satellite data
    - output_path: str, path to save denoised data (optional)
    """
    denoiser = SVDWaveletDenoiser(wavelet='db8')

    # Load your data
    data = denoiser.load_satellite_data(file_path, use_sample=False)

    if data is None:
        print(f"Failed to load data from {file_path}. Aborting processing.")
        return None

    # Apply denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        data,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4
    )

    # Visualize
    denoiser.visualize_denoising_results(results)

    # Save if requested
    if output_path:
        # Ensure denoised data is in a saveable format (e.g., 0-255 uint8 or 0-1 float)
        # Assuming the denoised data is in the range [0, 1] as per `np.clip(denoised, 0, 1)`
        if output_path.endswith('.npy'):
            np.save(output_path, results['final_denoised'])
        else:
            # Save as image - rescale to 0-255 and convert to uint8
            denoised_uint8 = (results['final_denoised'] * 255).astype(np.uint8)
            cv2.imwrite(output_path, denoised_uint8)
        print(f"Denoised data saved to: {output_path}")

    return results

if __name__ == "__main__":
    main()

    # Example for processing your own data:
    # results = process_your_satellite_data(
    #     "path/to/your/satellite_data.tif",
    #     "path/to/output/denoised_data.npy"
    # )
```

```python
# Execute the main function from ecc1aab3 to ensure global variables are set
# This code block is a replica of the necessary parts from ecc1aab3

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import uniform_filter
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import warnings
warnings.filterwarnings('ignore')

def install_dependencies():
    """Install required libraries for GeoTIFF processing."""
    import subprocess
    import sys

    print("Installing required libraries...")
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install",
                              "rasterio", "-q"])
        print("Successfully installed rasterio")
    except:
        print("Could not install rasterio. Trying alternative method...")


def load_geotiff_rasterio(image_path):
    """
    Load GeoTIFF using rasterio (RECOMMENDED for satellite data).
    """
    try:
        import rasterio

        print(f"Loading with rasterio...")
        with rasterio.open(image_path) as src:
            image = src.read()  # Shape: (bands, height, width)
            n_bands = src.count
            height = src.height
            width = src.width
            dtype = src.dtypes[0]

            print(f"Loaded GeoTIFF:")
            print(f"    - Bands: {n_bands}")
            print(f"    - Size: {width} x {height}")
            print(f"    - Data type: {dtype}")
            print(f"    - CRS: {src.crs}")

            image = np.moveaxis(image, 0, -1)

            return image, True

    except ImportError:
        print("rasterio not available")
        return None, False
    except Exception as e:
        print(f"rasterio error: {e}")
        return None, False


def load_geotiff_gdal(image_path):
    """
    Load GeoTIFF using GDAL (fallback method).
    """
    try:
        from osgeo import gdal

        print(f"  Loading with GDAL...")
        dataset = gdal.Open(image_path, gdal.GA_ReadOnly)

        if dataset is None:
            return None, False

        n_bands = dataset.RasterCount
        width = dataset.RasterXSize
        height = dataset.RasterYSize

        print(f"Loaded GeoTIFF:")
        print(f"    - Bands: {n_bands}")
        print(f"    - Size: {width} x {height}")

        bands = []
        for i in range(1, n_bands + 1):
            band = dataset.GetRasterBand(i)
            bands.append(band.ReadAsArray())

        image = np.stack(bands, axis=-1)

        dataset = None  # Close dataset
        return image, True

    except ImportError:
        print("GDAL not available")
        return None, False
    except Exception as e:
        print(f"GDAL error: {e}")
        return None, False


def load_geotiff_tifffile(image_path):
    """
    Load TIFF using tifffile library (works for some TIFFs).
    """
    try:
        import tifffile

        print(f"  Loading with tifffile...")
        image = tifffile.imread(image_path)

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        if len(image.shape) == 3:
            if image.shape[0] < image.shape[2]:
                image = np.moveaxis(image, 0, -1)

        return image, True

    except Exception as e:
        print(f"tifffile error: {e}")
        return None, False


def load_geotiff_opencv(image_path):
    """
    Load TIFF using OpenCV (limited multi-band support).
    """
    try:
        import cv2

        print(f"  Loading with OpenCV...")
        image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)

        if image is None:
            return None, False

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        if len(image.shape) == 3 and image.shape[2] == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        return image, True

    except Exception as e:
        print(f"OpenCV error: {e}")
        return None, False

def load_satellite_image(image_path):
    """
    Try multiple methods to load GeoTIFF satellite image.
    """
    import os

    print(f"\n[1] Loading satellite image from:")
    print(f"    {image_path}")

    if not os.path.exists(image_path):
        print(f"Error: File not found at {image_path}")
        return None

    image, success = load_geotiff_rasterio(image_path)
    if success and image is not None:
        return image

    image, success = load_geotiff_gdal(image_path)
    if success and image is not None:
        return image

    image, success = load_geotiff_tifffile(image_path)
    if success and image is not None:
        return image

    image, success = load_geotiff_opencv(image_path)
    if success and image is not None:
        return image

    if os.path.exists(image_path):
        raise RuntimeError(
            f"Could not load image from {image_path} with any available library.\n"
            "Please check the file format and consider installing rasterio: pip install rasterio"
        )
    else:
        raise FileNotFoundError(f"Image file not found at: {image_path}")


def normalize_to_uint8(image_array):
    """Normalize to 0-255 for display using percentile clipping."""
    if image_array.size == 0:
        return np.empty_like(image_array, dtype=np.uint8)

    p2, p98 = np.percentile(image_array, (2, 98))
    if p2 == p98:
        return np.full_like(image_array, 127, dtype=np.uint8)

    image_clipped = np.clip(image_array, p2, p98)
    image_normalized = ((image_clipped - p2) / (p98 - p2) * 255)
    return np.clip(image_normalized, 0, 255).astype(np.uint8)


def select_rgb_bands(image_array, rgb_indices=None):
    """Extract RGB bands from multi-band image."""
    print(f"\n[2] Extracting RGB bands...")
    print(f"    Image shape: {image_array.shape}")

    if len(image_array.shape) == 2:
        print("    Grayscale detected - converting to RGB")
        return np.stack([image_array] * 3, axis=-1)

    elif len(image_array.shape) == 3 and image_array.shape[2] == 3:
        print("    RGB image detected")
        return image_array

    elif len(image_array.shape) == 3 and image_array.shape[2] > 3:
        if rgb_indices is None:
            n_bands = image_array.shape[2]
            print(f"    Multi-band image: {n_bands} bands detected")
            print("    Trying common RGB band combinations...")

            if n_bands >= 4:
                print("    Using bands [3,2,1] for RGB (Sentinel-2/Landsat style)")
                rgb_indices = (3, 2, 1)
            else:
                print("    Using first 3 bands [0,1,2]")
                rgb_indices = (0, 1, 2)

        max_band_index = image_array.shape[2] - 1
        valid_indices = [idx for idx in rgb_indices if 0 <= idx <= max_band_index]

        if len(valid_indices) != 3:
            print(f"Warning: Could not find 3 valid RGB bands from indices {rgb_indices}. Using first 3 available bands.")
            valid_indices = list(range(min(3, image_array.shape[2])))
            if len(valid_indices) < 3:
                 raise ValueError(f"Image only has {image_array.shape[2]} bands, cannot select 3 for RGB.")

        print(f"    Selected RGB indices: {valid_indices}")
        return image_array[:, :, valid_indices]


    else:
        raise ValueError(f"Unexpected image shape: {image_array.shape}")


def add_gaussian_noise(image, sigma=25.0):
    """Add Gaussian noise."""
    image_float = image.astype(np.float64)
    noise = np.random.normal(0, sigma, image_float.shape)
    noisy_image = image_float + noise
    return noisy_image


def apply_mean_filter_multichannel(image, kernel_size=5):
    """Apply mean filter to each channel separately."""
    image_float = image.astype(np.float64)

    if len(image_float.shape) == 2:
        return uniform_filter(image_float, size=kernel_size)

    filtered_channels = []
    for i in range(image_float.shape[2]):
        filtered_channels.append(
            uniform_filter(image_float[:, :, i], size=kernel_size)
        )
    return np.stack(filtered_channels, axis=-1)


def calculate_metrics(original, processed, data_range=None):
    """Calculate PSNR and SSIM."""
    if data_range is None:
        if original.dtype == np.uint8:
            data_range = 255.0
        elif original.dtype == np.float64 or original.dtype == np.float32:
            data_range = np.max(original)

        if data_range is None or data_range <= 0:
             data_range = 1.0

    original_float = original.astype(np.float64)
    processed_float = processed.astype(np.float64)
    processed_clipped = np.clip(processed_float, original_float.min(), original_float.max())

    try:
        psnr_value = psnr(original_float, processed_clipped, data_range=data_range)
    except Exception as e:
        print(f"Error calculating PSNR: {e}")
        psnr_value = np.nan

    try:
        min_dim = min(original_float.shape[0], original_float.shape[1])
        win_size = min(7, min_dim)
        if win_size < 1:
            win_size = 1

        if len(original_float.shape) == 3:
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range,
                             channel_axis=2, win_size=win_size)
        else:
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range, win_size=win_size)
    except Exception as e:
        print(f"Error calculating SSIM: {e}")
        ssim_value = np.nan

    return psnr_value, ssim_value


def main_ecc1aab3():
    """Complete denoising pipeline."""

    print("="*70)
    print("SATELLITE IMAGE DENOISING - GEOTIFF COMPATIBLE (Re-run)")
    print("="*70)

    image_path = "/content/Sat_Image_Denoising/2_A.bmp"

    try:
        import os
        if not os.path.exists(image_path):
             raise FileNotFoundError(f"Image file not found at: {image_path}")

        image_array_raw = load_satellite_image(image_path)

        if image_array_raw is None:
             raise RuntimeError("Failed to load image.")

        if len(image_array_raw.shape) == 2:
             image_array = np.stack([image_array_raw]*3, axis=-1)
             print("Grayscale image loaded, converted to RGB.")
        else:
             image_array = select_rgb_bands(image_array_raw, None)

        image_array = image_array.astype(np.float64)

        print(f"\n[3] Normalizing data range...")
        max_original_value = np.max(image_array)
        if max_original_value > 0:
             image_array_normalized = image_array / max_original_value * 255.0
        else:
             image_array_normalized = image_array

        print(f"    Normalized range: [{image_array_normalized.min():.2f}, {image_array_normalized.max():.2f}]")

        print(f"\n[4] Adding Gaussian noise...")
        noise_sigma = 40.0
        print(f"    Noise level: σ = {noise_sigma}")

        noisy_image = add_gaussian_noise(image_array_normalized, sigma=noise_sigma)

        psnr_noisy_val, ssim_noisy_val = calculate_metrics(image_array_normalized, noisy_image, data_range=255.0)

        print(f"    Noisy - PSNR: {psnr_noisy_val:.2f} dB, SSIM: {ssim_noisy_val:.4f}")

        print(f"\n[5] Applying mean filter...")
        kernel_size = 5
        print(f"    Kernel size: {kernel_size}x{kernel_size}")

        filtered_image = apply_mean_filter_multichannel(noisy_image, kernel_size)

        psnr_filtered_val, ssim_filtered_val = calculate_metrics(image_array_normalized, filtered_image, data_range=255.0)

        print(f"    Filtered - PSNR: {psnr_filtered_val:.2f} dB, SSIM: {ssim_filtered_val:.4f}")
        print(f"    Improvement: +{psnr_filtered_val - psnr_noisy_val:.2f} dB")

        print(f"\n[6] Creating visualization...")

        fig, axes = plt.subplots(1, 3, figsize=(15, 5))

        axes[0].imshow(normalize_to_uint8(image_array_normalized))
        axes[0].set_title('Original (Clean)', fontsize=12, fontweight='bold')
        axes[0].axis('off')

        axes[1].imshow(normalize_to_uint8(noisy_image))
        axes[1].set_title(f'Noisy (σ={noise_sigma})\nPSNR: {psnr_noisy_val:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[1].axis('off')

        axes[2].imshow(normalize_to_uint8(filtered_image))
        axes[2].set_title(f'Mean Filtered ({kernel_size}x{kernel_size})\n'
                         f'PSNR: {psnr_filtered_val:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[2].axis('off')

        plt.tight_layout()
        plt.savefig('satellite_denoising_results.png', dpi=150, bbox_inches='tight')
        print(f"    ✓ Saved: satellite_denoising_results.png")
        plt.show()

        print("\n" + "="*70)
        print("RESULTS SUMMARY")
        print("="*70)
        print(f"  Image shape: {image_array_normalized.shape}")
        print(f"  Noise level: σ = {noise_sigma}")
        print(f"\n  {'Method':<20} {'PSNR (dB)':<12} {'SSIM':<10}")
        print(f"  {'-'*45}")
        print(f"  {'Noisy':<20} {psnr_noisy_val:<12.2f} {ssim_noisy_val:<10.4f}")
        print(f"  {'Mean Filter':<20} {psnr_filtered_val:<12.2f} {ssim_filtered_val:<10.4f}")
        print(f"  {'Improvement':<20} {f'+{psnr_filtered_val-psnr_noisy_val:.2f}':<12} {f'+{ssim_filtered_val-ssim_noisy_val:.4f}':<10}")
        print("="*70)

        return image_array_normalized, noisy_image, filtered_image, psnr_noisy_val, ssim_noisy_val, psnr_filtered_val, ssim_filtered_val, noise_sigma


    except FileNotFoundError as e:
        print(f"Error: {e}")
        print("Please ensure the image path is correct.")
        return None, None, None, None, None, None, None, None
    except Exception as e:
        print(f"An error occurred during processing: {e}")
        import traceback
        traceback.print_exc()
        return None, None, None, None, None, None, None, None

# Re-run the main pipeline from ecc1aab3
print("Re-running Mean Filter pipeline to ensure global variables are set...")
original_image_arr, noisy_image_arr, mean_filtered_image_arr, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma = main_ecc1aab3()

# Now, execute the SVD + Wavelet denoising method, using the preloaded noisy_image_arr.

#Denoised using SVD + Wavelet

import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim # Explicitly import ssim
import os

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path=None, use_sample=False, preloaded_image=None):
        """
        Load satellite data from various formats or use a preloaded image.

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration
        preloaded_image: numpy array, an already loaded image array to use directly

        Returns:
        data: numpy array, satellite data
        """
        if preloaded_image is not None:
            print("Using preloaded image for SVD + Wavelet denoising.")
            if preloaded_image.ndim == 3 and preloaded_image.shape[2] == 3:
                gray_image = np.mean(preloaded_image, axis=-1)
                noisy_data = img_as_float(gray_image)
            elif preloaded_image.ndim == 2:
                noisy_data = img_as_float(preloaded_image)
            else:
                print(f"Warning: Preloaded image has unexpected dimensions {preloaded_image.shape}. Attempting to use as-is.")
                noisy_data = img_as_float(preloaded_image)

            self.original_data = noisy_data
            print(f"Preloaded data shape: {noisy_data.shape}")
            return noisy_data

        if use_sample:
            base_data = img_as_float(data.camera())
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)
            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            if file_path is None:
                 print("Error: file_path is required when use_sample is False and no preloaded_image is given.")
                 return None

            loaded_image_array = None
            noisy_data = None

            try:
                from osgeo import gdal
                dataset = gdal.Open(file_path, gdal.GA_ReadOnly)
                if dataset:
                    raw_array_gdal = dataset.ReadAsArray()
                    if raw_array_gdal is not None:
                        if raw_array_gdal.ndim == 3:
                            if raw_array_gdal.shape[0] < raw_array_gdal.shape[1] and raw_array_gdal.shape[0] < raw_array_gdal.shape[2]:
                                raw_array_gdal = np.moveaxis(raw_array_gdal, 0, -1)
                            loaded_image_array = np.mean(raw_array_gdal, axis=-1)
                        elif raw_array_gdal.ndim == 2:
                            loaded_image_array = raw_array_gdal

                        if loaded_image_array is not None:
                            noisy_data = img_as_float(loaded_image_array)
                            print(f"Loaded satellite data with GDAL from: {file_path}")
                            print(f"Data shape: {noisy_data.shape}")
                            self.original_data = noisy_data
                            return noisy_data
                    print(f"GDAL could not process data from {file_path}.")
                else:
                    print(f"GDAL could not open {file_path}.")
            except ImportError:
                print("GDAL not available.")
            except Exception as e:
                print(f"GDAL error during loading: {e}")

            if noisy_data is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img_tif = tifffile.imread(file_path)
                    if img_tif is not None:
                         if img_tif.ndim == 3:
                            if img_tif.shape[0] < img_tif.shape[2]:
                                img_tif = np.moveaxis(img_tif, 0, -1)
                            img_tif = np.mean(img_tif, axis=-1)
                         loaded_image_array = img_tif

                         noisy_data = img_as_float(loaded_image_array)
                         print(f"Loaded satellite data with tifffile from: {file_path}")
                         print(f"Data shape: {noisy_data.shape}")
                         self.original_data = noisy_data
                         return noisy_data
                except Exception as e:
                    print(f"tifffile error during loading: {e}")

            if noisy_data is None:
                try:
                    img_cv = cv2.imread(file_path)
                    if img_cv is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img_cv.shape) == 3:
                        img_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                    loaded_image_array = img_cv

                    noisy_data = img_as_float(loaded_image_array)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")
                    self.original_data = noisy_data
                    return noisy_data
                except Exception as e:
                    print(f"OpenCV error during loading: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            print(f"Failed to load image from {file_path} with any available library.")
            return None

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            U, s, Vt = svd(data, full_matrices=False)
            rank = int(len(s) * rank_ratio)
            s_truncated = s.copy()
            s_truncated[rank:] = 0
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    block = data[i:i+block_size, j:j+block_size]
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0
                    denoised_block = U @ np.diag(s_truncated) @ Vt
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)
        sigma = self.estimate_noise_level(data)
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:
                level_thresh = 3 * sigma
                thresholds.append([level_thresh] * 3)

        coeffs_thresholded = [coeffs[0].copy()]

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0:
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0:
             snr = float('inf')
        else:
            snr = np.nan

        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             if original_norm.shape[2] in [1, 3, 4]:
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else:
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            coeffs = results['wavelet_coeffs']
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off')


        plt.tight_layout()
        plt.show()

        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}")
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf'))
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

def main_svd_wavelet_denoising():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    global original_image_arr, noisy_image_arr # Access the global variable

    if 'original_image_arr' not in globals() or original_image_arr is None:
        print("Error: original_image_arr is not available. Please ensure the mean filter cell was run and variables are accessible.")
        return

    # The SVDWaveletDenoiser expects a grayscale image normalized to 0-1.
    # original_image_arr is the 'clean' normalized RGB image (0-255).
    # We need to create a noisy version of this 'clean' image, and then pass it to the denoiser.
    # The previous cell `ecc1aab3` generates `noisy_image_arr` which is normalized RGB 0-255.
    # We convert `noisy_image_arr` to grayscale and normalize to 0-1 for this denoiser.

    if noisy_image_arr.ndim == 3 and noisy_image_arr.shape[2] == 3:
        grayscale_noisy_image = np.mean(noisy_image_arr, axis=-1) # Convert to grayscale
    else:
        grayscale_noisy_image = noisy_image_arr # Assume it's already grayscale

    # Normalize to 0-1 range
    satellite_data_for_denoising = img_as_float(grayscale_noisy_image)

    # Also provide the clean grayscale version for metrics calculation within the denoiser, if available
    if original_image_arr.ndim == 3 and original_image_arr.shape[2] == 3:
        grayscale_original_image = np.mean(original_image_arr, axis=-1)
    else:
        grayscale_original_image = original_image_arr

    clean_data_for_metrics = img_as_float(grayscale_original_image)

    print(f"Prepared noisy image (grayscale, normalized) for SVD + Wavelet. Shape: {satellite_data_for_denoising.shape}")

    if satellite_data_for_denoising is None:
        print("Failed to prepare satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data_for_denoising,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Update the 'original' in results to be the clean grayscale 0-1 image for accurate metrics
    results['original'] = clean_data_for_metrics

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(clean_data_for_metrics) # Pass clean data for comparison metrics

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try:
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data_for_denoising, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                clean_data_for_metrics, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")


# Run the SVD + Wavelet denoising pipeline
main_svd_wavelet_denoising()

```

```python
import os

# Define the expected path of the cloned repository and the image file
repo_path = "/content/drive/MyDrive/Satellite Project with Debojyoti Banerjee/all_band_vskp.tif"
image_file_path = os.path.join(repo_path, "2_A.bmp")

    if not os.path.isdir(repo_path):
        print(f"Failed to clone repository to {repo_path}. Please check the git clone command and network connectivity.")
    else:
        print("Repository cloned successfully.")

# Check if the image file exists after ensuring the repository is present
if not os.path.exists(image_file_path):
    print(f"Image file not found at {image_file_path} after cloning. Listing files in the repository directory:")
    # List files in the cloned directory to help diagnose
    if os.path.isdir(repo_path):
        !ls -R {repo_path}
    else:
        print("Repository directory does not exist.")
else:
    print(f"Image file found at {image_file_path}.")

# If the image file is now confirmed to exist, re-run the main processing code
# by calling the main_ecc1aab3 and then main_svd_wavelet_denoising functions
# as defined in the previous attempts to ensure global variables are set.
# I will replicate the setup needed for these functions to run.

# --- Start of re-defining ecc1aab3's components and main_ecc1aab3 function ---

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import uniform_filter
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
import warnings
warnings.filterwarnings('ignore')

def install_dependencies():
    """Install required libraries for GeoTIFF processing."""
    import subprocess
    import sys

    print("Installing required libraries...")
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install",
                              "rasterio", "-q"])
        print("Successfully installed rasterio")
    except:
        print("Could not install rasterio. Trying alternative method...")


def load_geotiff_rasterio(image_path):
    """
    Load GeoTIFF using rasterio (RECOMMENDED for satellite data).
    """
    try:
        import rasterio

        print(f"Loading with rasterio...")
        with rasterio.open(image_path) as src:
            image = src.read()  # Shape: (bands, height, width)
            n_bands = src.count
            height = src.height
            width = src.width
            dtype = src.dtypes[0]

            print(f"Loaded GeoTIFF:")
            print(f"    - Bands: {n_bands}")
            print(f"    - Size: {width} x {height}")
            print(f"    - Data type: {dtype}")
            print(f"    - CRS: {src.crs}")

            image = np.moveaxis(image, 0, -1)

            return image, True

    except ImportError:
        print("rasterio not available")
        return None, False
    except Exception as e:
        print(f"rasterio error: {e}")
        return None, False


def load_geotiff_gdal(image_path):
    """
    Load GeoTIFF using GDAL (fallback method).
    """
    try:
        from osgeo import gdal

        print(f"  Loading with GDAL...")
        dataset = gdal.Open(image_path, gdal.GA_ReadOnly)

        if dataset is None:
            return None, False

        n_bands = dataset.RasterCount
        width = dataset.RasterXSize
        height = dataset.RasterYSize

        print(f"Loaded GeoTIFF:")
        print(f"    - Bands: {n_bands}")
        print(f"    - Size: {width} x {height}")

        bands = []
        for i in range(1, n_bands + 1):
            band = dataset.GetRasterBand(i)
            bands.append(band.ReadAsArray())

        image = np.stack(bands, axis=-1)

        dataset = None  # Close dataset
        return image, True

    except ImportError:
        print("GDAL not available")
        return None, False
    except Exception as e:
        print(f"GDAL error: {e}")
        return None, False


def load_geotiff_tifffile(image_path):
    """
    Load TIFF using tifffile library (works for some TIFFs).
    """
    try:
        import tifffile

        print(f"  Loading with tifffile...")
        image = tifffile.imread(image_path)

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        if len(image.shape) == 3:
            if image.shape[0] < image.shape[2]:
                image = np.moveaxis(image, 0, -1)

        return image, True

    except Exception as e:
        print(f"tifffile error: {e}")
        return None, False


def load_geotiff_opencv(image_path):
    """
    Load TIFF using OpenCV (limited multi-band support).
    """
    try:
        import cv2

        print(f"  Loading with OpenCV...")
        image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)

        if image is None:
            return None, False

        print(f"Loaded TIFF:")
        print(f"    - Shape: {image.shape}")
        print(f"    - Data type: {image.dtype}")

        if len(image.shape) == 3 and image.shape[2] == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        return image, True

    except Exception as e:
        print(f"OpenCV error: {e}")
        return None, False

def load_satellite_image(image_path):
    """
    Try multiple methods to load GeoTIFF satellite image.
    """
    import os

    print(f"\n[1] Loading satellite image from:")
    print(f"    {image_path}")

    if not os.path.exists(image_path):
        print(f"Error: File not found at {image_path}")
        return None

    image, success = load_geotiff_rasterio(image_path)
    if success and image is not None:
        return image

    image, success = load_geotiff_gdal(image_path)
    if success and image is not None:
        return image

    image, success = load_geotiff_tifffile(image_path)
    if success and image is not None:
        return image

    image, success = load_geotiff_opencv(image_path)
    if success and image is not None:
        return image

    if os.path.exists(image_path):
        raise RuntimeError(
            f"Could not load image from {image_path} with any available library.\n"
            "Please check the file format and consider installing rasterio: pip install rasterio"
        )
    else:
        raise FileNotFoundError(f"Image file not found at: {image_path}")


def normalize_to_uint8(image_array):
    """Normalize to 0-255 for display using percentile clipping."""
    if image_array.size == 0:
        return np.empty_like(image_array, dtype=np.uint8)

    p2, p98 = np.percentile(image_array, (2, 98))
    if p2 == p98:
        return np.full_like(image_array, 127, dtype=np.uint8)

    image_clipped = np.clip(image_array, p2, p98)
    image_normalized = ((image_clipped - p2) / (p98 - p2) * 255)
    return np.clip(image_normalized, 0, 255).astype(np.uint8)


def select_rgb_bands(image_array, rgb_indices=None):
    """Extract RGB bands from multi-band image."""
    print(f"\n[2] Extracting RGB bands...")
    print(f"    Image shape: {image_array.shape}")

    if len(image_array.shape) == 2:
        print("    Grayscale detected - converting to RGB")
        return np.stack([image_array] * 3, axis=-1)

    elif len(image_array.shape) == 3 and image_array.shape[2] == 3:
        print("    RGB image detected")
        return image_array

    elif len(image_array.shape) == 3 and image_array.shape[2] > 3:
        if rgb_indices is None:
            n_bands = image_array.shape[2]
            print(f"    Multi-band image: {n_bands} bands detected")
            print("    Trying common RGB band combinations...")

            if n_bands >= 4:
                print("    Using bands [3,2,1] for RGB (Sentinel-2/Landsat style)")
                rgb_indices = (3, 2, 1)
            else:
                print("    Using first 3 bands [0,1,2]")
                rgb_indices = (0, 1, 2)

        max_band_index = image_array.shape[2] - 1
        valid_indices = [idx for idx in rgb_indices if 0 <= idx <= max_band_index]

        if len(valid_indices) != 3:
            print(f"Warning: Could not find 3 valid RGB bands from indices {rgb_indices}. Using first 3 available bands.")
            valid_indices = list(range(min(3, image_array.shape[2])))
            if len(valid_indices) < 3:
                 raise ValueError(f"Image only has {image_array.shape[2]} bands, cannot select 3 for RGB.")

        print(f"    Selected RGB indices: {valid_indices}")
        return image_array[:, :, valid_indices]


    else:
        raise ValueError(f"Unexpected image shape: {image_array.shape}")


def add_gaussian_noise(image, sigma=25.0):
    """Add Gaussian noise."""
    image_float = image.astype(np.float64)
    noise = np.random.normal(0, sigma, image_float.shape)
    noisy_image = image_float + noise
    return noisy_image


def apply_mean_filter_multichannel(image, kernel_size=5):
    """Apply mean filter to each channel separately."""
    image_float = image.astype(np.float64)

    if len(image_float.shape) == 2:
        return uniform_filter(image_float, size=kernel_size)

    filtered_channels = []
    for i in range(image_float.shape[2]):
        filtered_channels.append(
            uniform_filter(image_float[:, :, i], size=kernel_size)
        )
    return np.stack(filtered_channels, axis=-1)


def calculate_metrics(original, processed, data_range=None):
    """Calculate PSNR and SSIM."""
    if data_range is None:
        if original.dtype == np.uint8:
            data_range = 255.0
        elif original.dtype == np.float64 or original.dtype == np.float32:
            data_range = np.max(original)

        if data_range is None or data_range <= 0:
             data_range = 1.0

    original_float = original.astype(np.float64)
    processed_float = processed.astype(np.float64)
    processed_clipped = np.clip(processed_float, original_float.min(), original_float.max())

    try:
        psnr_value = psnr(original_float, processed_clipped, data_range=data_range)
    except Exception as e:
        print(f"Error calculating PSNR: {e}")
        psnr_value = np.nan

    try:
        min_dim = min(original_float.shape[0], original_float.shape[1])
        win_size = min(7, min_dim)
        if win_size < 1:
            win_size = 1

        if len(original_float.shape) == 3:
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range,
                             channel_axis=2, win_size=win_size)
        else:
            ssim_value = ssim(original_float, processed_clipped,
                             data_range=data_range, win_size=win_size)
    except Exception as e:
        print(f"Error calculating SSIM: {e}")
        ssim_value = np.nan

    return psnr_value, ssim_value


def main_ecc1aab3():
    """Complete denoising pipeline."""

    print("="*70)
    print("SATELLITE IMAGE DENOISING - GEOTIFF COMPATIBLE (Re-run)")
    print("="*70)

    # Use the verified image file path
    image_path = image_file_path

    try:
        import os
        if not os.path.exists(image_path):
             raise FileNotFoundError(f"Image file not found at: {image_path}")

        image_array_raw = load_satellite_image(image_path)

        if image_array_raw is None:
             raise RuntimeError("Failed to load image.")

        if len(image_array_raw.shape) == 2:
             image_array = np.stack([image_array_raw]*3, axis=-1)
             print("Grayscale image loaded, converted to RGB.")
        else:
             image_array = select_rgb_bands(image_array_raw, None)

        image_array = image_array.astype(np.float64)

        print(f"\n[3] Normalizing data range...")
        max_original_value = np.max(image_array)
        if max_original_value > 0:
             image_array_normalized = image_array / max_original_value * 255.0
        else:
             image_array_normalized = image_array

        print(f"    Normalized range: [{image_array_normalized.min():.2f}, {image_array_normalized.max():.2f}]")

        print(f"\n[4] Adding Gaussian noise...")
        noise_sigma = 40.0
        print(f"    Noise level: σ = {noise_sigma}")

        noisy_image = add_gaussian_noise(image_array_normalized, sigma=noise_sigma)

        psnr_noisy_val, ssim_noisy_val = calculate_metrics(image_array_normalized, noisy_image, data_range=255.0)

        print(f"    Noisy - PSNR: {psnr_noisy_val:.2f} dB, SSIM: {ssim_noisy_val:.4f}")

        print(f"\n[5] Applying mean filter...")
        kernel_size = 5
        print(f"    Kernel size: {kernel_size}x{kernel_size}")

        filtered_image = apply_mean_filter_multichannel(noisy_image, kernel_size)

        psnr_filtered_val, ssim_filtered_val = calculate_metrics(image_array_normalized, filtered_image, data_range=255.0)

        print(f"    Filtered - PSNR: {psnr_filtered_val:.2f} dB, SSIM: {ssim_filtered_val:.4f}")
        print(f"    Improvement: +{psnr_filtered_val - psnr_noisy_val:.2f} dB")

        print(f"\n[6] Creating visualization...")

        fig, axes = plt.subplots(1, 3, figsize=(15, 5))

        axes[0].imshow(normalize_to_uint8(image_array_normalized))
        axes[0].set_title('Original (Clean)', fontsize=12, fontweight='bold')
        axes[0].axis('off')

        axes[1].imshow(normalize_to_uint8(noisy_image))
        axes[1].set_title(f'Noisy (σ={noise_sigma})\nPSNR: {psnr_noisy_val:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[1].axis('off')

        axes[2].imshow(normalize_to_uint8(filtered_image))
        axes[2].set_title(f'Mean Filtered ({kernel_size}x{kernel_size})\n'
                         f'PSNR: {psnr_filtered_val:.2f} dB',
                         fontsize=12, fontweight='bold')
        axes[2].axis('off')

        plt.tight_layout()
        plt.savefig('satellite_denoising_results.png', dpi=150, bbox_inches='tight')
        print(f"    ✓ Saved: satellite_denoising_results.png")
        plt.show()

        print("\n" + "="*70)
        print("RESULTS SUMMARY")
        print("="*70)
        print(f"  Image shape: {image_array_normalized.shape}")
        print(f"  Noise level: σ = {noise_sigma}")
        print(f"\n  {'Method':<20} {'PSNR (dB)':<12} {'SSIM':<10}")
        print(f"  {'-'*45}")
        print(f"  {'Noisy':<20} {psnr_noisy_val:<12.2f} {ssim_noisy_val:<10.4f}")
        print(f"  {'Mean Filter':<20} {psnr_filtered_val:<12.2f} {ssim_filtered_val:<10.4f}")
        print(f"  {'Improvement':<20} {f'+{psnr_filtered_val-psnr_noisy_val:.2f}':<12} {f'+{ssim_filtered_val-ssim_noisy_val:.4f}':<10}")
        print("="*70)

        return image_array_normalized, noisy_image, filtered_image, psnr_noisy_val, ssim_noisy_val, psnr_filtered_val, ssim_filtered_val, noise_sigma


    except FileNotFoundError as e:
        print(f"Error: {e}")
        print("Please ensure the image path is correct.")
        return None, None, None, None, None, None, None, None
    except Exception as e:
        print(f"An error occurred during processing: {e}")
        import traceback
        traceback.print_exc()
        return None, None, None, None, None, None, None, None

# --- End of re-defining ecc1aab3's components and main_ecc1aab3 function ---


# --- Start of re-defining oyemYZjfY-k8's components and main_svd_wavelet_denoising function ---

from skimage import img_as_float # Ensure img_as_float is imported

class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path=None, use_sample=False, preloaded_image=None):
        """
        Load satellite data from various formats or use a preloaded image.

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration
        preloaded_image: numpy array, an already loaded image array to use directly

        Returns:
        data: numpy array, satellite data
        """
        if preloaded_image is not None:
            print("Using preloaded image for SVD + Wavelet denoising.")
            if preloaded_image.ndim == 3 and preloaded_image.shape[2] == 3:
                gray_image = np.mean(preloaded_image, axis=-1)
                noisy_data = img_as_float(gray_image)
            elif preloaded_image.ndim == 2:
                noisy_data = img_as_float(preloaded_image)
            else:
                print(f"Warning: Preloaded image has unexpected dimensions {preloaded_image.shape}. Attempting to use as-is.")
                noisy_data = img_as_float(preloaded_image)

            self.original_data = noisy_data
            print(f"Preloaded data shape: {noisy_data.shape}")
            return noisy_data

        if use_sample:
            base_data = img_as_float(data.camera())
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)
            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            if file_path is None:
                 print("Error: file_path is required when use_sample is False and no preloaded_image is given.")
                 return None

            loaded_image_array = None
            noisy_data = None

            try:
                from osgeo import gdal
                dataset = gdal.Open(file_path, gdal.GA_ReadOnly)
                if dataset:
                    raw_array_gdal = dataset.ReadAsArray()
                    if raw_array_gdal is not None:
                        if raw_array_gdal.ndim == 3:
                            if raw_array_gdal.shape[0] < raw_array_gdal.shape[1] and raw_array_gdal.shape[0] < raw_array_gdal.shape[2]:
                                raw_array_gdal = np.moveaxis(raw_array_gdal, 0, -1)
                            loaded_image_array = np.mean(raw_array_gdal, axis=-1)
                        elif raw_array_gdal.ndim == 2:
                            loaded_image_array = raw_array_gdal

                        if loaded_image_array is not None:
                            noisy_data = img_as_float(loaded_image_array)
                            print(f"Loaded satellite data with GDAL from: {file_path}")
                            print(f"Data shape: {noisy_data.shape}")
                            self.original_data = noisy_data
                            return noisy_data
                    print(f"GDAL could not process data from {file_path}.")
                else:
                    print(f"GDAL could not open {file_path}.")
            except ImportError:
                print("GDAL not available.")
            except Exception as e:
                print(f"GDAL error during loading: {e}")

            if noisy_data is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img_tif = tifffile.imread(file_path)
                    if img_tif is not None:
                         if img_tif.ndim == 3:
                            if img_tif.shape[0] < img_tif.shape[2]:
                                img_tif = np.moveaxis(img_tif, 0, -1)
                            img_tif = np.mean(img_tif, axis=-1)
                         loaded_image_array = img_tif

                         noisy_data = img_as_float(loaded_image_array)
                         print(f"Loaded satellite data with tifffile from: {file_path}")
                         print(f"Data shape: {noisy_data.shape}")
                         self.original_data = noisy_data
                         return noisy_data
                except Exception as e:
                    print(f"tifffile error during loading: {e}")

            if noisy_data is None:
                try:
                    img_cv = cv2.imread(file_path)
                    if img_cv is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img_cv.shape) == 3:
                        img_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                    loaded_image_array = img_cv

                    noisy_data = img_as_float(loaded_image_array)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")
                    self.original_data = noisy_data
                    return noisy_data
                except Exception as e:
                    print(f"OpenCV error during loading: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            print(f"Failed to load image from {file_path} with any available library.")
            return None

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            U, s, Vt = svd(data, full_matrices=False)
            rank = int(len(s) * rank_ratio)
            s_truncated = s.copy()
            s_truncated[rank:] = 0
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    block = data[i:i+block_size, j:j+block_size]
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0
                    denoised_block = U @ np.diag(s_truncated) @ Vt
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)
        sigma = self.estimate_noise_level(data)
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:
                level_thresh = 3 * sigma
                thresholds.append([level_thresh] * 3)

        coeffs_thresholded = [coeffs[0].copy()]

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0:
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0:
             snr = float('inf')
        else:
            snr = np.nan

        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             if original_norm.shape[2] in [1, 3, 4]:
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else:
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            coeffs = results['wavelet_coeffs']
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off')


        plt.tight_layout()
        plt.show()

        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}")
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf'))
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

def main_svd_wavelet_denoising():
    """
    Demonstrate SVD + Wavelet denoising on satellite data
    """
    print("=== SVD + WAVELET SATELLITE DATA DENOISING ===\n")

    denoiser = SVDWaveletDenoiser(wavelet='db8', mode='symmetric')

    global original_image_arr, noisy_image_arr # Access the global variable

    if 'original_image_arr' not in globals() or original_image_arr is None:
        print("Error: original_image_arr is not available. Please ensure the mean filter cell was run and variables are accessible.")
        return

    # The SVDWaveletDenoiser expects a grayscale image normalized to 0-1.
    # original_image_arr is the 'clean' normalized RGB image (0-255).
    # We need to create a noisy version of this 'clean' image, and then pass it to the denoiser.
    # The previous cell `ecc1aab3` generates `noisy_image_arr` which is normalized RGB 0-255.
    # We convert `noisy_image_arr` to grayscale and normalize to 0-1 for this denoiser.

    if noisy_image_arr.ndim == 3 and noisy_image_arr.shape[2] == 3:
        grayscale_noisy_image = np.mean(noisy_image_arr, axis=-1) # Convert to grayscale
    else:
        grayscale_noisy_image = noisy_image_arr # Assume it's already grayscale

    # Normalize to 0-1 range
    satellite_data_for_denoising = img_as_float(grayscale_noisy_image)

    # Also provide the clean grayscale version for metrics calculation within the denoiser, if available
    if original_image_arr.ndim == 3 and original_image_arr.shape[2] == 3:
        grayscale_original_image = np.mean(original_image_arr, axis=-1)
    else:
        grayscale_original_image = original_image_arr

    clean_data_for_metrics = img_as_float(grayscale_original_image)

    print(f"Prepared noisy image (grayscale, normalized) for SVD + Wavelet. Shape: {satellite_data_for_denoising.shape}")

    if satellite_data_for_denoising is None:
        print("Failed to prepare satellite data. Exiting.")
        return

    # Apply hybrid denoising
    results = denoiser.hybrid_svd_wavelet_denoise(
        satellite_data_for_denoising,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # Update the 'original' in results to be the clean grayscale 0-1 image for accurate metrics
    results['original'] = clean_data_for_metrics

    # Visualize results
    denoiser.visualize_denoising_results(results, show_coefficients=True)

    # Compare different methods
    comparison_results = denoiser.compare_methods(clean_data_for_metrics) # Pass clean data for comparison metrics

    # Test different wavelets
    print("\n=== TESTING DIFFERENT WAVELETS ===")
    wavelets = ['db4', 'db8', 'haar', 'bior2.2', 'coif2']

    for wav in wavelets:
        try:
            denoiser_wav = SVDWaveletDenoiser(wavelet=wav)
            results_wav = denoiser_wav.hybrid_svd_wavelet_denoise(
                satellite_data_for_denoising, svd_rank_ratio=0.8, wavelet_method='bayes'
            )
            metrics_wav = denoiser_wav.calculate_quality_metrics(
                clean_data_for_metrics, results_wav['final_denoised']
            )
            print(f"{wav}: PSNR = {metrics_wav['PSNR']:.2f} dB, SSIM = {metrics_wav['SSIM']:.4f}")
        except Exception as e:
            print(f"Testing wavelet {wav} failed: {e}")

# --- End of re-defining oyemYZjfY-k8's components and main_svd_wavelet_denoising function ---


# Run the mean filter pipeline first to set up global variables
print("Re-running Mean Filter pipeline to ensure global variables are set...")
global original_image_arr, noisy_image_arr, mean_filtered_image_arr, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma
original_image_arr, noisy_image_arr, mean_filtered_image_arr, psnr_noisy, ssim_noisy, psnr_filtered, ssim_filtered, noise_sigma = main_ecc1aab3()

# Only proceed with SVD + Wavelet if the mean filter pipeline succeeded in loading data
if original_image_arr is not None and noisy_image_arr is not None:
    print("\nMean Filter pipeline completed successfully. Proceeding with SVD + Wavelet denoising...")
    main_svd_wavelet_denoising()
else:
    print("\nSkipping SVD + Wavelet denoising due to failure in Mean Filter pipeline.")
```

## Analysis of Computational Cost vs. Denoising Performance

Based on the denoising methods applied (Mean, Median, SVD + Wavelet, Wiener + Wavelet, and Enhanced Denoising), here's an analysis of their trade-offs between computational cost and denoising performance:

### 1. Mean Filter (Cell `ecc1aab3`)

*   **Computational Cost:** **Very Low**. It involves simple arithmetic operations (summation and division) over a small window for each pixel. It scales linearly with the number of pixels and quadratically with the kernel size, making it one of the fastest denoising methods.
*   **Denoising Performance:** **Moderate to Low**. While effective at reducing Gaussian noise, it tends to blur edges and fine details significantly, leading to a loss of image sharpness. The PSNR achieved was **28.75 dB** and SSIM **0.6444** (on the 0-255 scale noisy image).

### 2. Median Filter (Cell `yhu2mI-KqBQi`)

*   **Computational Cost:** **Moderate**. It requires sorting pixel values within a window, which is computationally more intensive than averaging. The cost typically scales roughly with `N * log(N)` where N is the number of pixels in the kernel, making it slower than the mean filter.
*   **Denoising Performance:** **Moderate to Good**. It is generally better at preserving edges than the mean filter and is particularly effective against salt-and-pepper noise. It can still smooth out fine textures. The PSNR achieved was **23.01 dB** and SSIM **0.4925** (on the 0-255 scale noisy image).

### 3. SVD + Wavelet Denoising (Cell `bea6f720`)

*   **Computational Cost:** **High**. This method combines wavelet decomposition/reconstruction with Singular Value Decomposition (SVD). Wavelet transforms involve convolutions, and SVD itself can be computationally intensive, especially for large image patches. While patch-based SVD is used to manage this, the overall process is significantly more complex than simple spatial filters.
*   **Denoising Performance:** **Expected Good, Observed Modest.** This hybrid approach is designed to leverage the multi-resolution analysis of wavelets and the noise reduction capabilities of SVD to preserve details effectively. However, for the specific Gaussian noise added to `noisy_image_arr` (which was initially normalized to 0-255 and then downscaled to 0-1 for this method), the observed PSNR was **12.99 dB** and SSIM **0.2357**. This is lower than the Mean Filter's performance on the same noisy image (or a derivative of it). This discrepancy likely indicates that the noise characteristics or data scaling used in the SVD+Wavelet implementation were not ideally suited for the specific noise introduced in `ecc1aab3`, or that PSNR calculation against a normalized 0-1 range with a `data_range=1.0` makes direct comparison to 0-255 range PSNR difficult. In general, SVD+Wavelet is expected to outperform simpler filters for complex noise and detail preservation when properly tuned.

### 4. Wiener Filter + Wavelet Thresholding (Cell `8e54f2a6`)

*   **Computational Cost:** **High**. The Wiener filter typically operates in the frequency domain, requiring Fast Fourier Transforms (FFTs) and inverse FFTs, which are computationally expensive but efficient for global image processing. Combining this with wavelet transforms further adds to the cost.
*   **Denoising Performance:** **Good (Theoretical)**. The Wiener filter is optimal for stationary linear filters contaminated by additive noise, especially when signal and noise power spectral densities are known. When combined with wavelets, it can be very effective at removing noise while minimizing blurring and preserving important image features. While no direct comparable metrics were obtained from the notebook for this exact combination on the same noisy image from `ecc1aab3`, its theoretical performance is high for suitable noise models.

### 5. Enhanced Satellite Denoising (Cell `a9833787`)

*   **Computational Cost:** **Very High**. This is a comprehensive pipeline integrating multiple advanced techniques: multi-scale noise estimation, adaptive SVD rank selection (potentially block-wise), multi-wavelet adaptive thresholding, edge-preserving bilateral filtering, and contrast enhancement. Each step contributes significantly to the overall computational load.
*   **Denoising Performance:** **Potentially Very Good (Theoretical)**. This method aims for state-of-the-art denoising by adaptively handling various noise types and maximizing detail preservation and visual quality. The synthetic demo provided a PSNR of **16.92 dB**, which indicates good performance for its specific synthetic noise model, but again, is not directly comparable to the other methods without consistent testing conditions.

### Overall Trade-off

The analysis confirms a general trade-off:

*   **Simple filters (Mean, Median):** Offer **low computational cost** but provide **moderate denoising performance**, often with compromises on image detail and sharpness.
*   **Hybrid methods (SVD + Wavelet, Wiener + Wavelet):** Involve **higher computational costs** due to more complex mathematical operations (transforms, decompositions). They are designed to deliver **superior denoising performance** by preserving more intricate image details and adapting to different noise characteristics, although tuning and consistent experimental setup are crucial for realizing this potential.
*   **Enhanced/Comprehensive methods:** Have the **highest computational costs** due to their multi-stage, adaptive nature. They strive for the **best possible denoising performance** across a wider range of scenarios, aiming to minimize resolution loss and maximize visual quality.

The observed lower PSNR for SVD+Wavelet compared to the Mean Filter on the *same noisy image* in this specific execution highlights the importance of consistent normalization, data range assumptions, and proper parameter tuning when evaluating complex denoising algorithms, especially when comparing across implementations or different noise generation/metric calculation contexts.

I will get a list of all image files from the 'extracted_frames' directory, select three representative frames (first, middle, last), and then load, convert, display, and title them using Matplotlib.



```python
import os
import cv2
import matplotlib.pyplot as plt

# Define the directory where frames are extracted
output_dir = "extracted_frames"

# 1. Get a list of all image files in the 'extracted_frames' directory.
# Filter for common image extensions, assuming they are .png as per previous cells.
frame_files = sorted([f for f in os.listdir(output_dir) if f.endswith(('.png', '.jpg', '.jpeg'))])

if not frame_files:
    print(f"No image files found in '{output_dir}'. Please ensure frames were extracted successfully.")
else:
    print(f"Found {len(frame_files)} frames in '{output_dir}'.")

    # 2. Select 3 to 5 image file paths from this list to represent a sample.
    # Choose frames that are sufficiently spaced apart (e.g., first, middle, last) to show variation.
    num_frames_to_display = min(5, len(frame_files)) # Display up to 5 frames, or fewer if less are available

    selected_frame_indices = []
    if num_frames_to_display == 1:
        selected_frame_indices.append(len(frame_files) // 2) # Just the middle one
    elif num_frames_to_display > 1:
        # Ensure we pick distinct indices
        spacing = len(frame_files) // (num_frames_to_display + 1)
        if spacing == 0:
             spacing = 1

        # Use a more robust way to select evenly spaced frames including start and end
        if len(frame_files) > 0:
            selected_frame_indices.append(0) # First frame
            if len(frame_files) > 1:
                for i in range(1, num_frames_to_display - 1):
                    idx = i * (len(frame_files) - 1) // (num_frames_to_display - 1)
                    if idx not in selected_frame_indices:
                        selected_frame_indices.append(idx)
                if (len(frame_files) - 1) not in selected_frame_indices:
                    selected_frame_indices.append(len(frame_files) - 1) # Last frame

        # If we still don't have enough frames, fill with more middle ones
        while len(selected_frame_indices) < num_frames_to_display and len(selected_frame_indices) < len(frame_files):
            mid_idx = len(frame_files) // 2
            if mid_idx not in selected_frame_indices:
                selected_frame_indices.append(mid_idx)
            else:
                # Try adjacent indices if middle is already picked
                for offset in range(1, len(frame_files)):
                    if mid_idx - offset >= 0 and (mid_idx - offset) not in selected_frame_indices:
                        selected_frame_indices.append(mid_idx - offset)
                        break
                    if mid_idx + offset < len(frame_files) and (mid_idx + offset) not in selected_frame_indices:
                        selected_frame_indices.append(mid_idx + offset)
                        break
            selected_frame_indices.sort() # Keep them sorted for consistency

    selected_frame_paths = [os.path.join(output_dir, frame_files[i]) for i in selected_frame_indices]
    selected_frame_names = [frame_files[i] for i in selected_frame_indices]

    # 3. Create a Matplotlib figure with subplots to display each selected image.
    fig, axes = plt.subplots(1, num_frames_to_display, figsize=(5 * num_frames_to_display, 5))

    # Ensure axes is an array even for a single subplot
    if num_frames_to_display == 1:
        axes = [axes]

    print(f"Displaying {len(selected_frame_paths)} sample frames...")

    for i, frame_path in enumerate(selected_frame_paths):
        # a. Load the image using `cv2.imread()`.
        img_bgr = cv2.imread(frame_path)

        if img_bgr is None:
            print(f"Warning: Could not load image from {frame_path}")
            continue

        # b. Convert the image from BGR to RGB color format using `cv2.cvtColor()`.
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

        # c. Display the image in a subplot.
        axes[i].imshow(img_rgb)
        # d. Add a title to each subplot indicating the frame's filename.
        axes[i].set_title(selected_frame_names[i])
        # e. Turn off the axis for each subplot.
        axes[i].axis('off')

    # 5. Adjust the layout to prevent titles and images from overlapping.
    plt.tight_layout()
    # 6. Display the figure.
    plt.show()

    print("Sample frames displayed.")
```

## Estimate Noise Level and Visualize Distribution

### Subtask:
Estimate the noise standard deviation (sigma) from a relatively flat or homogeneous region of the `sample_frame_gray_normalized` image using a robust method (e.g., Median Absolute Deviation or from wavelet coefficients). Then, visualize the noise distribution by plotting a histogram of pixel intensity differences in that selected region to help determine the noise type.


**Reasoning**:
The subtask requires estimating noise and visualizing its distribution. To do this, I will first display the `sample_frame_gray_normalized` image to visually identify a flat region. Then, I will manually define a patch from this region, apply a mean filter for local smoothing, calculate pixel differences, estimate noise sigma using MAD, and finally plot a histogram of these differences to visualize the noise distribution.



```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.ndimage import uniform_filter

# Assuming sample_frame_gray_normalized is available from the previous step

# 1. Display sample_frame_gray_normalized to visually identify a flat region.
plt.figure(figsize=(10, 8))
plt.imshow(sample_frame_gray_normalized, cmap='gray')
plt.title('Original Frame (Grayscale, Normalized) - Select a flat region')
plt.axis('off')
plt.show()

print("Please visually identify a relatively flat region in the image displayed above.")
print("For this example, we will manually select a region. You can adjust these coordinates.")

# Manual selection of a patch (adjust coordinates and size as needed after visual inspection)
# Example coordinates for a 50x50 patch starting at row 100, column 150
start_row, end_row = 100, 150
start_col, end_col = 150, 200

# Ensure the selected region is within image bounds
h, w = sample_frame_gray_normalized.shape
if end_row > h or end_col > w:
    print("Warning: Selected patch exceeds image boundaries. Adjusting to fit.")
    end_row = min(end_row, h)
    end_col = min(end_col, w)
    start_row = max(0, end_row - 50) # Ensure patch size is at least roughly 50x50
    start_col = max(0, end_col - 50)

# 2. Manually define a small rectangular patch from the identified homogeneous region.
patch = sample_frame_gray_normalized[start_row:end_row, start_col:end_col]

print(f"Selected patch from coordinates: ({start_row}:{end_row}, {start_col}:{end_col})")
print(f"Patch shape: {patch.shape}")

# Visualize the selected patch
plt.figure(figsize=(4, 4))
plt.imshow(patch, cmap='gray')
plt.title('Selected Homogeneous Patch')
plt.axis('off')
plt.show()

# 3. Apply a small mean filter (e.g., 3x3 kernel) to the extracted patch.
kernel_size = 3
smoothed_patch = uniform_filter(patch, size=kernel_size)

# 4. Calculate the pixel intensity differences.
differences = (patch - smoothed_patch).flatten()

# 5. Estimate the noise standard deviation (sigma) using the Median Absolute Deviation (MAD).
mad = np.median(np.abs(differences - np.median(differences)))
sigma_estimate = mad / 0.6745

print(f"Estimated noise standard deviation (sigma): {sigma_estimate:.6f}")

# 6. Plot a histogram of the calculated pixel intensity differences.
plt.figure(figsize=(8, 5))
plt.hist(differences, bins=50, density=True, color='skyblue', edgecolor='black')
plt.title(f'Histogram of Pixel Intensity Differences (Estimated Noise σ: {sigma_estimate:.4f})')
plt.xlabel('Pixel Intensity Difference')
plt.ylabel('Probability Density')
plt.grid(True)
plt.show()

print("Noise estimation and visualization complete.")
```

## Prepare for Batch Denoising

### Subtask:
Initialize the SVDWaveletDenoiser class and get a list of all image files from the 'extracted_frames' directory.


**Reasoning**:
I will initialize the SVDWaveletDenoiser class and prepare a sorted list of image files from the 'extracted_frames' directory for batch processing. This ensures the denoising object is ready and the input files are organized.



```python
import numpy as np
import matplotlib.pyplot as plt
import pywt
import cv2
from scipy.linalg import svd
from skimage import data, img_as_float, exposure
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim
import os

# SVDWaveletDenoiser class definition (copied from cell fd0b0659 to ensure it's available)
class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        """
        Initialize SVD and Wavelet denoising system

        Parameters:
        wavelet: str, wavelet type ('db4', 'db8', 'haar', 'bior2.2', 'coif2')
        mode: str, boundary condition mode ('symmetric', 'periodization', 'zero')
        """
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path=None, use_sample=False, preloaded_image=None):
        """
        Load satellite data from various formats or use a preloaded image.

        Parameters:
        file_path: str, path to satellite data file
        use_sample: bool, use sample data for demonstration
        preloaded_image: numpy array, an already loaded image array to use directly

        Returns:
        data: numpy array, satellite data
        """
        if preloaded_image is not None:
            print("Using preloaded image for SVD + Wavelet denoising.")
            if preloaded_image.ndim == 3 and preloaded_image.shape[2] == 3:
                gray_image = np.mean(preloaded_image, axis=-1)
                noisy_data = img_as_float(gray_image)
            elif preloaded_image.ndim == 2:
                noisy_data = img_as_float(preloaded_image)
            else:
                print(f"Warning: Preloaded image has unexpected dimensions {preloaded_image.shape}. Attempting to use as-is.")
                noisy_data = img_as_float(preloaded_image)

            self.original_data = noisy_data
            print(f"Preloaded data shape: {noisy_data.shape}")
            return noisy_data

        if use_sample:
            base_data = img_as_float(data.camera())
            gaussian_noise = np.random.normal(0, 0.08, base_data.shape)
            salt_pepper = np.random.random(base_data.shape)
            salt_pepper[salt_pepper < 0.05] = -0.5
            salt_pepper[salt_pepper > 0.95] = 0.5
            salt_pepper[(salt_pepper >= 0.05) & (salt_pepper <= 0.95)] = 0
            noisy_data = base_data + gaussian_noise + salt_pepper * 0.1
            noisy_data = np.clip(noisy_data, 0, 1)
            print(f"Generated sample noisy satellite data: {noisy_data.shape}")

        else:
            if file_path is None:
                 print("Error: file_path is required when use_sample is False and no preloaded_image is given.")
                 return None

            loaded_image_array = None
            noisy_data = None

            try:
                from osgeo import gdal
                dataset = gdal.Open(file_path, gdal.GA_ReadOnly)
                if dataset:
                    raw_array_gdal = dataset.ReadAsArray()
                    if raw_array_gdal is not None:
                        if raw_array_gdal.ndim == 3:
                            if raw_array_gdal.shape[0] < raw_array_gdal.shape[1] and raw_array_gdal.shape[0] < raw_array_gdal.shape[2]:
                                raw_array_gdal = np.moveaxis(raw_array_gdal, 0, -1)
                            loaded_image_array = np.mean(raw_array_gdal, axis=-1)
                        elif raw_array_gdal.ndim == 2:
                            loaded_image_array = raw_array_gdal

                        if loaded_image_array is not None:
                            noisy_data = img_as_float(loaded_image_array)
                            print(f"Loaded satellite data with GDAL from: {file_path}")
                            print(f"Data shape: {noisy_data.shape}")
                            self.original_data = noisy_data
                            return noisy_data
                    print(f"GDAL could not process data from {file_path}.")
                else:
                    print(f"GDAL could not open {file_path}.")
            except ImportError:
                print("GDAL not available.")
            except Exception as e:
                print(f"GDAL error during loading: {e}")

            if noisy_data is None and file_path.lower().endswith(('.tif', '.tiff')):
                try:
                    import tifffile
                    img_tif = tifffile.imread(file_path)
                    if img_tif is not None:
                         if img_tif.ndim == 3:
                            if img_tif.shape[0] < img_tif.shape[2]:
                                img_tif = np.moveaxis(img_tif, 0, -1)
                            img_tif = np.mean(img_tif, axis=-1)
                         loaded_image_array = img_tif

                         noisy_data = img_as_float(loaded_image_array)
                         print(f"Loaded satellite data with tifffile from: {file_path}")
                         print(f"Data shape: {noisy_data.shape}")
                         self.original_data = noisy_data
                         return noisy_data
                except Exception as e:
                    print(f"tifffile error during loading: {e}")

            if noisy_data is None:
                try:
                    img_cv = cv2.imread(file_path)
                    if img_cv is None:
                        print(f"Error loading image with OpenCV: {file_path}")
                        return None
                    if len(img_cv.shape) == 3:
                        img_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                    loaded_image_array = img_cv

                    noisy_data = img_as_float(loaded_image_array)
                    print(f"Loaded satellite data with OpenCV from: {file_path}")
                    print(f"Data shape: {noisy_data.shape}")
                    self.original_data = noisy_data
                    return noisy_data
                except Exception as e:
                    print(f"OpenCV error during loading: {e}")
                    print(f"Error loading image with OpenCV: {file_path}")
                    return None

            print(f"Failed to load image from {file_path} with any available library.")
            return None

    def estimate_noise_level(self, data):
        """
        Estimate noise level using robust median estimator

        Parameters:
        data: numpy array, input data

        Returns:
        sigma: float, estimated noise standard deviation
        """
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        """
        Apply SVD-based denoising by truncating smaller singular values

        Parameters:
        data: numpy array, input noisy data
        rank_ratio: float, ratio of singular values to keep (0-1)
        block_size: int, size of blocks for patch-based SVD

        Returns:
        denoised: numpy array, SVD denoised data
        """
        print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            U, s, Vt = svd(data, full_matrices=False)
            rank = int(len(s) * rank_ratio)
            s_truncated = s.copy()
            s_truncated[rank:] = 0
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    block = data[i:i+block_size, j:j+block_size]
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0
                    denoised_block = U @ np.diag(s_truncated) @ Vt
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        """
        Apply wavelet thresholding denoising

        Parameters:
        data: numpy array, input data
        threshold_method: str, thresholding method ('bayes', 'sure', 'minimax', 'manual')
        threshold_mode: str, thresholding mode ('soft', 'hard')
        levels: int, number of decomposition levels

        Returns:
        denoised: numpy array, wavelet denoised data
        coeffs_original: list, original wavelet coefficients
        coeffs_thresholded: list, thresholded coefficients
        """
        print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)
        sigma = self.estimate_noise_level(data)
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:
                level_thresh = 3 * sigma
                thresholds.append([level_thresh] * 3)

        coeffs_thresholded = [coeffs[0].copy()]

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        """
        Apply hybrid SVD + Wavelet denoising

        Parameters:
        data: numpy array, input noisy data
        svd_first: bool, apply SVD before wavelet (True) or after (False)
        svd_rank_ratio: float, SVD rank ratio
        wavelet_method: str, wavelet threshold method
        wavelet_mode: str, wavelet threshold mode
        wavelet_levels: int, wavelet decomposition levels
        svd_block_size: int, block size for SVD

        Returns:
        dict: results containing denoised data and intermediate steps
        """
        print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        """
        Calculate denoising quality metrics

        Parameters:
        original: numpy array, original (clean) data
        denoised: numpy array, denoised data

        Returns:
        dict: quality metrics
        """
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0:
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0:
             snr = float('inf')
        else:
            snr = np.nan

        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             if original_norm.shape[2] in [1, 3, 4]:
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else:
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        """
        Visualize denoising results with before/after comparison
        """
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            coeffs = results['wavelet_coeffs']
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off')


        plt.tight_layout()
        plt.show()

        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}")
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        """
        Compare different SVD and wavelet parameter combinations

        Parameters:
        data: numpy array, input data
        methods_config: list, configurations to test
        """
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
                  f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            print(f"  SNR: {metrics['SNR']:.2f} dB")
            print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf'))
            print(f"\nBest method based on PSNR: {best_method['config']}")
            print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# 1. Initialize an instance of the `SVDWaveletDenoiser` class.
svd_wavelet_denoiser = SVDWaveletDenoiser()
print("Initialized SVDWaveletDenoiser class.")

# 2. Define the path to the 'extracted_frames' directory.
extracted_frames_dir = "extracted_frames"
print(f"Defined extracted frames directory: {extracted_frames_dir}")

# 3. List all files in the 'extracted_frames' directory.
all_files = os.listdir(extracted_frames_dir)

# 4. Filter this list to include only common image file extensions.
image_extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.tif', '.tiff')
image_files = [f for f in all_files if f.lower().endswith(image_extensions)]

# 5. Sort the list of image files to ensure consistent processing order.
sorted_image_files = sorted(image_files)

print(f"Found {len(sorted_image_files)} image files in '{extracted_frames_dir}'.")
```

## Apply SVD + Wavelet Denoising to All Frames

### Subtask:
Iterate through each extracted frame. For each frame: load it, convert it to grayscale and normalize it (as required by the SVDWaveletDenoiser), apply the SVD + Wavelet denoising, and then save the denoised frame to a new output directory.


**Reasoning**:
I will create a new directory for denoised frames, then iterate through each image file, load it, preprocess it, apply the SVD + Wavelet denoising, and save the result, following the provided instructions.



```python
import os
import cv2
import numpy as np
import pywt
from scipy.linalg import svd
from skimage import img_as_float
from sklearn.metrics import mean_squared_error
from skimage.metrics import structural_similarity as ssim
import matplotlib.pyplot as plt

# SVDWaveletDenoiser class definition (copied to ensure availability)
class SVDWaveletDenoiser:
    def __init__(self, wavelet='db8', mode='symmetric'):
        self.wavelet = wavelet
        self.mode = mode
        self.original_data = None
        self.denoised_data = None

    def load_satellite_data(self, file_path=None, use_sample=False, preloaded_image=None):
        if preloaded_image is not None:
            # print("Using preloaded image for SVD + Wavelet denoising.")
            if preloaded_image.ndim == 3 and preloaded_image.shape[2] == 3:
                gray_image = np.mean(preloaded_image, axis=-1)
                noisy_data = img_as_float(gray_image)
            elif preloaded_image.ndim == 2:
                noisy_data = img_as_float(preloaded_image)
            else:
                # print(f"Warning: Preloaded image has unexpected dimensions {preloaded_image.shape}. Attempting to use as-is.")
                noisy_data = img_as_float(preloaded_image)

            self.original_data = noisy_data
            # print(f"Preloaded data shape: {noisy_data.shape}")
            return noisy_data

        # Other loading logic is not directly used here but kept for completeness of the class definition.
        print("File loading is not implemented for this context.")
        return None

    def estimate_noise_level(self, data):
        coeffs = pywt.dwt2(data, self.wavelet, mode=self.mode)
        hh_coeffs = coeffs[1][2]
        sigma = np.median(np.abs(hh_coeffs)) / 0.6745

        # print(f"Estimated noise level (σ): {sigma:.6f}")
        return sigma

    def svd_denoise(self, data, rank_ratio=0.8, block_size=32):
        # print(f"Applying SVD denoising with rank ratio: {rank_ratio}")

        if block_size >= min(data.shape):
            U, s, Vt = svd(data, full_matrices=False)
            rank = int(len(s) * rank_ratio)
            s_truncated = s.copy()
            s_truncated[rank:] = 0
            denoised = U @ np.diag(s_truncated) @ Vt

        else:
            h, w = data.shape
            denoised = np.zeros_like(data)
            weight_map = np.zeros_like(data)

            step = block_size // 2

            for i in range(0, h - block_size + 1, step):
                for j in range(0, w - block_size + 1, step):
                    block = data[i:i+block_size, j:j+block_size]
                    U, s, Vt = svd(block, full_matrices=False)
                    rank = int(len(s) * rank_ratio)
                    s_truncated = s.copy()
                    s_truncated[rank:] = 0
                    denoised_block = U @ np.diag(s_truncated) @ Vt
                    denoised[i:i+block_size, j:j+block_size] += denoised_block
                    weight_map[i:i+block_size, j:j+block_size] += 1

            denoised = np.divide(denoised, weight_map,
                               out=np.zeros_like(denoised), where=weight_map!=0)

        # print("SVD denoising completed")
        return denoised

    def wavelet_threshold_denoise(self, data, threshold_method='bayes',
                                threshold_mode='soft', levels=4):
        # print(f"Applying wavelet thresholding: {threshold_method} method, {threshold_mode} mode")

        coeffs = pywt.wavedec2(data, self.wavelet, level=levels, mode=self.mode)
        sigma = self.estimate_noise_level(data)
        thresholds = []

        for i, detail_coeffs in enumerate(coeffs[1:], 1):
            lh, hl, hh = detail_coeffs

            if threshold_method == 'bayes':
                level_thresholds = []
                for subband in [lh, hl, hh]:
                    var_y = np.var(subband)
                    var_x = max(0, var_y - sigma**2)

                    if var_x > 0:
                        thresh = sigma**2 / np.sqrt(var_x)
                    else:
                        thresh = np.max(np.abs(subband))

                    level_thresholds.append(thresh)

                thresholds.append(level_thresholds)

            elif threshold_method == 'sure':
                n = lh.size
                level_thresh = sigma * np.sqrt(2 * np.log(n))
                thresholds.append([level_thresh] * 3)

            elif threshold_method == 'minimax':
                n = lh.size
                if n > 32:
                    level_thresh = sigma * (0.3936 + 0.1829 * np.log2(n))
                else:
                    level_thresh = 0
                thresholds.append([level_thresh] * 3)

            else:
                level_thresh = 3 * sigma
                thresholds.append([level_thresh] * 3)

        coeffs_thresholded = [coeffs[0].copy()]

        for i, (detail_coeffs, level_thresholds) in enumerate(zip(coeffs[1:], thresholds)):
            lh, hl, hh = detail_coeffs

            if threshold_mode == 'soft':
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='soft')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='soft')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='soft')
            else:
                lh_thresh = pywt.threshold(lh, level_thresholds[0], mode='hard')
                hl_thresh = pywt.threshold(hl, level_thresholds[1], mode='hard')
                hh_thresh = pywt.threshold(hh, level_thresholds[2], mode='hard')

            coeffs_thresholded.append((lh_thresh, hl_thresh, hh_thresh))

        denoised = pywt.waverec2(coeffs_thresholded, self.wavelet, mode=self.mode)
        denoised = np.clip(denoised, 0, 1)

        # print("Wavelet thresholding completed")
        return denoised, coeffs, coeffs_thresholded

    def hybrid_svd_wavelet_denoise(self, data, svd_first=True, svd_rank_ratio=0.8,
                                 wavelet_method='bayes', wavelet_mode='soft',
                                 wavelet_levels=4, svd_block_size=64):
        # print("\n=== HYBRID SVD + WAVELET DENOISING ===")

        if svd_first:
            # print("Step 1: Applying SVD denoising...")
            svd_denoised = self.svd_denoise(data, rank_ratio=svd_rank_ratio,
                                          block_size=svd_block_size)

            # print("Step 2: Applying wavelet thresholding to SVD result...")
            final_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                svd_denoised, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

        else:
            # print("Step 1: Applying wavelet thresholding...")
            wavelet_denoised, wavelet_coeffs, thresholded_coeffs = self.wavelet_threshold_denoise(
                data, threshold_method=wavelet_method,
                threshold_mode=wavelet_mode, levels=wavelet_levels
            )

            # print("Step 2: Applying SVD denoising to wavelet result...")
            final_denoised = self.svd_denoise(wavelet_denoised, rank_ratio=svd_rank_ratio,
                                            block_size=svd_block_size)
            svd_denoised = wavelet_denoised

        self.denoised_data = final_denoised

        return {
            'original': data,
            'svd_denoised': svd_denoised,
            'final_denoised': final_denoised,
            'wavelet_coeffs': wavelet_coeffs,
            'thresholded_coeffs': thresholded_coeffs
        }

    def calculate_quality_metrics(self, original, denoised):
        mse = mean_squared_error(original.flatten(), denoised.flatten())
        if mse > 0 and np.max(original) > 0:
            psnr = 20 * np.log10(np.max(original) / np.sqrt(mse))
        else:
            psnr = float('inf')

        signal_power = np.mean(original**2)
        noise_power = np.mean((original - denoised)**2)
        if noise_power > 0 and signal_power > 0:
             snr = 10 * np.log10(signal_power / noise_power)
        elif signal_power > 0 and noise_power == 0:
             snr = float('inf')
        else:
            snr = np.nan

        original_norm = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        denoised_norm = (denoised - np.min(denoised)) / (np.max(denoised) - np.min(denoised) + 1e-10)

        if len(original_norm.shape) == 2:
            ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
        elif len(original_norm.shape) == 3:
             if original_norm.shape[2] in [1, 3, 4]:
                 ssim_score = ssim(original_norm, denoised_norm, data_range=1.0, channel_axis=2, win_size=min(7, original_norm.shape[0], original_norm.shape[1]))
             else:
                 ssim_scores_channels = [ssim(original_norm[:,:,i], denoised_norm[:,:,i], data_range=1.0, win_size=min(7, original_norm.shape[0], original_norm.shape[1])) for i in range(original_norm.shape[2])]
                 ssim_score = np.mean(ssim_scores_channels)
        else:
            ssim_score = np.nan


        return {
            'MSE': mse,
            'PSNR': psnr,
            'SNR': snr,
            'SSIM': ssim_score
        }

    def visualize_denoising_results(self, results, show_coefficients=True):
        original = results['original']
        svd_denoised = results['svd_denoised']
        final_denoised = results['final_denoised']

        if show_coefficients:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        else:
            fig, axes = plt.subplots(1, 4, figsize=(20, 5))
            axes = axes.reshape(1, -1)

        original_display = (original - np.min(original)) / (np.max(original) - np.min(original) + 1e-10)
        svd_denoised_display = (svd_denoised - np.min(svd_denoised)) / (np.max(svd_denoised) - np.min(svd_denoised) + 1e-10)
        final_denoised_display = (final_denoised - np.min(final_denoised)) / (np.max(final_denoised) - np.min(final_denoised) + 1e-10)


        axes[0, 0].imshow(original_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 0].set_title('Original Noisy Satellite Data')
        axes[0, 0].axis('off')

        axes[0, 1].imshow(svd_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 1].set_title('SVD Denoised')
        axes[0, 1].axis('off')

        axes[0, 2].imshow(final_denoised_display, cmap='gray', vmin=0, vmax=1)
        axes[0, 2].set_title('SVD + Wavelet Denoised')
        axes[0, 2].axis('off')

        if show_coefficients:
            diff_range = np.max(original) - np.min(original)
            svd_diff = np.abs(original - svd_denoised)
            final_diff = np.abs(original - final_denoised)

            svd_diff_display = svd_diff / (diff_range + 1e-10)
            final_diff_display = final_diff / (diff_range + 1e-10)


            axes[1, 0].imshow(svd_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 0].set_title('SVD Noise Removed')
            axes[1, 0].axis('off')

            axes[1, 1].imshow(final_diff_display, cmap='hot', vmin=0, vmax=0.3)
            axes[1, 1].set_title('Total Noise Removed')
            axes[1, 1].axis('off')

            coeffs = results['wavelet_coeffs']
            try:
                coeffs_array, slices = pywt.coeffs_to_array(coeffs)
                axes[1, 2].imshow(np.abs(coeffs_array), cmap='viridis')
                axes[1, 2].set_title('Wavelet Coefficients')
                axes[1, 2].axis('off')
            except Exception as e:
                 print(f"Could not visualize wavelet coefficients: {e}")
                 axes[1, 2].set_title('Wavelet Coeffs (Error)')
                 axes[1, 2].axis('off')


        plt.tight_layout()
        plt.show()

        metrics = self.calculate_quality_metrics(original, final_denoised)

        print("\n=== DENOISING QUALITY METRICS ===")
        print(f"MSE: {metrics['MSE']:.6f}")
        print(f"PSNR: {metrics['PSNR']:.2f} dB")
        print(f"SNR: {metrics['SNR']:.2f} dB")
        print(f"SSIM: {metrics['SSIM']:.4f}")
        print("="*30)

    def compare_methods(self, data, methods_config=None):
        if methods_config is None:
            methods_config = [
                {'svd_rank': 0.7, 'wavelet_method': 'bayes', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.8, 'wavelet_method': 'sure', 'wavelet_mode': 'soft'},
                {'svd_rank': 0.9, 'wavelet_method': 'bayes', 'wavelet_mode': 'hard'},
                {'svd_rank': 0.8, 'wavelet_method': 'minimax', 'wavelet_mode': 'soft'},
            ]

        # print("\n=== COMPARING DENOISING METHODS ===")

        results_comparison = []

        for i, config in enumerate(methods_config):
            # print(f"\nMethod {i+1}: SVD rank={config['svd_rank']}, "
            #       f"Wavelet={config['wavelet_method']}, Mode={config['wavelet_mode']}")

            results = self.hybrid_svd_wavelet_denoise(
                data,
                svd_first=True,
                svd_rank_ratio=config['svd_rank'],
                wavelet_method=config['wavelet_method'],
                wavelet_mode=config['wavelet_mode']
            )

            metrics = self.calculate_quality_metrics(data, results['final_denoised'])

            # print(f"  PSNR: {metrics['PSNR']:.2f} dB")
            # print(f"  SNR: {metrics['SNR']:.2f} dB")
            # print(f"  SSIM: {metrics['SSIM']:.4f}")

            results_comparison.append({
                'config': config,
                'metrics': metrics,
                'denoised': results['final_denoised']
            })

        if results_comparison:
            best_method = max(results_comparison, key=lambda x: x['metrics']['PSNR'] if not np.isnan(x['metrics']['PSNR']) else -float('inf'))
            # print(f"\nBest method based on PSNR: {best_method['config']}")
            # print(f"Best PSNR: {best_method['metrics']['PSNR']:.2f} dB")
        else:
            print("\nNo methods were successfully compared.")


        return results_comparison

# Assuming svd_wavelet_denoiser is already initialized and sorted_image_files is available
# from the previous step ('ab25217c')

# 1. Create a new directory named 'denoised_frames_svd_wavelet' to store the processed images.
denoised_output_dir = "denoised_frames_svd_wavelet"
os.makedirs(denoised_output_dir, exist_ok=True)
print(f"Created directory: {denoised_output_dir}")

# Define the input directory for frames
input_frames_dir = "extracted_frames"

# 2. Loop through each image file path in the sorted_image_files list.
for i, filename in enumerate(sorted_image_files):
    # a. Construct the full input path by joining 'extracted_frames' with the current image filename.
    input_path = os.path.join(input_frames_dir, filename)

    # b. Load the image using cv2.imread().
    img_bgr = cv2.imread(input_path)

    if img_bgr is None:
        print(f"Warning: Could not load image from {input_path}. Skipping.")
        continue

    # c. Convert the loaded image to grayscale if it is a color image (check its dimensions).
    # Then, convert it to a float type and normalize its pixel values to the range [0, 1] using img_as_float().
    if len(img_bgr.shape) == 3:
        img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    else:
        img_gray = img_bgr  # Already grayscale

    # Convert to float and normalize to [0, 1]
    normalized_grayscale_image = img_as_float(img_gray)

    # d. Apply the SVD + Wavelet denoising method.
    # Instantiate denoiser if it's not globally available or if we want fresh instance per run/task
    # It was initialized in the previous step, so we reuse it.
    # svd_wavelet_denoiser = SVDWaveletDenoiser() # Re-instantiate if needed per frame

    denoising_results = svd_wavelet_denoiser.hybrid_svd_wavelet_denoise(
        normalized_grayscale_image,
        svd_first=True,
        svd_rank_ratio=0.8,
        wavelet_method='bayes',
        wavelet_mode='soft',
        wavelet_levels=4,
        svd_block_size=64
    )

    # e. Extract the final_denoised image from the results dictionary.
    final_denoised_image = denoising_results['final_denoised']

    # f. Convert the denoised image back to an 8-bit unsigned integer format (uint8)
    # with pixel values in the range [0, 255] for saving.
    denoised_uint8 = (final_denoised_image * 255).astype(np.uint8)

    # g. Construct the full output path for the denoised image.
    output_path = os.path.join(denoised_output_dir, filename)

    # h. Save the denoised image using cv2.imwrite().
    cv2.imwrite(output_path, denoised_uint8)

    # i. Print a progress message.
    if (i + 1) % 50 == 0 or i == len(sorted_image_files) - 1:
        print(f"Processed {i + 1}/{len(sorted_image_files)} frames: {filename}")

print("Batch denoising with SVD + Wavelet complete.")
```

```python
import os
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Define the directories
input_frames_dir = "extracted_frames"
denoised_output_dir = "denoised_frames_svd_wavelet"

# Get a list of all original and denoised image files
original_frame_files = sorted([f for f in os.listdir(input_frames_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp'))])
denoised_frame_files = sorted([f for f f in os.listdir(denoised_output_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp'))])

if not original_frame_files or not denoised_frame_files:
    print("Error: No original or denoised image files found. Please ensure all previous steps ran successfully.")
else:
    print(f"Found {len(original_frame_files)} original frames and {len(denoised_frame_files)} denoised frames.")

    # Select 3-5 representative frames (e.g., first, middle, last)
    num_frames_to_display = min(5, len(original_frame_files))

    selected_frame_indices = []
    if len(original_frame_files) > 0:
        selected_frame_indices.append(0) # First frame
        if num_frames_to_display > 1:
            # Calculate approximate spacing to get evenly distributed frames
            spacing = (len(original_frame_files) - 1) // (num_frames_to_display - 1)
            for i in range(1, num_frames_to_display - 1):
                idx = i * spacing
                if idx < len(original_frame_files) and idx not in selected_frame_indices:
                    selected_frame_indices.append(idx)
            # Ensure the last frame is included if there's enough room
            if (len(original_frame_files) - 1) not in selected_frame_indices:
                selected_frame_indices.append(len(original_frame_files) - 1)

    # Ensure unique and sorted indices
    selected_frame_indices = sorted(list(set(selected_frame_indices)))

    # Display up to num_frames_to_display
    actual_frames_to_display = selected_frame_indices[:num_frames_to_display]

    print(f"Displaying {len(actual_frames_to_display)} sample frames for comparison...")

    fig, axes = plt.subplots(2, len(actual_frames_to_display), figsize=(5 * len(actual_frames_to_display), 8))
    if len(actual_frames_to_display) == 1:
        axes = np.array([[axes[0]], [axes[1]]]) # Ensure axes is 2D for single frame

    for i, idx in enumerate(actual_frames_to_display):
        original_frame_name = original_frame_files[idx]
        denoised_frame_name = denoised_frame_files[idx] # Assuming same naming convention

        original_path = os.path.join(input_frames_dir, original_frame_name)
        denoised_path = os.path.join(denoised_output_dir, denoised_frame_name)

        # Load original frame
        img_original_bgr = cv2.imread(original_path)
        img_original_rgb = cv2.cvtColor(img_original_bgr, cv2.COLOR_BGR2RGB)

        # Load denoised frame (should be grayscale uint8)
        img_denoised_gray = cv2.imread(denoised_path, cv2.IMREAD_GRAYSCALE)

        # Display original frame
        axes[0, i].imshow(img_original_rgb)
        axes[0, i].set_title(f'Original: {original_frame_name}')
        axes[0, i].axis('off')

        # Display denoised frame
        axes[1, i].imshow(img_denoised_gray, cmap='gray')
        axes[1, i].set_title(f'SVD+Wavelet Denoised: {denoised_frame_name}')
        axes[1, i].axis('off')

    plt.tight_layout()
    plt.show()

    print("Sample original and SVD + Wavelet denoised frames displayed successfully.")
```
